Historical Background</div><div style="font-size:26px;opacity:.95;margin-top:6px;">From Unix to Linux (1960s – present)</div></div>

<div style="background:#F7F9FC;border-left:6px solid #1B2A49;border-radius:10px;padding:12px 16px;margin:10px 0 14px;color:#333;font-size:30px;line-height:1.55;">
  The history of Linux is easier to understand as <b>a chain of related ideas</b> rather than the story of one operating system:
</div>

<div style="display:grid;grid-template-columns:repeat(3,1fr);gap:14px;margin:12px 0;">

  <div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:16px;box-shadow:0 3px 10px rgba(0,0,0,.12);">
    <div style="font-size:60px;">🏛️</div>
    <div style="font-weight:800;color:#4C8BF5;font-size:36px;margin:5px 0 6px;">Unix</div>
    <div style="font-size:30px;color:#333;line-height:1.5;">
      Introduced a durable operating-system design tradition.
    </div>
  </div>

  <div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:16px;box-shadow:0 3px 10px rgba(0,0,0,.12);">
    <div style="font-size:60px;">🦬</div>
    <div style="font-weight:800;color:#FF7A00;font-size:36px;margin:5px 0 6px;">GNU</div>
    <div style="font-size:30px;color:#333;line-height:1.5;">
      Built a large body of free software around a Unix-like system.
    </div>
  </div>

  <div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:16px;box-shadow:0 3px 10px rgba(0,0,0,.12);">
    <div style="font-size:60px;">🐧</div>
    <div style="font-weight:800;color:#9E4CF5;font-size:36px;margin:5px 0 6px;">Linux</div>
    <div style="font-size:30px;color:#333;line-height:1.5;">
      Supplied the kernel that completed a practical GNU/Linux system.
    </div>
  </div>

</div>

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:20px 0 12px;font-size:32px;">
  Era-by-Era Summary
</h3>

<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">

<thead>
<tr>
  <th style="background:#FF7A00;color:white;padding:16px 18px;text-align:left;font-size:26px;width:13%;">
    When
  </th>
  <th style="background:#FF7A00;color:white;padding:16px 18px;text-align:left;font-size:26px;width:23%;">
    Milestone
  </th>
  <th style="background:#FF7A00;color:white;padding:16px 18px;text-align:left;font-size:26px;">
    What happened
  </th>
</tr>
</thead>

<tbody>

<tr>
  <td style="padding:16px 18px;background:#FFEDD9;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    1960s
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    Multics &amp; time-sharing
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    An ambitious time-sharing OS project involving MIT, General Electric and Bell Labs. It was not Linux, but it influenced the environment in which Unix emerged.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFFFFF;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    1969
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    Unix begins
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    At Bell Labs, <b>Ken Thompson</b> began developing a smaller operating system after Bell Labs withdrew from Multics. <b>Dennis Ritchie</b> became a key collaborator.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFEDD9;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    Early 1970s
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    C &amp; portability
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    <b>Dennis Ritchie</b> developed the C language. Unix was rewritten in C, making it much easier to port between computer architectures.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFFFFF;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    1970s
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    Unix spreads
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    Unix became influential in universities and research. Berkeley developed <b>BSD</b>, an important Unix branch.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFEDD9;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    1983
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    GNU
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    <b>Richard Stallman</b> announced GNU, a project to develop a Unix-compatible system whose software could be freely studied, modified and shared.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFFFFF;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    1985
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    FSF
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    The <b>Free Software Foundation</b> was founded to support the free software movement.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFEDD9;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    1989
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    GPL
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    The first <b>GNU General Public License</b> was released, using copyleft to preserve users' freedoms.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFFFFF;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    1991
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    Linux kernel
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    <b>Linus Torvalds</b>, then a University of Helsinki student, announced the project in August 1991 and released the first Linux kernel in October 1991.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFEDD9;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    1990s
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    Distributions
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    Linux combined with GNU tools and other software formed complete systems. <b>Debian</b> and <b>Slackware</b> became early Linux distributions.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFFFFF;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    1998
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    Open Source
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    The term <b>Open Source</b> gained momentum through the <b>Open Source Initiative</b>.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFEDD9;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    2000s–2010s
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    Infrastructure
  </td>
  <td style="padding:16px 18px;background:#FFEDD9;color:#222;font-size:26px;line-height:1.55;vertical-align:top;">
    Linux expanded across servers, embedded systems, networking and supercomputing. <b>Android</b> adopted the Linux kernel, and Linux became central to cloud infrastructure.
  </td>
</tr>

<tr>
  <td style="padding:16px 18px;background:#FFFFFF;color:#FF7A00;font-size:26px;font-weight:800;vertical-align:top;">
    2020s–2026
  </td>
  <td style="padding:16px 18px;background:#FFFFFF;color:#222;font-size:26px;font-weight:800;vertical-align:top;">
    Current era
  </td>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:20px 0 10px;font-size:32px;">
  🐧 2.1 What Exactly Is Linux?
</h3>

<p style="font-size:30px;line-height:1.6;color:#333;">
  <b>Linux is the name of the kernel</b> — the privileged core of the operating system
  that manages CPU scheduling, memory, devices, filesystems, networking and system calls.
  A complete Linux-based desktop system also includes user-space libraries, utilities,
  a shell, graphical components, package management and applications. Therefore,
  <b>“GNU/Linux”</b> can be a more precise term for systems that combine the Linux kernel
  with GNU software, while <b>“Linux distribution”</b> is the common practical term for
  a complete packaged system.
</p>

<div style="text-align:center;margin:20px 0;">
  <img src="images/Linux_System_Layers.jpg"
       alt="Figure 2 — Simplified Linux system layers."
       style="width:78%;max-width:100%;border-radius:14px;border:3px solid #9E4CF5;box-shadow:0 6px 18px rgba(0,0,0,.18);">
  <div style="font-size:26px;color:#666;font-style:italic;margin-top:8px;">
    Figure 2 — Simplified Linux system layers.
  </div>
</div>

<div style="text-align:center;margin:20px 0;">
  <img src="images/Linux_Distribution_Architecture.jpg"
       alt="Linux kernel, GNU tools, applications and package management form a Linux distribution."
       style="width:92%;max-width:100%;border-radius:14px;border:3px solid #9E4CF5;box-shadow:0 6px 18px rgba(0,0,0,.18);">
  <div style="font-size:26px;color:#666;font-style:italic;margin-top:8px;">
    Linux kernel + GNU tools + applications + package management = Linux distribution
  </div>
</div>

  </td>
  <td style="padding:11px 14px;background:#E8F0FE;color:#222;font-size:28px;line-height:1.5;">

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:20px 0 10px;font-size:32px;">
2.4 Step-by-Step: Ubuntu in a Virtual Machine
</h3>

<div style="background:#F2FAF4;border-left:5px solid #34A853;border-radius:10px;padding:12px 16px;margin:10px 0 14px;color:#444;font-size:28px;line-height:1.5;">
  Follow these steps to install Ubuntu safely inside a virtual machine without modifying the host operating system.
</div>

<div style="margin:10px 0;">

  <!-- 1 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#34A853;font-size:38px;font-weight:800;line-height:1.5;">
      1.
    </div>
    <div style="background:#E6F4EA;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;">
      Install <b>VMware Workstation</b> or <b>VirtualBox</b> on the host. Enable CPU virtualisation in UEFI/BIOS if the hypervisor reports that it is disabled.
    </div>
  </div>

  <!-- 2 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#34A853;font-size:38px;font-weight:800;line-height:1.5;">
      2.
    </div>
    <div style="background:#FFFFFF;border:1px solid #DCE3EF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;">
      Download the approved <b>Ubuntu Desktop ISO</b> from the official Ubuntu release page. Verify the release name and architecture.
    </div>
  </div>

  <!-- 3 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#34A853;font-size:38px;font-weight:800;line-height:1.5;">
      3.
    </div>
    <div style="background:#E6F4EA;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;">
      Create a new VM and choose <b>Linux / Ubuntu</b> as the guest operating system.
    </div>
  </div>

  <!-- 4 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#34A853;font-size:38px;font-weight:800;line-height:1.5;">
      4.
    </div>
    <div style="background:#FFFFFF;border:1px solid #DCE3EF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;">
      Allocate resources according to the current Ubuntu requirements and the host's available capacity. Give the VM enough <b>RAM</b>, <b>CPU cores</b>, and <b>storage</b> for the intended laboratory tasks.
    </div>
  </div>

  <!-- 5 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#34A853;font-size:38px;font-weight:800;line-height:1.5;">

    </div>
    <div style="background:#E6F4EA;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;">

  </div>

  <!-- 6 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#34A853;font-size:38px;font-weight:800;line-height:1.5;">

  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#34A853;font-size:38px;font-weight:800;line-height:1.5;">

    <div style="background:#E6F4EA;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;">

  </div>

  <!-- 8 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">

  <!-- 9 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:20px 0 10px;font-size:32px;">
  2.5 Step-by-Step: Live USB
</h3>

<div style="background:#F3F7FE;border-left:5px solid #4C8BF5;border-radius:10px;padding:10px 14px;margin:10px 0 14px;color:#444;font-size:28px;line-height:1.5;">
  Create a bootable USB drive and use it to start Ubuntu without installing it on the computer.
</div>

<div style="margin:10px 0;">

  <!-- 1 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#4C8BF5;font-size:38px;font-weight:800;line-height:1.5;">
      1.
    </div>
    <div style="background:#E8F0FE;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;">

    </div>
  </div>

  <!-- 2 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#4C8BF5;font-size:38px;font-weight:800;line-height:1.5;">
      2.
    </div>
    <div style="background:#FFFFFF;border:1px solid #DCE3EF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;">

    </div>

  <!-- 3 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">

    </div>
  </div>

  <!-- 4 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#4C8BF5;font-size:38px;font-weight:800;line-height:1.5;">
      4.

    <div style="background:#FFFFFF;border:1px solid #DCE3EF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;">

    </div>
  </div>

  <!-- 5 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#4C8BF5;font-size:38px;font-weight:800;line-height:1.5;">

    </div>
  </div>

  <!-- 6 -->
  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#4C8BF5;font-size:38px;font-weight:800;line-height:1.5;">

    </div>

    </div>
  </div>

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 6px;font-size:32px;">
  2.6 Step-by-step: Dual Boot with Windows
</h3>

<div style="margin:12px 0;">

  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#FF7A00;font-size:32px;font-weight:800;line-height:1.5;">1.</div>
    <div style="background:#FFFFFF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;border-left:5px solid #FFEDD9;">
      <b>Back up</b> important files to independent storage before modifying partitions.
    </div>
  </div>

  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#FF7A00;font-size:32px;font-weight:800;line-height:1.5;">2.</div>
    <div style="background:#FFFFFF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;border-left:5px solid #FFEDD9;">
      In Windows, confirm that the PC uses <b>UEFI</b> and check whether
      <b>BitLocker / device encryption</b> is enabled. Save the
      <b>recovery key</b> before changing disk configuration.
    </div>
  </div>

  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#FF7A00;font-size:32px;font-weight:800;line-height:1.5;">3.</div>
    <div style="background:#FFFFFF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;border-left:5px solid #FFEDD9;">
      Use Windows <b>Disk Management</b> to <b>shrink</b> an existing
      partition and create unallocated space. Do not delete or format
      the Windows system partition just to make space.
    </div>
  </div>

  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#FF7A00;font-size:32px;font-weight:800;line-height:1.5;">4.</div>
    <div style="background:#FFFFFF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;border-left:5px solid #FFEDD9;">
      Create the Ubuntu bootable USB and boot the computer in
      <b>UEFI mode</b>.
    </div>
  </div>

  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#FF7A00;font-size:32px;font-weight:800;line-height:1.5;">5.</div>
    <div style="background:#FFFFFF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;border-left:5px solid #FFEDD9;">
      In the installer, choose the option that uses the prepared
      <b>free space</b>. For manual partitioning, verify the disk and
      partition identifiers carefully.
    </div>
  </div>

  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#FF7A00;font-size:32px;font-weight:800;line-height:1.5;">6.</div>
    <div style="background:#FFFFFF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;border-left:5px solid #FFEDD9;">
      Reuse the <b>existing EFI System Partition</b> rather than
      creating another one unnecessarily. Mount the Linux root
      filesystem at <code>/</code>. A separate <code>/home</code>
      partition is optional.
    </div>
  </div>

  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#FF7A00;font-size:32px;font-weight:800;line-height:1.5;">7.</div>
    <div style="background:#FFFFFF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;border-left:5px solid #FFEDD9;">
      Carefully check the <b>installation summary</b> and target disk
      before proceeding. Treat <b>“Erase disk”</b> as a destructive operation.
    </div>
  </div>

  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#FF7A00;font-size:32px;font-weight:800;line-height:1.5;">8.</div>
    <div style="background:#FFFFFF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;border-left:5px solid #FFEDD9;">
      Complete the boot configuration. On a typical UEFI installation,
      the firmware and bootloader configuration should allow the installed
      operating systems to be selected at startup.
    </div>
  </div>

  <div style="display:flex;align-items:flex-start;margin:10px 0;">
    <div style="flex:0 0 44px;color:#FF7A00;font-size:32px;font-weight:800;line-height:1.5;">9.</div>
    <div style="background:#FFFFFF;padding:12px 16px;border-radius:10px;flex:1;font-size:30px;line-height:1.5;color:#222;border-left:5px solid #FFEDD9;">
      Boot each operating system once and confirm that the system,
      files, and network connection work correctly.
    </div>
  </div>

</div>

<div style="background:#FFFFFF;border-left:8px solid #EA4335;border-radius:10px;padding:14px 18px;margin:14px 0;color:#222;">

  <div style="font-weight:800;color:#EA4335;font-size:32px;margin-bottom:6px;">
    ⛔ Critical Warning — “Erase disk”
  </div>

  <div style="font-size:30px;line-height:1.55;color:#222;">
    Selecting <b>Erase disk</b> can destroy the existing operating system

    <b>disposable lab machine</b> or perform the procedure in a

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 6px;">
 2.7 Partitioning Concepts
</h3>

<p style="font-size:30px;line-height:1.6;color:#333;">
  On modern Linux systems using <b>UEFI</b>, the disk normally uses a
  <b>GPT</b> partition table. A typical installation includes an
  <b>EFI System Partition</b> formatted as FAT32 and a Linux filesystem
  mounted at <code>/</code>. A separate <code>/home</code> partition is
  optional, and <b>swap can be implemented as a swapfile</b> rather than
  a dedicated partition. Older installation guides may present separate
  <code>/boot</code> and swap partitions as mandatory, but these are
  better treated as <i>teaching examples</i> rather than universal
  requirements.
</p>

  <td style="padding:9px 12px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:28px;">
    Optional
  </td>
  <td style="padding:9px 12px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:28px;">
    Can separate personal data from the operating system.
  </td>
</tr>

<tr>
  <td style="padding:9px 12px;border-bottom:1px solid #e3e8f0;background:#FFFFFF;color:#222;font-size:28px;font-weight:700;">
    <code>swap</code>
  </td>
  <td style="padding:9px 12px;border-bottom:1px solid #e3e8f0;background:#FFFFFF;color:#222;font-size:28px;">
    Virtual memory and hibernation support
  </td>
  <td style="padding:9px 12px;border-bottom:1px solid #e3e8f0;background:#FFFFFF;color:#222;font-size:28px;">
    Implementation-dependent
  </td>
  <td style="padding:9px 12px;border-bottom:1px solid #e3e8f0;background:#FFFFFF;color:#222;font-size:28px;">
    Modern Ubuntu can use a swapfile instead of a dedicated partition.
  </td>
</tr>

<tr>
  <td style="padding:9px 12px;background:#F3E8FF;color:#222;font-size:28px;font-weight:700;">
    <code>/boot</code>
  </td>
  <td style="padding:9px 12px;background:#F3E8FF;color:#222;font-size:28px;">
    Boot-related files
  </td>
  <td style="padding:9px 12px;background:#F3E8FF;color:#222;font-size:28px;">
    Usually not a separate partition
  </td>
  <td style="padding:9px 12px;background:#F3E8FF;color:#222;font-size:28px;">
    The directory may exist inside the root filesystem; a separate partition is only needed in particular layouts.
  </td>
</tr>

</tbody>
</table>

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 6px;font-size:32px;">
2.8 Boot Loader: GRUB
</h3>

<p style="font-size:30px;line-height:1.6;color:#222;">
  A <b>boot loader</b> starts an operating system. In a common Ubuntu
  <b>UEFI</b> installation, <b>GRUB</b> participates in the boot process
  and can display a menu containing Linux and other detected operating
  systems. The boot loader is <b>not the kernel</b>; it loads the kernel
  or transfers control to the next stage of the boot process.
</p>

<div style="text-align:center;margin:20px 0;">
  <img src="images/Boot_Loader_GRUB.jpg"
       alt="Firmware → boot loader → kernel → init system → user space"
       style="width:100%;max-width:100%;border-radius:14px;border:3px solid #EA4335;box-shadow:0 6px 18px rgba(0,0,0,.18);">
  <div style="font-size:26px;color:#666;font-style:italic;margin-top:8px;">
    Firmware → Boot Loader → Kernel → Init System → User Space
  </div>
</div>

<p style="font-size:30px;line-height:1.6;color:#222;">
  <b>Read-only inspection commands:</b>
</p>

````python
%%bash
# efibootmgr / /boot/efi need root or a UEFI system; errors are normal in containers and WSL
(sudo -n efibootmgr -v 2>&1 || echo "efibootmgr: needs sudo or is unavailable here") | head -20
echo "----- /boot/efi"
ls /boot/efi 2>&1 | head
echo "----- /boot"
ls /boot | head -20
````

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 6px;font-size:32px;">
2.9 Linux Distribution Families
</h3>

<div style="text-align:center;margin:20px 0;">
  <img src="images/Linux_Distribution_Families.png"
       alt="Distribution families — same kernel, different packaging and policies."
       style="width:92%;max-width:100%;border-radius:14px;border:3px solid #34A853;box-shadow:0 6px 18px rgba(0,0,0,.18);">
  <div style="font-size:26px;color:#666;font-style:italic;margin-top:8px;">
    Distribution families — same kernel, different packaging and policies.
  </div>
</div>

<p style="font-size:30px;line-height:1.6;color:#222;">
  A <b>distribution</b> packages a kernel, user-space tools, libraries,
  package-management infrastructure, installers and applications into
  a usable operating system. The distribution choice affects
  <b>package-management commands, release policies and default software</b>,
  while the Linux kernel remains the central component. <b>Ubuntu</b>
  is derived from Debian, while <b>Fedora</b> belongs to the Red Hat family.
</p>

<p style="font-size:30px;line-height:1.6;color:#222;">
  <b>Identify the distribution and package-manager family:</b>
</p>

````python
%%bash
cat /etc/os-release
echo "-----"
apt --version 2>/dev/null || true
dnf --version 2>/dev/null || true
pacman --version 2>/dev/null || true
````

**A nicer view of the same information** — this Python cell reads `/etc/os-release` and renders a colorful ID card:

````python
import platform, html
try:
    info = dict(l.rstrip().split("=",1) for l in open("/etc/os-release") if "=" in l)
    info = {k: v.strip('"') for k, v in info.items()}
    rows = [("🐧 Distribution", info.get("PRETTY_NAME","?")), ("🏷️ ID / family", info.get("ID","?") + " / " + info.get("ID_LIKE","—")),
            ("🔢 Version", info.get("VERSION","?")), ("⚙️ Kernel", platform.release()), ("🧱 Architecture", platform.machine()), ("🖥️ Hostname", platform.node())]
    cells = "".join(f'<div style="background:rgba(255,255,255,.15);padding:10px 14px;border-radius:10px;"><div style="font-size:12px;opacity:.85;">{a}</div><div style="font-size:17px;font-weight:700;">{html.escape(b)}</div></div>' for a,b in rows)
    display(HTML(f'<div style="background:linear-gradient(135deg,#4C8BF5,#9E4CF5);color:white;padding:18px;border-radius:16px;box-shadow:0 6px 18px rgba(0,0,0,.25);"><div style="font-size:22px;font-weight:800;margin-bottom:10px;">System ID Card</div><div style="display:grid;grid-template-columns:repeat(3,1fr);gap:10px;">{cells}</div></div>'))
except FileNotFoundError:
    display(HTML('<div style="background:#FCE8E6;border-left:8px solid #EA4335;padding:12px;border-radius:8px;">⚠️ /etc/os-release not found — this is not a Linux system. Run on Linux, a VM or WSL.</div>'))
````

    font-size:32px;
">
3.1 Software Repositories and Package Management
</h3>

<p style="
    font-size:30px;
    line-height:1.65;
    color:#222;
">
  Distributions obtain software from <b>repositories</b>.
  A package manager resolves <b>dependencies</b>, downloads packages,
  verifies metadata, installs files and records the installed state.
  On Debian/Ubuntu the standard tool is <b>APT</b>.
</p>

<div style="text-align:center;margin:20px 0;">
  <img src="images/APT_Workflow.png"
       alt="Core APT workflow"
       style="
         width:100%;
         max-width:100%;
         border-radius:14px;
         border:3px solid #4C8BF5;
         box-shadow:0 6px 18px rgba(0,0,0,.18);
       ">
  <div style="
      font-size:26px;
      color:#666;
      font-style:italic;
      margin-top:8px;
  ">
    Core APT workflow
  </div>
</div>

<div style="
    background:#E8F0FE;
    border-left:8px solid #4C8BF5;
    border-radius:10px;
    padding:18px 22px;
    margin:16px 0;
    color:#222;
">

<div style="
    font-weight:800;
    color:#4C8BF5;
    font-size:32px;
    margin-bottom:8px;
">
Why update before install?
</div>

<div style="
    font-size:30px;
    line-height:1.65;
">
  <code>apt update</code> refreshes <b>package metadata</b> —
  it does <b>not</b> upgrade anything.
  <code>apt upgrade</code> performs the actual upgrades.
  Keeping them separate helps you understand
  <b>repository metadata</b> versus <b>installed software</b>.
</div>

</div>

<p style="
    font-size:30px;
    line-height:1.6;
    color:#222;
">
  <b>Installing packages needs root privileges, so run this block in a terminal:</b>
</p>

````python
sudo apt update
apt search htop
apt show htop
sudo apt install htop
htop            # press q to quit
sudo apt remove htop
````

    font-size:32px;
">
  3.3 The Command Line and Terminal
</h3>

<p style="
    font-size:30px;
    line-height:1.65;
    color:#222;
">
  The <b>CLI</b> lets you interact with the operating system by typing commands.
  A <b>terminal emulator</b> provides the window; the <b>shell</b>, such as Bash,
  interprets commands and starts programs; and the <b>kernel</b> performs
  the privileged work.
</p>

<p style="
    font-size:30px;
    line-height:1.65;
    color:#222;
">
  Two Unix ideas are especially important:
  many resources are represented as <b>files</b>, and configuration
  is commonly stored in <b>text files</b>.
</p>

<div style="text-align:center;margin:20px 0;">
  <img src="images/Command_Anatomy.jpg"
       alt="Command anatomy: command, options, arguments"
       style="
         width:80%;
         max-width:100%;
         border-radius:14px;
         border:3px solid #1B2A49;
         box-shadow:0 6px 18px rgba(0,0,0,.18);
       ">
  <div style="
      font-size:26px;
      color:#666;
      font-style:italic;
      margin-top:8px;
  ">
    Command anatomy: command, options, arguments
  </div>
</div>

<div style="
    background:#FFF6D6;
    border-left:8px solid #F5B301;
    border-radius:10px;
    padding:18px 22px;
    margin:16px 0;
    color:#222;
">

<div style="
    font-weight:800;
    color:#B07A00;
    font-size:30px;
    margin-bottom:8px;
">
  Correction from Older Notes
</div>

<div style="
    font-size:30px;
    line-height:1.65;
">
  Options are <b>not</b> universally “single letters after
  <code>-</code>” or “words after <code>--</code>”.
  These are common <b>conventions</b>, not universal rules.
  Always check <code>man</code> or <code>--help</code>.
  Commands are also generally <b>case-sensitive</b>.
</div>

</div>

<p style="
    font-size:30px;
    line-height:1.6;
    color:#222;
">
  <b>Open a terminal</b> with
  <kbd>Ctrl</kbd> + <kbd>Alt</kbd> + <kbd>T</kbd>
  and practise basic navigation:
</p>

````python
%%bash
pwd
ls -lah | head -15
cd /etc && pwd
cd ~ && pwd
cd .. && pwd
````

    margin:16px 0;
    color:#222;
">

<div style="
    font-weight:800;
    color:#34A853;
    font-size:30px;
    margin-bottom:8px;
">
  Least Privilege
</div>

<div style="
    font-size:30px;
    line-height:1.65;
">
  Grant only the permissions a task needs.
  Use <code>sudo</code> for <i>specific</i> administrative commands
  instead of performing normal work as <code>root</code>.
  <b>Never paste a privileged command you do not understand.</b>
</div>

</div>

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">3.6.1 Linux /proc Filesystem</h3>

<p style="font-size:26px;line-height:1.7;color:#222;margin:10px 0;">
The <b style="color:#4C8BF5;">/proc</b> directory is a <b>virtual filesystem</b> that provides information about the running Linux kernel, hardware, processes, and system resources. Most entries are not ordinary files stored on disk; their contents are generated by the kernel when you read them.
</p>

<div style="background:#E8F0FE;border-left:8px solid #4C8BF5;border-radius:10px;padding:15px 20px;margin:14px 0;color:#222;">
<div style="font-size:28px;font-weight:800;color:#4C8BF5;margin-bottom:6px;">Important Idea</div>
<div style="font-size:24px;line-height:1.7;">
Think of <code>/proc</code> as a <b>window into the running system</b>. You can read information about the CPU, memory, processes, and network configuration without opening the computer or inspecting hardware directly.
</div>
</div>

<div style="font-size:28px;font-weight:800;color:#9E4CF5;margin:22px 0 10px;">System Information</div>

<div style="border:2px solid #9E4CF5;border-radius:12px;padding:14px 18px;margin:10px 0;">
<div style="font-size:25px;font-weight:800;color:#9E4CF5;"><code>/proc/cpuinfo</code></div>
<div style="font-size:22px;line-height:1.6;color:#222;">Information about the CPU, including processor model, cores, and supported features.</div>
<div style="background:#F3E8FF;padding:9px 12px;border-radius:7px;margin-top:8px;font-size:21px;"><code>cat /proc/cpuinfo</code></div>
</div>

<div style="border:2px solid #34A853;border-radius:12px;padding:14px 18px;margin:10px 0;">
<div style="font-size:25px;font-weight:800;color:#34A853;"><code>/proc/meminfo</code></div>
<div style="font-size:22px;line-height:1.6;color:#222;">Memory statistics, such as total RAM, available memory, free memory, and cached data.</div>
<div style="background:#E6F4EA;padding:9px 12px;border-radius:7px;margin-top:8px;font-size:21px;"><code>cat /proc/meminfo</code></div>
</div>

<div style="border:2px solid #FF7A00;border-radius:12px;padding:14px 18px;margin:10px 0;">
<div style="font-size:25px;font-weight:800;color:#FF7A00;"><code>/proc/uptime</code></div>
<div style="font-size:22px;line-height:1.6;color:#222;">Shows how many seconds the system has been running and the accumulated idle time.</div>
<div style="background:#FFEDD9;padding:9px 12px;border-radius:7px;margin-top:8px;font-size:21px;"><code>cat /proc/uptime</code></div>
</div>
<div style="border:2px solid #4C8BF5;border-radius:12px;padding:14px 18px;margin:10px 0;">
<div style="font-size:25px;font-weight:800;color:#4C8BF5;"><code>/proc/version</code></div>
<div style="font-size:22px;line-height:1.6;color:#222;">Shows Linux kernel version and build information.</div>
<div style="background:#E8F0FE;padding:9px 12px;border-radius:7px;margin-top:8px;font-size:21px;"><code>cat /proc/version</code></div>
</div>

<div style="border:2px solid #00A6A6;border-radius:12px;padding:14px 18px;margin:10px 0;">
<div style="font-size:25px;font-weight:800;color:#008B8B;"><code>/proc/loadavg</code></div>
<div style="font-size:22px;line-height:1.6;color:#222;">Shows system load averages over 1, 5, and 15 minutes. These values represent runnable or uninterruptibly waiting tasks, not simply CPU utilization percentages.</div>
<div style="background:#DDF6F6;padding:9px 12px;border-radius:7px;margin-top:8px;font-size:21px;"><code>cat /proc/loadavg</code></div>
</div>

<div style="font-size:28px;font-weight:800;color:#9E4CF5;margin:22px 0 10px;">Boot, Mounts, and Storage</div>

<div style="font-size:22px;line-height:1.7;color:#222;">
<ul>
<li><code>/proc/cmdline</code> — kernel parameters used during boot.</li>
<li><code>/proc/mounts</code> — filesystems currently mounted and their mount points.</li>
<li><code>/proc/swaps</code> — active swap areas and their usage.</li>
<li><code>/proc/filesystems</code> — filesystem types supported by the kernel.</li>
<li><code>/proc/diskstats</code> — block-device I/O statistics, including reads and writes.</li>
<li><code>/proc/stat</code> — overall CPU and system statistics, including interrupts and context switches.</li>
<li><code>/proc/interrupts</code> — interrupt counts for CPUs and devices.</li>
</ul>
</div>

<div style="background:#FFF3E0;border-left:8px solid #FF7A00;border-radius:10px;padding:15px 20px;margin:14px 0;color:#222;">
<div style="font-size:27px;font-weight:800;color:#FF7A00;margin-bottom:6px;">Try It</div>
<div style="font-size:22px;line-height:1.7;">
Compare the following commands. The first reads mount information directly from <code>/proc</code>; the second provides a convenient overview of mounted filesystems.
</div>
<div style="font-size:21px;margin-top:10px;"><code>cat /proc/mounts</code></div>
<div style="font-size:21px;margin-top:6px;"><code>df -h</code></div>
</div>










<div style="font-size:28px;font-weight:800;color:#34A853;margin:22px 0 10px;">Network Information</div>

<div style="font-size:22px;line-height:1.7;color:#222;">
<ul>
<li><code>/proc/net/dev</code> — network-interface statistics, including received and transmitted packets and bytes.</li>
<li><code>/proc/net/tcp</code> — information about TCP sockets represented in the kernel's interface format.</li>
<li><code>/proc/net/route</code> — IPv4 routing-table information.</li>
<li><code>/proc/net/sockstat</code> — socket usage statistics.</li>
</ul>
</div>

<div style="font-size:28px;font-weight:800;color:#4C8BF5;margin:22px 0 10px;">Process Information: /proc/PID</div>

<p style="font-size:23px;line-height:1.7;color:#222;">
Linux creates a directory for each running process. <code>PID</code> means <b>Process ID</b>, a number that identifies a process. For example, if a process has PID <code>1234</code>, its information may be available under <code>/proc/1234/</code>.
</p>

<div style="border:2px solid #4C8BF5;border-radius:12px;padding:14px 18px;margin:10px 0;">
<div style="font-size:24px;font-weight:800;color:#4C8BF5;"><code>/proc/PID/status</code></div>
<div style="font-size:22px;line-height:1.6;color:#222;">Readable process details, including its state, memory usage, and thread count.</div>
<div style="background:#E8F0FE;padding:9px 12px;border-radius:7px;margin-top:8px;font-size:20px;"><code>cat /proc/$$/status</code></div>
<div style="font-size:19px;color:#444;margin-top:5px;"><code>$$</code> expands to the current shell's PID in Bash.</div>
</div>

<div style="border:2px solid #9E4CF5;border-radius:12px;padding:14px 18px;margin:10px 0;">
<div style="font-size:24px;font-weight:800;color:#9E4CF5;"><code>/proc/PID/cmdline</code></div>
<div style="font-size:22px;line-height:1.6;color:#222;">Contains the command-line arguments used to start a process. Arguments are separated by null characters, so the raw output may look unusual.</div>
<div style="background:#F3E8FF;padding:9px 12px;border-radius:7px;margin-top:8px;font-size:20px;"><code>cat /proc/$$/cmdline</code></div>
</div>

<div style="border:2px solid #34A853;border-radius:12px;padding:14px 18px;margin:10px 0;">
<div style="font-size:24px;font-weight:800;color:#34A853;"><code>/proc/PID/fd</code></div>
<div style="font-size:22px;line-height:1.6;color:#222;">Contains symbolic links to the files, pipes, sockets, and other objects opened by a process.</div>
<div style="background:#E6F4EA;padding:9px 12px;border-radius:7px;margin-top:8px;font-size:20px;"><code>ls -l /proc/$$/fd</code></div>
</div>

<div style="border:2px solid #FF7A00;border-radius:12px;padding:14px 18px;margin:10px 0;">
<div style="font-size:24px;font-weight:800;color:#FF7A00;"><code>/proc/PID/io</code></div>
<div style="font-size:22px;line-height:1.6;color:#222;">Reports certain process I/O counters, such as bytes read and written and the number of read/write system calls.</div>
<div style="background:#FFEDD9;padding:9px 12px;border-radius:7px;margin-top:8px;font-size:20px;"><code>cat /proc/$$/io</code></div>
</div>

<div style="background:#E6F4EA;border-left:8px solid #34A853;border-radius:10px;padding:16px 22px;margin:18px 0;color:#222;">
<div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:6px;">Key Takeaway</div>
<div style="font-size:23px;line-height:1.7;">
<ul>
<li><code>/proc</code> provides information about the running system.</li>
<li><code>/proc/PID</code> provides information about an individual process.</li>
<li>Most entries are virtual and kernel-generated, not ordinary files stored permanently on disk.</li>
<li>Many entries are readable by regular users, but access to some process and system details may be restricted.</li>
</ul>
</div>
</div>


````python
!ls /proc | head -20
!cat /proc/cpuinfo | head -15
!grep -E 'MemTotal|MemAvailable' /proc/meminfo
!cat /proc/uptime
!ls /proc/$$/fd
````

  <div style="

  <div style="

    </div>
    <div style="

      <code>/home/student/file.txt</code>.
    </div>
  </div>

  <div style="

    </div>

      <code>./file.txt</code>.
      <code>.</code> = current directory,
      <code>..</code> = parent directory.
    </div>
  </div>

<h3 style="
    color:#4C8BF5;

    padding-left:12px;
    margin:14px 0 10px;
    font-size:32px;
">
  3.7 Navigation and File Inspection Commands
</h3>


````python
%%bash
cd /var/log && pwd
cd - >/dev/null; cd ~; pwd
find ~ -maxdepth 2 -type f -name "*.txt" 2>/dev/null | head
echo "----- os-release"; cat /etc/os-release | head -4
echo "----- head /etc/passwd"; head -n 10 /etc/passwd
echo "----- tail dpkg.log"; tail -n 20 /var/log/dpkg.log 2>/dev/null || echo "(no dpkg.log)"
````

<h3 style="
    color:#9E4CF5;
    border-left:8px solid #9E4CF5;
    padding-left:12px;
    margin:14px 0 10px;
    font-size:32px;
">
  3.8 Filesystems, Disks and Partitions
</h3>

<p style="
    font-size:30px;
    line-height:1.65;
    color:#222;
    margin:10px 0 16px;
">
  A disk is divided into <b>partitions</b>; each partition can hold a
  <b>filesystem</b>. Modern systems expose disks as
  <code>/dev/sda</code>, <code>/dev/sdb</code> or
  <code>/dev/nvme0n1</code> (NVMe partitions:
  <code>/dev/nvme0n1p1</code>).
  The older <code>/dev/hda</code>, <code>/dev/hdb</code> naming is
  <b>mostly historical</b>.
</p>

<div style="text-align:center;margin:20px 0;">
  <img src="images/SATA_NVMe_Device_Naming.jpg"
       alt="SATA/USB vs. NVMe device naming"
       style="
         width:82%;
         max-width:100%;
         border-radius:14px;
         border:3px solid #9E4CF5;
         box-shadow:0 6px 18px rgba(0,0,0,.18);
       ">
  <div style="
      font-size:26px;
      color:#666;
      font-style:italic;
      margin-top:8px;
  ">
    SATA/USB vs. NVMe device naming
  </div>
</div>

<div style="
    background:#E6F4EA;
    border-left:8px solid #34A853;
    border-radius:10px;
    padding:18px 22px;
    margin:16px 0;
    color:#222;
">
  <div style="
      font-weight:800;
      color:#34A853;
      font-size:30px;
      margin-bottom:8px;
  ">
    Read-only First
  </div>

  <div style="
      font-size:30px;
      line-height:1.65;
  ">
    Use <code>lsblk</code>, <code>findmnt</code>, <code>df</code> and
    <code>blkid</code> <b>before</b> any partition editor.
    <code>mkfs</code>, <code>fdisk</code> writes, <code>parted</code>,
    <code>wipefs</code> and <code>dd</code> can destroy data when misused.
  </div>
</div>

<p style="
    font-size:30px;
    line-height:1.6;
    color:#222;
    margin:12px 0 16px;
">
  <b>Inspect storage without modifying it:</b>
</p>

````python
%%bash
lsblk -o NAME,SIZE,FSTYPE,TYPE,MOUNTPOINTS 2>/dev/null || lsblk
echo "----- findmnt"; findmnt | head -12
echo "----- df -hT"; df -hT | head -12
echo "----- fdisk / blkid (need sudo)"
sudo -n fdisk -l 2>&1 | head -5; sudo -n blkid 2>&1 | head -5
````

**Visualise disk usage** — a donut chart of the root filesystem (read-only):

````python
import shutil
import matplotlib.pyplot as plt
total, used, free = shutil.disk_usage("/")
gb = 1024**3
fig, ax = plt.subplots(figsize=(5.5, 5.5))
wedges, _ = ax.pie([used, free], colors=["#EA4335", "#34A853"], startangle=90, wedgeprops=dict(width=0.38, edgecolor="white", linewidth=3))
ax.text(0, 0.08, f"{used/total:.0%}", ha="center", va="center", fontsize=34, fontweight="bold", color="#1B2A49")
ax.text(0, -0.18, "used", ha="center", fontsize=13, color="#555")
ax.legend(wedges, [f"Used  {used/gb:.1f} GiB", f"Free  {free/gb:.1f} GiB"], loc="lower center", ncol=2, frameon=False, bbox_to_anchor=(0.5, -0.06))
ax.set_title(f"Root filesystem  ( / )  —  {total/gb:.1f} GiB total", fontsize=14, fontweight="bold", color="#1B2A49")
plt.tight_layout(); plt.show()
````

<h3 style="color:#00A6A6;border-left:8px solid #00A6A6;padding-left:12px;margin:14px 0 6px;"> 3.10  systemd targets instead of runlevels</h3><p style="font-size:15px;">Old Unix/Linux systems described modes as <b>runlevels 0–6</b>. Modern Ubuntu uses <b>systemd</b>, so learn <b>targets</b> as the current idea. Runlevel numbers still matter historically — many textbooks and guides use them.</p><div style="text-align:center;margin:14px 0;"><div style="font-size:13px;color:#666;font-style:italic;margin-top:8px;">Traditional runlevels mapped to systemd targets</div></div><table style="border-collapse:separate;border-spacing:0;width:100%;margin:12px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);"><thead><tr><th style="background:#00A6A6;color:white;padding:10px 12px;text-align:left;font-size:14px;">Runlevel</th><th style="background:#00A6A6;color:white;padding:10px #e3e8f0;background:#DDF6F6;color:#222;font-size:14px;"><code>poweroff.target</code></td><td style="padding:9px 12px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:14px;">Power off</td></tr><tr><td style="padding:9px 12px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:14px;font-weight:700;">1</td><td style="padding:9px 12px;border-bottom:1px solid / single-user style maintenance</td></tr><tr><td style="padding:9px 12px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:14px;font-weight:700;">2 / 3 / 4</td><td #e3e8f0;background:#DDF6F6;color:#222;font-size:14px;">Multi-user text mode (historical meaning varies by distro)</td></tr><tr><td style="padding:9px 12px;border-bottom:1px solid <code>/etc/systemd/system/default.target</code></p>

````python
%%bash
systemctl get-default 2>&1
echo "-----"
systemctl list-units --type=target --no-pager 2>&1 | head -12
echo "-----"
ls -l /etc/systemd/system/default.target 2>&1
````

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.1 Command-Line Basics</h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">A <b>CLI</b> (Command Line Interface) is a way to talk to the operating system or to programs by <b>typing commands</b>. Every command follows the same general pattern:</p>
<div style="display:flex;justify-content:center;flex-wrap:wrap;margin:16px 0;"><div style="text-align:center;margin:8px 10px;"><div style="background:#E8F0FE;border:3px solid #4C8BF5;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:44px;font-weight:800;color:#4C8BF5;">grep</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">command</div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#FFEDD9;border:3px solid #FF7A00;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:44px;font-weight:800;color:#FF7A00;">-n</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">short option<br>(one dash)</div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#F3E8FF;border:3px solid #9E4CF5;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:44px;font-weight:800;color:#9E4CF5;">--color</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">long option<br>(two dashes)</div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#E6F4EA;border:3px solid #34A853;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:44px;font-weight:800;color:#34A853;">configuration</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">pattern</div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#DDF6F6;border:3px solid #00A6A6;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:44px;font-weight:800;color:#00A6A6;">linux-commands</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">file</div></div></div>
<div style="background:#E6F4EA;border-left:8px solid #34A853;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#34A853;font-size:30px;margin-bottom:6px;">Reading a SYNOPSIS</div><div style="font-size:28px;line-height:1.6;">Each command has a <b>synopsis</b> that shows its general usage, for example <code>ls [options] [path]</code>. Square brackets <b>[ ]</b> mean the part is <b>optional</b>.</div></div>

<div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">Case-sensitive</div><div style="font-size:26px;color:#333;line-height:1.5;">Linux commands and file names distinguish upper and lower case: <code>ls</code> ≠ <code>LS</code>.</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">Switches / options</div><div style="font-size:26px;color:#333;line-height:1.5;">Switches add features and make a command flexible and interactive.</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">Short options</div><div style="font-size:26px;color:#333;line-height:1.5;">One letter after a single dash: <code>-l</code>. Several can share one dash: <code>-l -t -h</code> = <code>-lth</code>.</div></div>
<div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">Long options</div><div style="font-size:26px;color:#333;line-height:1.5;">A word or phrase after <b>two dashes</b>: <code>--color</code>. Each long option stands alone.</div></div>
</div>

````python
%%bash
echo "--- three equivalent spellings"
ls -l -a /etc | head -3
ls -la /etc | head -3
ls --all -l /etc | head -3
echo
echo "--- commands are case-sensitive"
LS 2>&1 || true
````

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.2 Navigation</h3>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#9E4CF5;color:white;padding:14px 16px;text-align:left;font-size:27px;width:30%;">Command</th><th style="background:#9E4CF5;color:white;padding:14px 16px;text-align:left;font-size:27px;width:70%;">What it does</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:26px;font-weight:700;"><code>pwd</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:26px;"><b>p</b>rint <b>w</b>orking <b>d</b>irectory — shows the current path</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:26px;font-weight:700;"><code>cd directory</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:26px;"><b>c</b>hange <b>d</b>irectory — enter a directory, e.g. <code>cd Desktop</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:26px;font-weight:700;"><code>cd /</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:26px;">Go to the <b>root</b> directory</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:26px;font-weight:700;"><code>cd ..</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:26px;">Go one level <b>up</b> (like the Up button of a file explorer)</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:26px;font-weight:700;"><code>cd -</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:26px;">Jump <b>back</b> to the previous directory (like the Back button)</td></tr>
<tr><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:26px;font-weight:700;"><code>clear</code></td><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:26px;">Clear the terminal screen (scroll up to see old output)</td></tr>
</tbody>
</table>

<div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:30px;margin-bottom:6px;">⛔ Power commands — shown as text only</div><div style="font-size:28px;line-height:1.6;"><code>reboot</code> restarts the machine and <code>halt</code> shuts it down. They normally need administrator rights (<code>sudo</code>), and they are deliberately <b>not</b> provided as runnable cells.</div></div>

````python
%%bash
cd ~;   echo "Home     : $(pwd)"
cd /etc; echo "Now in   : $(pwd)"
cd ..;  echo "One up   : $(pwd)"
cd -                       # prints the directory it jumps back to
````

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.3 Listing Files with <code>ls</code></h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>ls</code> (<b>l</b>i<b>s</b>t) shows the files and directories of the current path or of a path you give. Without a path it works on the current directory.</p>
<div style="display:grid;grid-template-columns:repeat(4,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">-S</div><div style="font-size:26px;color:#333;line-height:1.5;">Sort by <b>size</b> (default order is alphabetical)</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">-t</div><div style="font-size:26px;color:#333;line-height:1.5;">Sort by <b>modification time</b></div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">-l</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>Long</b> list format — more details</div></div>
<div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">-a</div><div style="font-size:26px;color:#333;line-height:1.5;">Show <b>hidden</b> files too</div></div>
</div>
<div style="background:#FFEDD9;border-left:8px solid #FF7A00;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#FF7A00;font-size:30px;margin-bottom:6px;">Hidden files</div><div style="font-size:28px;line-height:1.6;">To hide a file in Linux, start its name with a <b>dot</b>. In a graphical file manager choose <i>View → Show Hidden Files</i>.</div></div>

<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">One command can combine several switches — here <b>long format</b>, sorted by <b>time</b>, with <b>human-readable</b> sizes:</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:25px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> ls -lth</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">total 62K</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">drwxr-xr-x. 2 root root 4.0K Dec  3 22:25 Documents</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">drwxr-xr-x. 2 root root 4.0K Dec  2 20:47 Downloads</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">drwxr-xr-x. 5 root root 4.0K Nov 29 04:38 Progs</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">drwxr-xr-x. 2 root root 4.0K Nov 29 04:34 Desktop</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">-rw-r--r--. 1 root root  46K Nov 28 15:22 install.log</div>
</div>
<p style="font-size:24px;line-height:1.65;color:#222;margin:10px 0;">The trailing dot after the permissions is an <b>SELinux</b> marker (common on Red Hat–based systems); Ubuntu normally omits it.</p>

````python
%%bash
cd /etc
echo "--- newest first (-t), long (-l), human sizes (-h)"
ls -lth | head -6
echo "--- biggest first (-S)"
ls -lSh | head -5
echo "--- hidden files in your home directory (-a)"
ls -a ~ | head -10
````

<h3 style="color:#00A6A6;border-left:8px solid #00A6A6;padding-left:12px;margin:14px 0 10px;font-size:32px;">Reading the <code>ls -l</code> output</h3>
<p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">From left to right every line shows:</p>
<div style="display:grid;grid-template-columns:repeat(4,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">1 · Permissions</div><div style="font-size:26px;color:#333;line-height:1.5;">e.g. <code>drwxr-xr-x</code></div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">2 · Links</div><div style="font-size:26px;color:#333;line-height:1.5;">Number of links — for a directory it reflects its subdirectories</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">3 · Owner</div><div style="font-size:26px;color:#333;line-height:1.5;">User that owns the file or directory</div></div>
<div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">4 · Group</div><div style="font-size:26px;color:#333;line-height:1.5;">Group the item belongs to</div></div>
<div style="background:#FCE8E6;border-top:6px solid #EA4335;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#EA4335;font-size:28px;margin-bottom:8px;">5 · Size</div><div style="font-size:26px;color:#333;line-height:1.5;">Bytes by default; <code>-h</code> makes it readable (4.0K, 46K)</div></div>
<div style="background:#DDF6F6;border-top:6px solid #00A6A6;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#00A6A6;font-size:28px;margin-bottom:8px;">6 · Modified</div><div style="font-size:26px;color:#333;line-height:1.5;">Date of the last change</div></div>
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">7 · Name</div><div style="font-size:26px;color:#333;line-height:1.5;">File or directory name</div></div>
</div>
<p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><b>The permission string has 4 parts:</b></p>
<div style="display:flex;justify-content:center;flex-wrap:wrap;margin:16px 0;"><div style="text-align:center;margin:8px 10px;"><div style="background:#F3E8FF;border:3px solid #9E4CF5;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:46px;font-weight:800;color:#9E4CF5;">d</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">Part 1 — type<br><b>d</b> directory · <b>-</b> file · <b>l</b> link</div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#E8F0FE;border:3px solid #4C8BF5;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:46px;font-weight:800;color:#4C8BF5;">rwx</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">Part 2 — <b>owner</b> (user)</div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#E6F4EA;border:3px solid #34A853;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:46px;font-weight:800;color:#34A853;">r-x</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">Part 3 — <b>group</b></div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#FFEDD9;border:3px solid #FF7A00;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:46px;font-weight:800;color:#FF7A00;">r-x</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">Part 4 — <b>others</b></div></div></div>
<p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><code>r</code> read · <code>w</code> write · <code>x</code> execute · <code>-</code> permission absent</p>

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.4 Permissions — <code>chmod</code></h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>chmod</code> (<b>ch</b>ange <b>mod</b>e) changes the permissions of files and directories. Permissions decide <b>who</b> may read, write or execute an item.</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:25px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> chmod [options] value file-path-and-name</div>
</div>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Besides letters, every permission has a <b>number</b>. The three digits (owner · group · others) are the <b>sum</b> of the permissions granted:</p>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#EA4335;color:white;padding:14px 16px;text-align:left;font-size:25px;width:15%;">Digit</th><th style="background:#EA4335;color:white;padding:14px 16px;text-align:left;font-size:25px;width:25%;">Letters</th><th style="background:#EA4335;color:white;padding:14px 16px;text-align:left;font-size:25px;width:60%;">Meaning</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;font-weight:700;">7</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;"><code>rwx</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;">4 + 2 + 1 — read, write, execute</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;">6</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;"><code>rw-</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">4 + 2 — read, write</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;font-weight:700;">5</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;"><code>r-x</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;">4 + 1 — read, execute</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;">4</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;"><code>r--</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">read only</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;font-weight:700;">2</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;"><code>-w-</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;">write only</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;">1</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;"><code>--x</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">execute only</td></tr>
<tr><td style="padding:11px 16px;background:#FCE8E6;color:#222;font-size:24px;font-weight:700;">0</td><td style="padding:11px 16px;background:#FCE8E6;color:#222;font-size:24px;"><code>---</code></td><td style="padding:11px 16px;background:#FCE8E6;color:#222;font-size:24px;">no permission</td></tr>
</tbody>
</table>

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">Worked example</h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">A directory has <b>rwxr-x--x</b>:</p>
<div style="display:flex;justify-content:center;flex-wrap:wrap;margin:16px 0;"><div style="text-align:center;margin:8px 10px;"><div style="background:#E8F0FE;border:3px solid #4C8BF5;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:52px;font-weight:800;color:#4C8BF5;">7</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">owner<br>4+2+1 = <b>rwx</b></div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#E6F4EA;border:3px solid #34A853;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:52px;font-weight:800;color:#34A853;">5</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">group<br>4+0+1 = <b>r-x</b></div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#FFEDD9;border:3px solid #FF7A00;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:52px;font-weight:800;color:#FF7A00;">1</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">others<br>0+0+1 = <b>--x</b></div></div></div>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Its mode is therefore <b>751</b>. We now want: <b>owner keeps everything</b>, <b>group may only read</b>, <b>others get nothing</b>:</p>
<div style="display:flex;justify-content:center;flex-wrap:wrap;margin:16px 0;"><div style="text-align:center;margin:8px 10px;"><div style="background:#E8F0FE;border:3px solid #4C8BF5;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:52px;font-weight:800;color:#4C8BF5;">7</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">owner 4+2+1</div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#E6F4EA;border:3px solid #34A853;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:52px;font-weight:800;color:#34A853;">4</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">group 4+0+0</div></div><div style="text-align:center;margin:8px 10px;"><div style="background:#FFEDD9;border:3px solid #FF7A00;border-radius:12px;padding:10px 22px;font-family:monospace;font-size:52px;font-weight:800;color:#FF7A00;">0</div><div style="font-size:24px;color:#444;margin-top:6px;max-width:270px;">others 0+0+0</div></div></div>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:25px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> chmod 740 Desktop/mydirectory</div>
</div>
<p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">New permission: <b>rwxr-----</b> = 740. Useful options: <code>-v</code> prints what changed, <code>-R</code> also applies the mode to all <b>sub-directories and files</b>.</p>

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">Symbolic mode: <code>u</code> <code>g</code> <code>o</code> <code>a</code> with <code>=</code> <code>+</code> <code>-</code></h3>
<p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><code>u</code> owner (user) · <code>g</code> group · <code>o</code> others · <code>a</code> all.  Starting from a file with mode <b>754</b>:</p>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#EA4335;color:white;padding:14px 16px;text-align:left;font-size:25px;width:44%;">Command</th><th style="background:#EA4335;color:white;padding:14px 16px;text-align:left;font-size:25px;width:41%;">Effect</th><th style="background:#EA4335;color:white;padding:14px 16px;text-align:left;font-size:25px;width:15%;">New mode</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;font-weight:700;"><code>chmod u=rwx,g=rx,o=r myfile</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;"><b>Replaces</b> all previous permissions → 754</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;">754</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;"><code>chmod o=rw myfile</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">Only the “others” part becomes rw</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">756</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;font-weight:700;"><code>chmod o+x myfile</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;"><b>Add</b> execute for others</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:24px;">757</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;"><code>chmod g-r myfile</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;"><b>Remove</b> read from the group</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">717</td></tr>
<tr><td style="padding:11px 16px;background:#FCE8E6;color:#222;font-size:24px;font-weight:700;"><code>chmod a+w myfile</code></td><td style="padding:11px 16px;background:#FCE8E6;color:#222;font-size:24px;">Add write for <b>all</b> users</td><td style="padding:11px 16px;background:#FCE8E6;color:#222;font-size:24px;">737</td></tr>
</tbody>
</table>
<p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">In a graphical file manager: right-click the item → <i>Properties</i> → <i>Permissions</i> tab.</p>

````python
%%bash
mkdir -p /tmp/exp3_lab && cd /tmp/exp3_lab
touch myfile
chmod 754 myfile;   stat -c '%a  %A  (754)' myfile
chmod o=rw myfile;  stat -c '%a  %A  (o=rw)' myfile
chmod o+x myfile;   stat -c '%a  %A  (o+x)' myfile
chmod g-r myfile;   stat -c '%a  %A  (g-r)' myfile
chmod a+w myfile;   stat -c '%a  %A  (a+w)' myfile
chmod -v 600 myfile
````

````python
from IPython.display import display, HTML

def sym(mode):
    return "".join(("r" if int(d) & 4 else "-") + ("w" if int(d) & 2 else "-") + ("x" if int(d) & 1 else "-") for d in str(mode))

examples = [(751, "the directory from the example"), (740, "owner everything, group read, others nothing"),
            (754, "u=rwx,g=rx,o=r"), (756, "after  o=rw"), (757, "after  o+x"), (717, "after  g-r"), (737, "after  a+w"),
            (644, "typical text file"), (755, "typical program / directory"), (600, "private file")]
cols = ["#E8F0FE", "#E6F4EA", "#FFEDD9"]
rows = ""
for m, why in examples:
    s = sym(m)
    parts = "".join(f'<td style="padding:8px 14px;background:{cols[i]};font-family:monospace;font-size:20px;font-weight:700;">{s[i*3:i*3+3]}</td>' for i in range(3))
    rows += f'<tr><td style="padding:8px 14px;font-weight:800;font-size:20px;">{m}</td>{parts}<td style="padding:8px 14px;font-size:18px;color:#444;">{why}</td></tr>'
display(HTML('<table style="border-collapse:collapse;border:2px solid #EA4335;border-radius:10px;overflow:hidden;">'
             '<tr style="background:#EA4335;color:white;"><th style="padding:8px 14px;">mode</th><th>owner</th><th>group</th><th>others</th><th style="text-align:left;padding:8px 14px;">note</th></tr>'
             + rows + '</table>'))
````

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.5 Owners and Privileges — <code>chown</code>, <code>su</code>, <code>sudo</code></h3>
<div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">chown</div><div style="font-size:26px;color:#333;line-height:1.5;">Change the <b>owner and group</b> of a file.<br><code>sudo chown username:groupname file</code></div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">su</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>S</b>witch <b>u</b>ser inside the terminal.<br><code>su username</code><br>Return with <code>exit</code>.</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">sudo</div><div style="font-size:26px;color:#333;line-height:1.5;">Run <b>one</b> administrative command with root rights, without switching to root or knowing the root password.</div></div>
</div>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#4C8BF5;color:white;padding:14px 16px;text-align:left;font-size:26px;width:35%;">Prompt</th><th style="background:#4C8BF5;color:white;padding:14px 16px;text-align:left;font-size:26px;width:65%;">Meaning</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E8F0FE;color:#222;font-size:25px;font-weight:700;"><code>student@oslab:~&#36;</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E8F0FE;color:#222;font-size:25px;">Regular (non-root) user — the prompt ends with <b>&#36;</b></td></tr>
<tr><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:25px;font-weight:700;"><code>root@oslab:~#</code></td><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:25px;">Root user — the prompt ends with <b>#</b></td></tr>
</tbody>
</table>
<p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">When you type a password in a terminal, <b>nothing appears</b> (no <code>&#42;</code> or ●). Type carefully and press <kbd>Enter</kbd>.</p>

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">How <code>sudo</code> differs between distributions</h3>
<div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🟠</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">Debian / Ubuntu</div><div style="font-size:26px;color:#333;line-height:1.5;">The <b>root account is disabled</b> by default. The user created during installation can use <code>sudo</code> immediately — put <code>sudo</code> at the start of an administrative command.</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🔴</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">Red Hat–based</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>sudo</b> is often <b>not enabled</b> by default for ordinary users; administrators log in as root or add users to the sudoers.</div></div>
</div>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:25px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo passwd root</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">[sudo] password for student:</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">New password:</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">Retype new password:</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">passwd: password updated successfully</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> su -          # or:  su root</div>
</div>
<div style="background:#E6F4EA;border-left:8px solid #34A853;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:6px;">Security note</div><div style="font-size:26px;line-height:1.6;">Enabling the root account is optional. Prefer <code>sudo</code> for single commands; for a root shell use <code>sudo -i</code>.</div></div>

````python
%%bash
id
echo "--- owner:group of your home directory (what chown would change)"
stat -c '%U:%G  %n' ~
echo "--- prompt symbol for this user: $( [ "$(id -u)" -eq 0 ] && echo '#  (root)' || echo '$  (regular user)' )"
echo "--- does sudo work without a password here?"
sudo -n true 2>/dev/null && echo "yes" || echo "no - sudo asks for a password (or is not installed)"
````

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.6 User and Group Management</h3>

<p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">
Linux manages access to files and system resources using <b>users</b>, <b>groups</b>, and numeric identifiers. These commands modify system accounts, so practice on a <b>VM</b>.
</p>

<div style="background:#E6F4EA;border-left:8px solid #34A853;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;">
<div style="font-weight:800;color:#24853B;font-size:27px;margin-bottom:8px;">1. Core Concepts</div>
<div style="font-size:24px;line-height:1.8;">
<b>User:</b> An account that can access the system, such as <code>student</code>.<br>
<b>Group:</b> A collection of users that can share permissions, such as <code>students</code> or <code>docker</code>.<br>
<b>UID:</b> The numeric identifier assigned to a user.<br>
<b>GID:</b> The numeric identifier assigned to a group.<br>
<b>Primary Group:</b> The user's main group, commonly used as the default group for new files.<br>
<b>Supplementary Groups:</b> Additional groups that provide further access.
</div>
</div>

<p style="font-size:25px;line-height:1.65;color:#222;margin:12px 0;">
For example, <code>student</code> may have <code>student</code> as the primary group and <code>sudo</code> and <code>adm</code> as supplementary groups.
</p>

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:24px 0 10px;font-size:29px;">2. Creating a User — <code>useradd</code></h3>

<p style="font-size:24px;line-height:1.65;color:#222;">General syntax:</p>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:10px 0;font-size:24px;color:#222;">
<code>sudo useradd [options] username</code>
</div>

<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.12);">
<thead><tr><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:23px;width:20%;">Option</th><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:23px;">Purpose</th></tr></thead>
<tbody>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>-m</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Create the user's home directory.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>-g</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Specify the primary group.</td></tr>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>-G</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Specify supplementary groups, separated by commas.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>-s</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Set the default login shell.</td></tr>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>-u</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Specify a UID.</td></tr>
</tbody>
</table>


<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Example 1 — Create a user</div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo useradd -m student</pre>
<p style="font-size:22px;line-height:1.6;margin:8px 0 0;">Creates the <code>student</code> account and its home directory, typically <code>/home/student</code>.</p>
</div>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Example 2 — Add supplementary groups</div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo useradd -m -G sudo,adm student</pre>
<p style="font-size:22px;line-height:1.6;margin:8px 0 0;"><code>student</code> is created with membership in <code>sudo</code> and <code>adm</code>, assuming those groups exist.</p>
</div>

<div style="background:#FFF4E5;border-left:8px solid #FF7A00;border-radius:10px;padding:14px 20px;margin:14px 0;color:#222;">
<div style="font-weight:800;color:#B85C00;font-size:24px;margin-bottom:6px;">Security Note</div>
<div style="font-size:22px;line-height:1.65;">Members of <code>sudo</code> can typically run authorized administrative commands. The <code>adm</code> group may provide access to certain system logs. Grant membership only when necessary.</div>
</div>



<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:24px 0 10px;font-size:29px;">3. Setting a Password — <code>passwd</code></h3>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:10px 0;color:#222;">
<pre style="font-size:23px;white-space:pre-wrap;margin:0;">sudo passwd student</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">Sets or changes the user's password. Linux normally does not display characters while you type the password.</p>
</div>

<div style="background:#E6F4EA;border-left:8px solid #34A853;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;color:#24853B;font-size:24px;margin-bottom:6px;">Remember</div>
<div style="font-size:22px;line-height:1.65;">Creating an account does not necessarily set a password. Without <code>-m</code>, <code>useradd</code> does not normally create the home directory unless system configuration specifies otherwise.</div>
</div>

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:24px 0 10px;font-size:29px;">4. Modifying a User — <code>usermod</code></h3>

<p style="font-size:24px;line-height:1.65;color:#222;">Use <code>usermod</code> to modify an existing account.</p>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:10px 0;font-size:23px;color:#222;">
<code>sudo usermod [options] username</code>
</div>

<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.12);">
<thead><tr><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:23px;width:20%;">Option</th><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:23px;">Purpose</th></tr></thead>
<tbody>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>-aG</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Append supplementary group memberships without replacing existing ones.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>-G</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Replace the supplementary group list.</td></tr>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>-g</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Change the primary group.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>-l</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Change the login name.</td></tr>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>-s</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Change the default shell.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>-L / -U</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Lock or unlock the password, respectively.</td></tr>
</tbody>
</table>






<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Example 1 — Add a user to a group</div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo usermod -aG docker student</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">Adds <code>docker</code> to the supplementary groups of <code>student</code>, preserving existing memberships.</p>
</div>

<div style="background:#FFF4E5;border-left:8px solid #FF7A00;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;color:#B85C00;font-size:24px;margin-bottom:6px;">Why is <code>-aG</code> important?</div>
<div style="font-size:22px;line-height:1.65;">If the user belongs to <code>sudo</code> and <code>adm</code>, using <code>usermod -G docker student</code> may remove those supplementary memberships. Using <code>-aG</code> adds <code>docker</code> instead. Membership in <code>docker</code> can grant root-equivalent privileges on many systems.</div>
</div>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Example 2 — Change the shell</div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">which fish
sudo usermod -s /usr/bin/fish student</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">Sets Fish as the default shell if it is installed and the path is correct. The shell determines how commands entered in the terminal are interpreted.</p>
</div>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Example 3 — Rename a user</div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo usermod -l newstudent student</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">Changes the login name but does not automatically rename the home directory.</p>
</div>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Example 4 — Lock and unlock a password</div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo usermod -L student
sudo usermod -U student</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">These commands lock and unlock password authentication. They do not necessarily disable every possible login method.</p>
</div>


<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:24px 0 10px;font-size:29px;">5. Checking User IDs — <code>id</code></h3>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:10px 0;color:#222;">
<pre style="font-size:23px;white-space:pre-wrap;margin:0;">id student</pre>
<p style="font-size:22px;line-height:1.6;margin:8px 0;">Example output:</p>
<pre style="font-size:21px;white-space:pre-wrap;margin:0;">uid=1001(student) gid=1001(student) groups=1001(student),4(adm),27(sudo),999(docker)</pre>
</div>

<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;">
<thead><tr><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:23px;">Command output</th><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:23px;">Meaning</th></tr></thead>
<tbody>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>uid=1001(student)</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">User ID and username.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>gid=1001(student)</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Primary group ID and group name.</td></tr>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>groups=...</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Group IDs and names.</td></tr>
</tbody>
</table>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Useful variations</div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">id
id -u student
id -g student
id -G student
id -Gn student</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;"><code>id</code> shows the current user's information. The <code>-u</code> and <code>-g</code> options show the user and primary group IDs; <code>-G</code> shows all group IDs, while <code>-Gn</code> shows group names.</p>
</div>




<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:24px 0 10px;font-size:29px;">6. Listing Groups — <code>groups</code></h3>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:10px 0;color:#222;">
<pre style="font-size:23px;white-space:pre-wrap;margin:0;">groups student</pre>
<p style="font-size:22px;line-height:1.6;margin:8px 0;">Example output:</p>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">student : student adm sudo docker</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">Lists the groups associated with <code>student</code>. Use <code>groups</code> without a username to check the current user's groups.</p>
</div>

<p style="font-size:23px;line-height:1.65;color:#222;">Unlike <code>groups</code>, <code>id</code> also displays numeric identifiers such as UID and GID.</p>

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:24px 0 10px;font-size:29px;">7. Deleting a User — <code>userdel</code></h3>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:10px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Delete the account only</div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo userdel student</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 12px;">Removes the account but does not necessarily remove its home directory.</p>
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Delete the account and home directory</div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo userdel -r student</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">The <code>-r</code> option also attempts to remove the home directory and local mail spool.</p>
</div>

<div style="background:#FCE8E6;border-left:8px solid #D93025;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;color:#B3261E;font-size:24px;margin-bottom:6px;">Warning — Data Loss</div>
<div style="font-size:22px;line-height:1.65;">Check for important files and create a backup before running <code>userdel -r</code>. Files outside the home directory may remain on the system.</div>
</div>


<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:24px 0 10px;font-size:29px;">8 Creating and Managing Groups</h3>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Create a group — <code>groupadd</code></div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo groupadd developers</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">Creates the <code>developers</code> group. It does not automatically add any users.</p>
</div>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Add a user to the group</div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo usermod -aG developers student
groups student</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">Adds <code>student</code> to <code>developers</code> and then checks the user's group memberships.</p>
</div>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Rename a group — <code>groupmod</code></div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo groupmod -n programmers developers</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">Renames <code>developers</code> to <code>programmers</code>. The usernames of its members do not change.</p>
</div>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Delete a group — <code>groupdel</code></div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">sudo groupdel programmers</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">Deletes the group if system constraints permit. It does not delete the accounts of its members. A group that is still the primary group of a user generally cannot be deleted.</p>
</div>

<div style="background:#F3F5F7;border-radius:10px;padding:14px 20px;margin:12px 0;color:#222;">
<div style="font-weight:800;font-size:24px;margin-bottom:8px;">Look up group information — <code>getent</code></div>
<pre style="font-size:22px;white-space:pre-wrap;margin:0;">getent group programmers</pre>
<p style="font-size:22px;line-height:1.65;margin:8px 0 0;">Displays the group entry if it is available through the system's configured account databases.</p>
</div>



<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:24px 0 10px;font-size:29px;">9. Command Summary</h3>

<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.12);">
<thead><tr><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:23px;width:32%;">Command</th><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:23px;">Main Function</th></tr></thead>
<tbody>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>useradd</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Create a user.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>usermod</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Modify an existing user.</td></tr>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>passwd</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Set or change a password.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>id</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Display user and group IDs.</td></tr>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>groups</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">List group memberships.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>userdel</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Delete a user account.</td></tr>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>groupadd</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Create a group.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>groupmod</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Modify or rename a group.</td></tr>
<tr><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;"><code>groupdel</code></td><td style="padding:10px 16px;background:#E6F4EA;font-size:22px;">Delete a group.</td></tr>
<tr><td style="padding:10px 16px;background:white;font-size:22px;"><code>getent group</code></td><td style="padding:10px 16px;background:white;font-size:22px;">Look up group information.</td></tr>
</tbody>
</table>

<div style="background:#E6F4EA;border-left:8px solid #34A853;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;">
<div style="font-weight:800;color:#24853B;font-size:27px;margin-bottom:6px;">The Most Important Distinction</div>
<div style="font-size:24px;line-height:1.7;">Creating a user, adding a user to a group, and granting file permissions are three different operations. Adding <code>student</code> to <code>labteam</code> does not automatically grant access to every file owned by that group. File ownership and permissions must also allow the required access.</div>
</div>

````python
%%bash
me=$(id -un)
echo "--- id and groups"
id "$me"
groups "$me"
echo "--- account record: name, UID, GID, home, shell"
getent passwd "$me" | cut -d: -f1,3,4,6,7
echo "--- members of the sudo / adm groups"
getent group sudo adm
````

<h3 style="color:#00A6A6;border-left:8px solid #00A6A6;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.7 Searching — <code>find</code></h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>find</code> locates files by <b>name, owner, group, type, permissions, date</b> and more. It searches <b>all sub-directories</b> of the given path.</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:25px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> find [path] [options] file-expression</div>
</div>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#00A6A6;color:white;padding:14px 16px;text-align:left;font-size:26px;width:48%;">Task</th><th style="background:#00A6A6;color:white;padding:14px 16px;text-align:left;font-size:26px;width:52%;">Command</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:25px;font-weight:700;">File named <i>test</i> in the current path</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:25px;"><code>find -name test</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;font-weight:700;">Names that <b>start with</b> <i>test</i></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;"><code>find -name "test&#42;"</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:25px;font-weight:700;"><i>text.txt</i> under <i>/home/student/</i></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:25px;"><code>find /home/student/ -name text.txt</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;font-weight:700;"><i>linux.pdf</i>, <b>ignoring case</b></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;"><code>find -iname linux.pdf</code></td></tr>
<tr><td style="padding:11px 16px;background:#DDF6F6;color:#222;font-size:25px;font-weight:700;">Files of <b>10 MB or more</b></td><td style="padding:11px 16px;background:#DDF6F6;color:#222;font-size:25px;"><code>find -size +10M</code></td></tr>
</tbody>
</table>

````python
%%bash
mkdir -p /tmp/exp3_lab/find_demo && cd /tmp/exp3_lab/find_demo
touch test test1 test.txt text.txt linux.pdf LiNuX.pdf
truncate -s 11M big.bin
echo "--- -name test";        find . -name test
echo "--- -name 'test*'";     find . -name 'test*'
echo "--- -iname linux.pdf";  find . -iname linux.pdf
echo "--- -size +10M";        find . -size +10M
echo "--- explicit path";     find /tmp/exp3_lab/find_demo -name text.txt
````

<h3 style="color:#00A6A6;border-left:8px solid #00A6A6;padding-left:12px;margin:14px 0 10px;font-size:32px;">Searching text — <code>grep</code></h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>grep</code> prints every <b>line</b> that contains a word or string.</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:25px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> grep [options] pattern file(s)</div>
</div>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#00A6A6;color:white;padding:14px 16px;text-align:left;font-size:25px;width:32%;">Option / pattern</th><th style="background:#00A6A6;color:white;padding:14px 16px;text-align:left;font-size:25px;width:68%;">Meaning</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:24px;font-weight:700;"><code>--color</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:24px;">Highlight the match in the output</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;"><code>-n</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">Show line numbers</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:24px;font-weight:700;"><code>-c</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:24px;">Only print the <b>count</b> of matching lines</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;"><code>-i</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;"><b>Ignore case</b></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:24px;font-weight:700;"><code>-l</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#DDF6F6;color:#222;font-size:24px;">Only list the <b>names of files</b> that match</td></tr>
<tr><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:24px;font-weight:700;"><code>^word</code></td><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:24px;">Lines that <b>start with</b> word (a leading space breaks the match)</td></tr>
</tbody>
</table>
<p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">Search several files with a space or a wildcard (<code>grep configuration &#42;.txt</code>). A phrase must be quoted: <code>grep 'configuration file' &#42;</code></p>

````python
%%bash
mkdir -p /tmp/exp3_lab && cd /tmp/exp3_lab
cat > linux-commands <<'EOF'
Yum uses a configuration file at /etc/yum.conf.
Additional configuration files are also read from the directories
there is nothing special on this line
Also this configuration is set to it,
There is an uppercase line
The inclusion of external configuration files is supported.
EOF
echo "configuration notes" > notes.txt; echo "nothing here" > other.txt
echo "--- --color -n";           grep --color -n configuration linux-commands
echo "--- -c (count)";           grep -c configuration linux-commands
echo "--- -i there";             grep -i there linux-commands
echo "--- ^there (line start)";  grep '^there' linux-commands
echo "--- -l (file names)";      grep -l configuration *.txt linux-commands
echo "--- quoted phrase";        grep 'configuration file' linux-commands
````

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.8 Small but Useful: <code>sort</code> · <code>who</code> · <code>whereis</code> · <code>which</code></h3>
<div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">sort</div><div style="font-size:26px;color:#333;line-height:1.5;">Sorts the <b>lines</b> of a text file: <code>sort [options] [file]</code></div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">who</div><div style="font-size:26px;color:#333;line-height:1.5;">Shows the users <b>currently logged in</b>.</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">whereis</div><div style="font-size:26px;color:#333;line-height:1.5;">Locates the <b>binary, source and manual</b> of a command. <code>-b</code> binary · <code>-m</code> manual · <code>-s</code> source</div></div>
<div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">which</div><div style="font-size:26px;color:#333;line-height:1.5;">Shows <b>which executable file</b> runs when you type a command.</div></div>
</div>

````python
%%bash
echo "--- sort"
printf 'pear\napple\nfig\nbanana\n' | sort
echo "--- sort -r (reverse)"
printf 'pear\napple\nfig\nbanana\n' | sort -r
echo "--- who: users logged in right now"
who || true
echo "--- whereis (all), binaries only (-b), manuals only (-m)"
whereis ls; whereis -b ls; whereis -m ls
echo "--- which: the file that actually runs"
which ls
````

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.9 File and Directory Management</h3>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#FF7A00;color:white;padding:14px 16px;text-align:left;font-size:25px;width:14%;">Command</th><th style="background:#FF7A00;color:white;padding:14px 16px;text-align:left;font-size:25px;width:38%;">What it does</th><th style="background:#FF7A00;color:white;padding:14px 16px;text-align:left;font-size:25px;width:48%;">Notes</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FFEDD9;color:#222;font-size:24px;font-weight:700;"><code>mkdir</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FFEDD9;color:#222;font-size:24px;">Create a new directory</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FFEDD9;color:#222;font-size:24px;"><code>mkdir path/name</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;"><code>rmdir</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">Remove a directory</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">Only <b>empty</b> directories; cannot delete files</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FFEDD9;color:#222;font-size:24px;font-weight:700;"><code>rm</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FFEDD9;color:#222;font-size:24px;">Remove files</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FFEDD9;color:#222;font-size:24px;"><code>-r</code> removes a directory <b>with all its contents</b></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;"><code>cp</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">Copy files and directories</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;"><code>cp source destination</code> — directories need <code>-r</code></td></tr>
<tr><td style="padding:11px 16px;background:#FFEDD9;color:#222;font-size:24px;font-weight:700;"><code>mv</code></td><td style="padding:11px 16px;background:#FFEDD9;color:#222;font-size:24px;">Move <b>and/or rename</b> in one step</td><td style="padding:11px 16px;background:#FFEDD9;color:#222;font-size:24px;"><code>mv old&#95;path new&#95;path</code></td></tr>
</tbody>
</table>
<div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:28px;margin-bottom:6px;">⚠️ No recycle bin</div><div style="font-size:26px;line-height:1.6;"><code>rm</code> deletes immediately — there is no undo. Check the path <b>before</b> pressing <kbd>Enter</kbd>, especially with <code>-r</code>.</div></div>

<div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">Tab completion</div><div style="font-size:26px;color:#333;line-height:1.5;">Type part of a command or name and press <kbd>Tab</kbd> to complete it. If several names share the prefix it fills the common part; <b>press Tab twice</b> to list all options.</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">Spaces in names</div><div style="font-size:26px;color:#333;line-height:1.5;">Put a backslash before the space — <code>cd My&#92; Folder</code> — or quote the name. Otherwise it is read as <b>two</b> separate names.</div></div>
</div>

````python
%%bash
mkdir -p /tmp/exp3_lab && cd /tmp/exp3_lab
rm -r files_demo 2>/dev/null || true
mkdir files_demo && cd files_demo
mkdir -p docs/archive backup 'My Folder'
echo "report v1" > docs/report.txt
cp docs/report.txt docs/report_copy.txt           # copy a file
cp -r docs backup/docs_backup                     # copy a directory (-r)
mv docs/report_copy.txt docs/archive/old.txt      # move + rename at once
ls -R
echo "--- rmdir refuses non-empty directories"
rmdir docs 2>&1 || true
echo "--- a name with a space"
cd My\ Folder && pwd && cd ..
echo "--- rm -r removes a directory with everything inside"
rm -r backup && ls
````

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.10 Archiving — <code>tar</code></h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>tar</code> (<b>T</b>ape <b>Ar</b>chive) packs many files into <b>one file</b> while keeping the folder structure and metadata.</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:25px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> tar [options] new&#95;name.tar file&#95;or&#95;directory&#95;name(s)</div>
</div>
<div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">-c</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>C</b>reate an archive</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">-f</div><div style="font-size:26px;color:#333;line-height:1.5;">The archive is this <b>f</b>ile</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">-x</div><div style="font-size:26px;color:#333;line-height:1.5;">E<b>x</b>tract an archive</div></div>
<div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">-u</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>U</b>pdate files already in the archive</div></div>
<div style="background:#FCE8E6;border-top:6px solid #EA4335;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#EA4335;font-size:28px;margin-bottom:8px;">-r</div><div style="font-size:26px;color:#333;line-height:1.5;">Append new items to an archive (placed at the top level)</div></div>
<div style="background:#DDF6F6;border-top:6px solid #00A6A6;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#00A6A6;font-size:28px;margin-bottom:8px;">-v</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>V</b>erbose — list the files processed</div></div>
</div>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#4C8BF5;color:white;padding:14px 16px;text-align:left;font-size:24px;width:34%;">Task</th><th style="background:#4C8BF5;color:white;padding:14px 16px;text-align:left;font-size:24px;width:66%;">Command</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E8F0FE;color:#222;font-size:23px;font-weight:700;">Archive two files</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E8F0FE;color:#222;font-size:23px;"><code>tar -cf Documents/myarchive.tar Desktop/e1 Desktop/e2</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:23px;font-weight:700;">Extract them</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:23px;"><code>tar -xvf myarchive.tar</code></td></tr>
<tr><td style="padding:11px 16px;background:#E8F0FE;color:#222;font-size:23px;font-weight:700;">Archive + compress with <b>xz</b></td><td style="padding:11px 16px;background:#E8F0FE;color:#222;font-size:23px;"><code>tar -cJf archive.tar.xz files…</code></td></tr>
</tbody>
</table>
<p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">Other compressors work too: <code>-z</code> (gzip) and <code>-j</code> (bzip2).</p>

````python
%%bash
mkdir -p /tmp/exp3_lab/tar_demo && cd /tmp/exp3_lab/tar_demo
mkdir -p Desktop Documents extracted
echo "experiment one" > Desktop/e1;  seq 1 2000 > Desktop/e2
echo "--- create (-c) an archive file (-f)"
tar -cf Documents/myarchive.tar Desktop/e1 Desktop/e2
echo "--- list its content (-t)"
tar -tf Documents/myarchive.tar
echo "--- extract (-x) verbosely (-v) into another folder (-C)"
tar -xvf Documents/myarchive.tar -C extracted
echo "--- archive + compression: xz (J) and gzip (z)"
tar -cJf Documents/archive.tar.xz Desktop/e1 Desktop/e2
tar -czf Documents/archive.tar.gz Desktop/e1 Desktop/e2
ls -l Documents
````

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.11 Text Processing — <code>cat</code> · <code>sed</code> · <code>awk</code></h3>
<div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">cat</div><div style="font-size:26px;color:#333;line-height:1.5;">Prints one or more text files one after another. <code>-n</code> adds <b>line numbers</b>.</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">sed</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>S</b>tream <b>ed</b>itor — works line by line: search, replace, delete, insert. The file is <b>unchanged</b> unless you use <code>-i</code>.</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">awk</div><div style="font-size:26px;color:#333;line-height:1.5;">A text-processing language: splits each line into <b>fields</b> (&#36;1, &#36;2 …) to extract, filter and report.</div></div>
</div>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:24px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sed 's/old&#95;text/new&#95;text/g' filename</div>
<div style="white-space:pre-wrap;color:#8FA3C7;"># s = substitute · g = global (every match on each line)</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sed 's/error/warning/g' log.txt</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> awk '{print &#36;1, &#36;3}' filename</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> awk -F: '{print &#36;1, &#36;7}' /etc/passwd     # -F: sets the separator</div>
</div>

````python
%%bash
mkdir -p /tmp/exp3_lab && cd /tmp/exp3_lab
printf 'error: disk full\ninfo: backup done\nerror: timeout\n' > log.txt
echo "--- cat -n";                           cat -n log.txt
echo "--- sed: output changes, file does not"; sed 's/error/warning/g' log.txt
echo "--- the file is still untouched";      cat log.txt
echo "--- sed -i edits the file in place";   sed -i 's/error/warning/g' log.txt; cat log.txt
echo "--- awk: user name and shell (fields 1 and 7)"
awk -F: '{print $1, $7}' /etc/passwd | head -5
````

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.12 Disk Management — <code>fdisk</code></h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>fdisk</code> manages the <b>partition table</b>: create, delete, change the type and list partitions. Newer versions support <b>GPT</b> as well as classic <b>MBR</b>.</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:25px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo fdisk [options] device</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo fdisk -l /dev/sda      # partition table of one disk</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo fdisk -l               # all attached disks</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo fdisk /dev/sdb         # interactive mode</div>
</div>
<div style="display:grid;grid-template-columns:repeat(4,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">m</div><div style="font-size:26px;color:#333;line-height:1.5;">help</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">p</div><div style="font-size:26px;color:#333;line-height:1.5;">print table</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">n</div><div style="font-size:26px;color:#333;line-height:1.5;">new partition</div></div>
<div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">d</div><div style="font-size:26px;color:#333;line-height:1.5;">delete partition</div></div>
<div style="background:#FCE8E6;border-top:6px solid #EA4335;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#EA4335;font-size:28px;margin-bottom:8px;">t</div><div style="font-size:26px;color:#333;line-height:1.5;">change type</div></div>
<div style="background:#DDF6F6;border-top:6px solid #00A6A6;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#00A6A6;font-size:28px;margin-bottom:8px;">w</div><div style="font-size:26px;color:#333;line-height:1.5;">write &amp; exit</div></div>
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">q</div><div style="font-size:26px;color:#333;line-height:1.5;">quit <b>without</b> saving</div></div>
</div>

<p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><b>Creating a partition</b> (full walk-through in Experiment 16):</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:24px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo fdisk /dev/sdb</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">Command (m for help): n</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">Partition type: p</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">Partition number (1-4, default 1): 1</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">First sector (2048-...): 2048</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">Last sector: +5G</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">Command (m for help): w</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo partprobe /dev/sdb     # re-read the table without rebooting</div>
</div>
<div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:28px;margin-bottom:6px;">⛔ fdisk edits the disk directly</div><div style="font-size:26px;line-height:1.6;">One wrong device name can destroy data. Identify the disk with <code>lsblk</code> first and practise only on a <b>disposable VM disk</b>. On a GPT disk the “primary/extended” question does not appear.</div></div>

````python
%%bash
echo "--- block devices (safe, read-only)"
lsblk 2>&1 | head -12
echo "--- partition tables (needs sudo - may be denied here)"
sudo -n fdisk -l 2>&1 | head -6
````

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.13 <code>mount</code> and <code>umount</code></h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>mount</code> <b>attaches</b> a filesystem to a directory of the existing tree so its files become reachable — on a physical device or a virtual (<i>loopback</i>) one. Other books say “connect” or “mount”.</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:24px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo mount [options] device mount&#95;point</div>
<div style="white-space:pre-wrap;color:#8FA3C7;"># 1) a USB drive (device name found earlier with fdisk / lsblk)</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo mkdir /media/usb</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo mount /dev/sdb1 /media/usb</div>
<div style="white-space:pre-wrap;color:#8FA3C7;"># 2) look inside an ISO without writing anything to disk</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo mkdir -p /mnt/iso</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo mount -o loop /path/to/ubuntu.iso /mnt/iso</div>
</div>
<p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><code>umount</code> <b>detaches</b> it. All changed data in memory is <b>flushed</b> to disk first, which prevents corruption before you remove a drive or power off:</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:24px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo umount /media/usb        # by mount point</div>
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> sudo umount /dev/sdb1         # or by device</div>
</div>
<p style="font-size:25px;line-height:1.65;color:#222;margin:10px 0;">Desktop systems usually mount automatically, but you must still <b>umount / eject</b> manually before pulling a removable drive.</p>

````python
%%bash
echo "--- a few currently mounted filesystems"
mount | head -6
echo "--- the filesystem that holds /"
findmnt / 2>&1
````

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.14 Copying Raw Data — <code>dd</code></h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>dd</code> copies and converts data at the <b>byte / block</b> level. It works directly with raw blocks and devices, so it is used for low-level jobs: raw files, disk backups, disk simulation.</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:25px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> dd if=/dev/zero of=file.img bs=1M count=100</div>
</div>
<div style="display:grid;grid-template-columns:repeat(4,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">if=/dev/zero</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>Input</b> file. <code>/dev/zero</code> yields endless zero bytes (0x00).</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">of=file.img</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>Output</b> file; created if it does not exist.</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">bs=1M</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>Block size</b>: read / write 1 MB at a time. Larger blocks copy faster.</div></div>
<div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">count=100</div><div style="font-size:26px;color:#333;line-height:1.5;">Number of blocks to copy → 100 × 1 MB = <b>100 MB</b>.</div></div>
</div>
<div style="text-align:center;margin:14px 0;"><div style="font-size:26px;color:#666;font-style:italic;margin-top:8px;">Figure 3-2 — Creating an empty file for testing or teaching</div></div>
<p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">Output: <b>100+0 records in / out</b> (blocks read and written), then bytes copied, time and speed.</p>

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">More <code>dd</code> uses</h3>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#EA4335;color:white;padding:14px 16px;text-align:left;font-size:24px;width:34%;">Task</th><th style="background:#EA4335;color:white;padding:14px 16px;text-align:left;font-size:24px;width:66%;">Command</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:23px;font-weight:700;">Back up a disk</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:23px;"><code>dd if=/dev/sda of=backup.img bs=4M</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:23px;font-weight:700;">Write an ISO to a USB stick</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:23px;"><code>dd if=image.iso of=/dev/sdb bs=4M status=progress</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:23px;font-weight:700;">Copy text, converting case</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#FCE8E6;color:#222;font-size:23px;"><code>dd if=input.txt of=output.txt conv=ucase</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:23px;font-weight:700;">File with random data</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:23px;"><code>dd if=/dev/urandom of=random.img bs=1M count=10</code></td></tr>
<tr><td style="padding:11px 16px;background:#FCE8E6;color:#222;font-size:23px;font-weight:700;">Recover a damaged partition</td><td style="padding:11px 16px;background:#FCE8E6;color:#222;font-size:23px;"><code>dd if=/dev/sda1 of=/dev/sdb1 bs=4M conv=noerror,sync</code></td></tr>
</tbody>
</table>
<div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:28px;margin-bottom:6px;">⛔ One typo erases a disk</div><div style="font-size:26px;line-height:1.6;"><code>dd</code> writes straight to devices: putting <code>of=/dev/sda</code> by mistake overwrites the <b>entire disk</b>. Verify with <code>lsblk</code> and double-check <code>if</code> / <code>of</code> before running.</div></div>

````python
%%bash
mkdir -p /tmp/exp3_lab && cd /tmp/exp3_lab
echo "--- 10 blocks of 1 MiB from /dev/zero (the manual makes 100 MB)"
dd if=/dev/zero of=file.img bs=1M count=10 2>&1
ls -lh file.img
echo "--- conv=ucase converts text to upper case"
echo "hello linux" | dd conv=ucase 2>/dev/null
echo "--- random content"
dd if=/dev/urandom of=random.img bs=1M count=2 2>&1 | tail -1
ls -lh random.img
````

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.16 Processes — <code>kill</code> · <code>killall</code> · <code>ps</code></h3>
<div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">kill</div><div style="font-size:26px;color:#333;line-height:1.5;">Send a <b>signal</b> to a process by <b>PID</b> (process ID): <code>kill [options] PID</code>. Find PIDs with <code>ps</code> or <code>top</code>.</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">killall</div><div style="font-size:26px;color:#333;line-height:1.5;">Terminate processes by <b>name</b>: <code>killall [options] name</code>.</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">ps</div><div style="font-size:26px;color:#333;line-height:1.5;">A <b>snapshot</b> of the processes at the moment you run it — not interactive.</div></div>
</div>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#4C8BF5;color:white;padding:14px 16px;text-align:left;font-size:25px;width:38%;">Signal</th><th style="background:#4C8BF5;color:white;padding:14px 16px;text-align:left;font-size:25px;width:62%;">Meaning</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E8F0FE;color:#222;font-size:24px;font-weight:700;"><code>kill PID</code> (15 / TERM)</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E8F0FE;color:#222;font-size:24px;">Polite request to finish — the process can clean up</td></tr>
<tr><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:24px;font-weight:700;"><code>kill -9 PID</code> / <code>killall -9 name</code></td><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:24px;"><b>Force</b> termination (signal 9) even if running — last resort</td></tr>
</tbody>
</table>

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;"><code>ps</code> options</h3>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#4C8BF5;color:white;padding:14px 16px;text-align:left;font-size:25px;width:30%;">ps option</th><th style="background:#4C8BF5;color:white;padding:14px 16px;text-align:left;font-size:25px;width:70%;">Meaning</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E8F0FE;color:#222;font-size:24px;font-weight:700;"><code>-e</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E8F0FE;color:#222;font-size:24px;">All processes</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;"><code>-f</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;">Full format (more details)</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E8F0FE;color:#222;font-size:24px;font-weight:700;"><code>-p PID</code></td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E8F0FE;color:#222;font-size:24px;">Only the given process ID</td></tr>
<tr><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:24px;font-weight:700;"><code>--ppid PPID</code></td><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:24px;">Only children of the given parent ID — e.g. <code>ps -f --ppid 2</code></td></tr>
</tbody>
</table>

````python
%%bash
echo "--- all processes, full format (first lines)"
ps -ef | head -5
echo "--- children of PID 2 (kernel threads)"
ps -f --ppid 2 2>&1 | head -5
echo "--- start a harmless background process, find it, stop it"
sleep 300 &
pid=$!; disown $pid
ps -f -p $pid
kill $pid
sleep 0.5
ps -p $pid >/dev/null && echo "still running" || echo "process $pid terminated"
````

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;"><code>top</code> — live process monitor</h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>top</code> lists all running processes in real time, refreshed regularly.</p>
<div style="text-align:center;margin:14px 0;"><div style="font-size:26px;color:#666;font-style:italic;margin-top:8px;">Figure 3-3 — Output of top</div></div>

<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">

<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:22px;font-weight:700;">PID / USER</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:22px;">Process ID · owner of the process</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:22px;font-weight:700;">PR</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:22px;">Scheduling priority chosen by the kernel. <tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:22px;font-weight:700;">VIRT / RES / SHR</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:22px;">Virtual memory (swap + shared) · physical RAM · memory shared with other processes</td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:22px;font-weight:700;">S</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#F3E8FF;color:#222;font-size:22px;">State: <b>R</b> running · <b>S</b> sleeping · <b>Z</b> zombie · <b>T</b> stopped (<b>I</b> idle kernel thread)</td></tr>

<tr><td style="padding:11px 16px;background:#F3E8FF;color:#222;font-size:22px;font-weight:700;">TIME+ / COMMAND</td><td style="padding:11px 16px;background:#F3E8FF;color:#222;font-size:22px;">Total CPU time so far · name of the command</td></tr>
</tbody>
</table>
<div style="background:#F3E8FF;border-left:8px solid #9E4CF5;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#9E4CF5;font-size:27px;margin-bottom:6px;">/proc — a window into the kernel</div><div style="font-size:25px;line-height:1.6;"><code>/proc</code> is a <b>virtual filesystem</b> the kernel builds in memory. Tools like <code>top</code> read live data from files such as <code>/proc/interrupts</code> (hardware interrupts) and <code>/proc/stat</code> (CPU, process and interrupt statistics).</div></div>

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;"><code>htop</code> — <code>top</code> with a friendlier interface</h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>htop</code> is an interactive, real-time monitor that shows CPU, memory and more. It runs <b>inside the terminal</b> (no separate GUI). Install it with <code>sudo apt-get install htop</code>. Use <kbd>Page Up</kbd> / <kbd>Page Down</kbd> to scroll and <kbd>q</kbd> to quit.</p>
<div style="text-align:center;margin:14px 0;"><div style="font-size:26px;color:#666;font-style:italic;margin-top:8px;">Figure 3-4 — Output of htop</div></div>
<div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">🟩 Green</div><div style="font-size:26px;color:#333;line-height:1.5;">Memory used by processes</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">🟦 Blue</div><div style="font-size:26px;color:#333;line-height:1.5;">Buffers</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">🟨 Yellow</div><div style="font-size:26px;color:#333;line-height:1.5;">Cache</div></div>
</div>
<p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">Top right: running tasks, <b>load average</b> (three numbers = last <b>1, 5 and 15</b> minutes) and uptime. Click a column header to <b>sort</b> by it.</p>

````python
%%bash
echo "--- load average (1, 5, 15 min) + runnable/total tasks"
cat /proc/loadavg
echo "--- top in batch mode (one snapshot)"
top -b -n 1 | head -12
echo "--- /proc: live kernel data"
grep -E 'MemTotal|MemAvailable|SwapTotal' /proc/meminfo
head -n 1 /proc/stat | cut -c1-70
head -n 3 /proc/interrupts | cut -c1-72
````

````python
import os
from IPython.display import display, HTML
try:
    l1, l5, l15 = open("/proc/loadavg").read().split()[:3]
    mem = {}
    for line in open("/proc/meminfo"):
        key, val = line.split(":"); mem[key] = int(val.split()[0])
    total, avail = mem["MemTotal"], mem["MemAvailable"]
    pct = (total - avail) * 100 / total
    cpus = os.cpu_count() or 1
    def tone(x): return "#34A853" if float(x) < cpus * 0.7 else ("#FF7A00" if float(x) < cpus else "#EA4335")
    cards = "".join(f'<div style="background:rgba(255,255,255,.15);padding:10px 16px;border-radius:10px;text-align:center;"><div style="font-size:13px;opacity:.85;">load {n}</div>'
                    f'<div style="font-size:26px;font-weight:800;color:{tone(v)};background:white;border-radius:8px;padding:2px 10px;margin-top:4px;">{v}</div></div>'
                    for n, v in (("1 min", l1), ("5 min", l5), ("15 min", l15)))
    display(HTML('<div style="background:linear-gradient(120deg,#1B2A49,#4C8BF5);color:white;padding:16px 20px;border-radius:14px;">'
                 f'<div style="font-weight:800;font-size:18px;margin-bottom:10px;">📈 Live snapshot from /proc &nbsp; ({cpus} CPU cores)</div>'
                 f'<div style="display:flex;gap:12px;">{cards}</div>'
                 f'<div style="margin-top:12px;font-size:14px;">RAM used: <b>{pct:.0f}%</b></div>'
                 f'<div style="background:rgba(255,255,255,.25);border-radius:8px;height:14px;"><div style="width:{pct:.0f}%;background:#34A853;height:14px;border-radius:8px;"></div></div></div>'))
except Exception as e:
    display(HTML(f"<b>/proc is not available here:</b> {e}"))
````

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.17 Network, Open Files and Boot Services</h3>
<div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">netstat</div><div style="font-size:26px;color:#333;line-height:1.5;">Shows network <b>packet statistics and open ports</b> — handy for monitoring and troubleshooting. Modern replacement: <code>ss -tuln</code>.</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">lsof</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>L</b>i<b>s</b>t <b>o</b>pen <b>f</b>iles: files, sockets, pipes, devices and the processes using them. When <code>umount</code> fails, <code>lsof /media/usb</code> shows what keeps the disk busy.</div></div>
<div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">chkconfig</div><div style="font-size:26px;color:#333;line-height:1.5;">Legacy (Red Hat / SysV) tool that lists services started at boot per <b>runlevel 0–6</b>. On systemd use <code>systemctl list-unit-files</code>.</div></div>
</div>
<div style="text-align:center;margin:14px 0;"><div style="font-size:26px;color:#666;font-style:italic;margin-top:8px;">Figure 3-5 — Output of netstat -a | more</div></div>

<p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Example of the legacy <code>chkconfig --list</code> output (a number = runlevel):</p>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:24px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> chkconfig --list</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">cpuspeed  0:off  1:on   2:on   3:on   4:on   5:on   6:off</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">vsftpd    0:off  1:off  2:off  3:off  4:off  5:off  6:off</div>
<div style="white-space:pre-wrap;color:#B9C6DD;">network   0:off  1:off  2:on   3:on   4:on   5:on   6:off</div>
</div>

````python
%%bash
echo "--- listening sockets with ss (modern netstat)"
ss -tuln 2>&1 | head -8
echo "--- netstat -a | head (package net-tools)"
command -v netstat >/dev/null && netstat -a | head -8 || echo "netstat not installed: sudo apt install net-tools"
echo "--- lsof: open files of this shell"
command -v lsof >/dev/null && lsof -p $$ 2>/dev/null | head -6 || echo "lsof not installed: sudo apt install lsof"
echo "--- modern chkconfig: services enabled at boot"
systemctl list-unit-files --type=service --state=enabled --no-pager 2>&1 | head -8
````

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.18 Help and Documentation</h3>
<div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;">
<div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">man</div><div style="font-size:26px;color:#333;line-height:1.5;">Shows the <b>manual</b> of a command. <code>man ls</code>. Scroll with <kbd>u</kbd> / <kbd>d</kbd>, quit with <kbd>q</kbd>.</div></div>
<div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">-h / --help</div><div style="font-size:26px;color:#333;line-height:1.5;">Defined for almost every command: prints a short built-in help. <code>ls --help</code></div></div>
</div>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:12px 0;font-family:monospace;font-size:25px;line-height:1.75;color:#E6EDF7;">
<div style="white-space:pre-wrap;"><span style="color:#7EE787;">&#36;</span> man [options] command-or-application-name</div>

</div>

````python
%%bash
echo "--- ls --help (first lines)"
ls --help | head -6
echo "--- one-line description"
whatis ls 2>&1 | head -2
echo "--- manual page as plain text (lines 1-12)"
man grep 2>&1 | col -b | sed -n '1,12p' 
````

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.19 Software Repositories and <code>apt</code></h3>
<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><b>Repositories</b> (repos) are software sources from which packages are downloaded, installed and updated. Each distribution has its own, usually several. The main addresses are listed in <code>/etc/apt/sources.list</code> (recent Ubuntu releases keep them in <code>/etc/apt/sources.list.d/</code>).</p>
<p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><code>apt-get</code> / <code>apt</code> is the package manager of <b>Debian-based</b> systems (Debian, Mint, Ubuntu) — the counterpart of <code>yum</code> (now <code>dnf</code>) on Red Hat–based systems. It is <b>multi-command</b>: first the tool, then the action.</p>
<table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);">
<thead><tr><th style="background:#34A853;color:white;padding:14px 16px;text-align:left;font-size:25px;width:44%;">Task</th><th style="background:#34A853;color:white;padding:14px 16px;text-align:left;font-size:25px;width:56%;">Command</th></tr></thead>
<tbody>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:24px;font-weight:700;">Install a package</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:24px;"><code>sudo apt-get install unrar</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;font-weight:700;">Same with the shorter tool</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:24px;"><code>sudo apt install fish</code></td></tr>
<tr><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:24px;font-weight:700;">Remove (keeps configuration files)</td><td style="padding:11px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:24px;"><code>sudo apt-get remove unrar</code></td></tr>
<tr><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:24px;font-weight:700;">Remove <b>and</b> delete configuration</td><td style="padding:11px 16px;background:#ffffff;color:#222;font-size:24px;"><code>sudo apt-get remove --purge unrar</code></td></tr>
</tbody>
</table>
<p style="font-size:25px;line-height:1.65;color:#222;margin:10px 0;">The prompt shows a normal user, so <code>sudo</code> comes first. To install any other package just change its name.</p>

````python
%%bash
echo "--- repository definitions"
cat /etc/apt/sources.list 2>&1 | head -5
ls /etc/apt/sources.list.d/ 2>&1
echo "--- is a package installed / available? (read-only)"
apt-cache policy htop 2>&1 | head -4
dpkg -s bash 2>&1 | head -3
````

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">☠️ Command safety: do not run blindly</h3>
<div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:29px;margin-bottom:6px;">⛔ These commands can modify or destroy data</div><div style="font-size:27px;line-height:1.6;">Use them <b>only</b> on an explicitly identified lab device or a disposable VM, <b>with instructor approval</b>. Shown as text, not runnable cells.</div></div>
<div style="background:#1B2A49;border-radius:12px;padding:16px 22px;margin:10px 0;font-family:monospace;font-size:24px;line-height:1.9;color:#FCA5A5;"><div>rm -r ...                       # no recycle bin</div><div>sudo chmod -R ...               # permissions on a whole tree</div><div>sudo userdel -r ...             # deletes the home directory too</div><div>sudo dd if=... of=/dev/...</div><div>sudo fdisk /dev/...             # edits the partition table</div><div>sed -i ...                      # edits the file in place</div><div>kill -9 ... / killall -9 ...    # no chance to clean up</div><div>sudo reboot  /  sudo halt</div></div>
<p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">Before any of these: <b>check the path or device name, double-check it, and confirm a backup exists.</b></p>

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">🧠 Self-check quiz — Experiment 3</h3>
<p style="font-size:24px;line-height:1.65;color:#222;margin:10px 0;">Click a question to reveal the answer.</p>
<details style="background:#E8F0FE;border:2px solid #4C8BF5;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#4C8BF5;font-size:27px;">❓ Why does typing <code>LS</code> fail although <code>ls</code> works?</summary><div style="margin-top:8px;color:#222;font-size:25px;line-height:1.55;">✅ Linux is <b>case-sensitive</b>: <code>LS</code> and <code>ls</code> are different names.</div></details>
<details style="background:#E6F4EA;border:2px solid #34A853;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#34A853;font-size:27px;">❓ What does <code>ls -lth</code> do?</summary><div style="margin-top:8px;color:#222;font-size:25px;line-height:1.55;">✅ Three single-letter options share one dash: <b>l</b>ong format, sorted by modification <b>t</b>ime, <b>h</b>uman-readable sizes.</div></details>
<details style="background:#F3E8FF;border:2px solid #9E4CF5;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#9E4CF5;font-size:27px;">❓ A directory has <code>rwxr-x--x</code> (751). Which mode keeps the owner’s rights, gives the group read only and others nothing?</summary><div style="margin-top:8px;color:#222;font-size:25px;line-height:1.55;">✅ <b>740</b> = 7 (rwx) + 4 (r--) + 0 (---), i.e. <code>rwxr-----</code>: <code>chmod 740 dir</code>.</div></details>
<details style="background:#FFEDD9;border:2px solid #FF7A00;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#FF7A00;font-size:27px;">❓ What is the difference between <code>su</code> and <code>sudo</code>?</summary><div style="margin-top:8px;color:#222;font-size:25px;line-height:1.55;">✅ <code>su</code> <b>switches</b> to another user account; <code>sudo</code> runs <b>one command</b> with administrator rights without switching.</div></details>
<details style="background:#FCE8E6;border:2px solid #EA4335;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#EA4335;font-size:27px;">❓ <code>rmdir</code> refuses to delete a directory. What now?</summary><div style="margin-top:8px;color:#222;font-size:25px;line-height:1.55;">✅ It only removes <b>empty</b> directories. Delete the contents first or use <code>rm -r</code> — carefully, there is no undo.</div></details>
<details style="background:#DDF6F6;border:2px solid #00A6A6;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#00A6A6;font-size:27px;">❓ Does <code>sed 's/error/warning/g' log.txt</code> change <code>log.txt</code>?</summary><div style="margin-top:8px;color:#222;font-size:25px;line-height:1.55;">✅ No — it prints the modified text. Add <code>-i</code> to edit the file <b>in place</b>.</div></details>
<details style="background:#E8F0FE;border:2px solid #4C8BF5;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#4C8BF5;font-size:27px;">❓ Why is <code>dd</code> dangerous?</summary><div style="margin-top:8px;color:#222;font-size:25px;line-height:1.55;">✅ It writes directly to devices; a wrong <code>of=/dev/sda</code> overwrites the <b>whole disk</b>. Check with <code>lsblk</code> first.</div></details>
<details style="background:#E6F4EA;border:2px solid #34A853;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#34A853;font-size:27px;">❓ What are the three numbers of the load average?</summary><div style="margin-top:8px;color:#222;font-size:25px;line-height:1.55;">✅ The average load over the last <b>1, 5 and 15 minutes</b>.</div></details>

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 6px;">🧠 Self-check quiz</h3><p style="font-size:14px;color:#555;">Click a question to reveal the answer.</p><details style="background:#F3E8FF;border:2px solid #9E4CF5;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#9E4CF5;">❓ Is “Linux” the whole operating system?</summary><div style="margin-top:8px;color:#222;font-size:15px;">✅ No. <b>Linux is the kernel</b>. A complete system adds GNU tools, libraries, a shell, desktop, package management and apps — a <b>distribution</b> bundles them.</div></details><details style="background:#E6F4EA;border:2px solid #34A853;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#34A853;">❓ Which freedoms require access to source code?</summary><div style="margin-top:8px;color:#222;font-size:15px;">✅ <b>Freedom 1</b> (study/modify) and <b>Freedom 3</b> (distribute modified versions).</div></details><details style="background:#E8F0FE;border:2px solid #4C8BF5;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#4C8BF5;">❓ Which installation mode is the safest for practising?</summary><div style="margin-top:8px;color:#222;font-size:15px;">✅ A <b>virtual machine</b> — it is isolated from the physical disk and supports snapshots.</div></details><details style="background:#FFEDD9;border:2px solid #FF7A00;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#FF7A00;">❓ Does <code>apt update</code> upgrade installed software?</summary><div style="margin-top:8px;color:#222;font-size:15px;">✅ No. It only <b>refreshes package metadata</b>. <code>apt upgrade</code> does the upgrading.</div></details><details style="background:#DDF6F6;border:2px solid #00A6A6;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#00A6A6;">❓ First ping to 1.1.1.1 works, second to example.com fails — what is wrong?</summary><div style="margin-top:8px;color:#222;font-size:15px;">✅ <b>DNS resolution</b>. IP connectivity is fine; check <code>resolvectl status</code>.</div></details><details style="background:#FCE8E6;border:2px solid #EA4335;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#EA4335;">❓ What is the modern equivalent of runlevel 5?</summary><div style="margin-top:8px;color:#222;font-size:15px;">✅ <code>graphical.target</code>.</div></details><details style="background:#FCE8E6;border:2px solid #EA4335;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#EA4335;">❓ Name two commands that can destroy data if misused.</summary><div style="margin-top:8px;color:#222;font-size:15px;">✅ e.g. <code>dd</code>, <code>mkfs</code>, <code>wipefs</code>, <code>fdisk</code> (write mode), <code>rm -rf</code>.</div></details><details style="background:#F3E8FF;border:2px solid #9E4CF5;border-radius:10px;padding:10px 14px;margin:8px 0;"><summary style="cursor:pointer;font-weight:700;color:#9E4CF5;">❓ Is GRUB the kernel?</summary><div style="margin-top:8px;color:#222;font-size:15px;">✅ No. GRUB is a <b>boot loader</b>; it loads or hands control to the kernel.</div></details>
