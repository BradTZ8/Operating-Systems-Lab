<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.1 What Is a Shell Script?</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">A <b>shell script</b> is a program that is <b>interpreted and executed by the shell</b>. The shell receives scripts and command-line commands and runs them. A script can change the filesystem, start programs, print messages and much more.</p><div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">📜</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">One command per line</div><div style="font-size:26px;color:#333;line-height:1.5;">Every line is a standard shell or Linux command — the same commands you can already type in the terminal.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🔁</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">Run once, repeat forever</div><div style="font-size:26px;color:#333;line-height:1.5;">Executing the file runs all its commands automatically; you do not retype them next time.</div></div><div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">⚡</div><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">No compilation</div><div style="font-size:26px;color:#333;line-height:1.5;">Scripts are <b>interpreted</b> directly by the shell — there is no compile step.</div></div></div>

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.1.1 Why Learn Shell Scripting?</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Even an administrator who never writes scripts needs to <b>read</b> them to understand system behaviour. When Linux starts, the system scripts in <code>/etc/rc.d</code> run; they contain the configuration and the services of the machine. To analyse the system, an administrator must know how these <b>startup scripts</b> work and, when necessary, modify them.</p><div style="background:#E6F4EA;border-left:8px solid #34A853;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#34A853;font-size:30px;margin-bottom:6px;">Typical uses</div><div style="font-size:28px;line-height:1.6;">automating <b>daily tasks</b> · producing <b>error reports</b> · <b>monitoring</b> the system · <b>synchronisation</b> of files and data.</div></div><div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:30px;margin-bottom:6px;">When NOT to use shell scripting</div><div style="font-size:28px;line-height:1.6;">Shell scripts are a poor choice for <b>heavy programs</b> (lots of computation or mathematics), for <b>complex</b> applications and for <b>graphical</b> programs.</div></div>

file</code></td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:26px;">send the output of <code>command</code> to <code>file</code></td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:26px;font-weight:700;"><code>command &lt; file</code></td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:26px;">read the input of <code>command</code> from <code>file</code></td></tr><tr><td style="padding:10px 16px;background:#E6F4EA;color:#222;font-size:26px;font-weight:700;"><code>command1 | command2</code></td><td style="padding:10px 16px;background:#E6F4EA;color:#222;font-size:26px;">the output of <code>command1</code> becomes the input of <code>command2</code></td></tr></tbody></table>

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">A) The &gt; operator</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Here <code>ls</code> lists the current folder, but <b>nothing appears</b> in the terminal (Figure 4-3) because the output was written to <code>lsresults.txt</code>.</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">ls &gt; lsresults.txt</span></div></div><div style="text-align:center;margin:16px 0;"><img src="images/Redirect_Output.jpg" alt="Figure 4-3 — Using the &gt; operator and the resulting file" style="width:80%;max-width:100%;border-radius:14px;border:3px solid #4C8BF5;box-shadow:0 6px 18px rgba(0,0,0,.18);"><div style="font-size:24px;color:#666;font-style:italic;margin-top:8px;">Figure 4-3 — Using the &gt; operator and the resulting file</div></div><div style="background:#FFEDD9;border-left:8px solid #FF7A00;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#FF7A00;font-size:30px;margin-bottom:6px;">Why is lsresults.txt in its own listing?</div><div style="font-size:28px;line-height:1.6;">The shell <b>creates</b> the file first, then runs <code>ls</code>, which therefore sees the new file and writes its name too.</div></div>

```python
%%bash
mkdir -p /tmp/exp4_lab && cd /tmp/exp4_lab
ls > lsresults.txt
echo "--- nothing was printed above; the output is inside the file:"
cat lsresults.txt
```

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">B) The &lt; operator</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">The file <code>lsresults.txt</code> is given to <code>wc</code> as input; with <b>-w</b> it counts the words. A second example sorts the same file alphabetically:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">wc</span><span style="color:#E6EDF7;"> -w &lt; lsresults.txt</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">sort</span><span style="color:#E6EDF7;"> &lt; lsresults.txt</span></div></div><div style="background:#FFEDD9;border-left:8px solid #FF7A00;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#FF7A00;font-size:30px;margin-bottom:6px;">Remember</div><div style="font-size:28px;line-height:1.6;">The <b>input</b> comes from the file, but the <b>result</b> is still displayed in the terminal.</div></div>

```python
%%bash
cd /tmp/exp4_lab
printf 'pear\napple\nfig\nbanana\n' > fruit.txt
echo "--- wc -w < fruit.txt"; wc -w < fruit.txt
echo "--- sort < fruit.txt";  sort < fruit.txt
```

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">C) The pipe |</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>ls -l /bin/</code> prints a very long list. Adding <code>more</code> through a pipe shows it <b>page by page</b>: <b>Enter</b> moves one line, <b>Space</b> one page.</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">ls -l /bin/ | more</span></div></div><div style="text-align:center;margin:16px 0;"><img src="images/Pipe_More.jpg" alt="Figure 4-4 — The pipe operator" style="width:72%;max-width:100%;border-radius:14px;border:3px solid #9E4CF5;box-shadow:0 6px 18px rgba(0,0,0,.18);"><div style="font-size:24px;color:#666;font-style:italic;margin-top:8px;">Figure 4-4 — The pipe operator</div></div><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Pipes also let you filter. The left command (<code>ls</code>) feeds the right command (<code>grep</code>); every line containing <b>server</b> is shown. It is as if you saved the <code>ls</code> output in a text file and then ran <code>grep</code> on that file. <b>Only the output of the second command</b> reaches the terminal.</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">ls -l /bin | </span><span style="color:#FF9E64;">grep</span><span style="color:#E6EDF7;"> server --color</span></div></div>

<div style="background:#DDF6F6;border-left:8px solid #00A6A6;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#00A6A6;font-size:30px;margin-bottom:6px;">Practice</div><div style="font-size:28px;line-height:1.6;">Run <code>find /bin -name "&#42;server&#42;"</code> and compare its result with the pipe example above.</div></div><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>head</code> prints only the first lines. Without a number it shows the <b>first 10</b> lines. Here the sorted file is passed on and only <b>5</b> lines are displayed:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">sort</span><span style="color:#E6EDF7;"> lsresults.txt | </span><span style="color:#FF9E64;">head</span><span style="color:#E6EDF7;"> -n 5</span></div></div>

```python
%%bash
cd /tmp/exp4_lab
echo "--- ls -l /bin | grep server   (lines containing 'server')"
ls -l /bin | grep server || echo "(no program with 'server' in its name on this machine)"
echo "--- find /bin -name '*server*'"
find /bin -name '*server*' 2>/dev/null | head -5
echo "--- sort lsresults.txt | head -n 5"
sort lsresults.txt | head -n 5
```

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">Combining redirection operators</h3><div style="background:#E6F4EA;border-left:8px solid #34A853;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#34A853;font-size:30px;margin-bottom:6px;">Example 1 — two redirections at once</div><div style="font-size:28px;line-height:1.6;"><code>ps | sort &gt; pssort.txt</code> — the output of <code>ps</code> is sorted by <code>sort</code> and the result is stored in <code>pssort.txt</code>.</div></div><div style="background:#E8F0FE;border-left:8px solid #4C8BF5;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#4C8BF5;font-size:30px;margin-bottom:6px;">Example 2 — a chain of pipes</div><div style="font-size:28px;line-height:1.6;">Lines 5 to 10 (if they exist) of the sorted search results for the word <i>information</i> in the <code>.conf</code> files of <code>/etc</code>:</div></div><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">grep</span><span style="color:#E6EDF7;"> information /etc/&#42;.conf | </span><span style="color:#FF9E64;">sort</span><span style="color:#E6EDF7;"> | </span><span style="color:#FF9E64;">head</span><span style="color:#E6EDF7;"> -n 10 | </span><span style="color:#FF9E64;">tail</span><span style="color:#E6EDF7;"> -n 5</span></div></div>

```python
%%bash
cd /tmp/exp4_lab
echo "--- ps | sort > pssort.txt"
ps | sort > pssort.txt; head -3 pssort.txt
echo "--- grep information /etc/*.conf | sort | head -n 10 | tail -n 5"
grep -s information /etc/*.conf | sort | head -n 10 | tail -n 5 || true
```

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">Several commands on one line — the semicolon</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">To send several separate commands to the interpreter at once, end each group with <b>;</b> :</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">sort</span><span style="color:#E6EDF7;"> lsresults.txt &gt; temp; </span><span style="color:#FF9E64;">head</span><span style="color:#E6EDF7;"> -n 5 &lt; temp; rm temp</span></div></div><div style="margin:10px 0;"><div style="display:flex;align-items:flex-start;margin:10px 0;"><div style="flex:0 0 44px;color:#FF7A00;font-size:36px;font-weight:800;line-height:1.5;">1.</div><div style="background:#FFEDD9;padding:12px 16px;border-radius:10px;flex:1;font-size:26px;line-height:1.5;color:#222;">Sorts <code>lsresults.txt</code> and puts the result in a second file called <code>temp</code>.</div></div><div style="display:flex;align-items:flex-start;margin:10px 0;"><div style="flex:0 0 44px;color:#FF7A00;font-size:36px;font-weight:800;line-height:1.5;">2.</div><div style="background:#FFFFFF;border:1px solid #DCE3EF;padding:12px 16px;border-radius:10px;flex:1;font-size:26px;line-height:1.5;color:#222;">Gives <code>temp</code> to <code>head</code> as input, so only its first 5 lines are shown.</div></div><div style="display:flex;align-items:flex-start;margin:10px 0;"><div style="flex:0 0 44px;color:#FF7A00;font-size:36px;font-weight:800;line-height:1.5;">3.</div><div style="background:#FFEDD9;padding:12px 16px;border-radius:10px;flex:1;font-size:26px;line-height:1.5;color:#222;">Deletes <code>temp</code>.</div></div></div><div style="background:#FFEDD9;border-left:8px solid #FF7A00;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#FF7A00;font-size:30px;margin-bottom:6px;">Important</div><div style="font-size:28px;line-height:1.6;">Commands separated by <code>;</code> work <b>independently</b>: a command ignores the input and output of the previous one unless you pass data through a temporary file (as above) or another method. <b>Try it:</b> rewrite this example using only redirection operators.</div></div>

<h3 style="color:#00A6A6;border-left:8px solid #00A6A6;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.3 Shells</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Commands reach the operating system through the terminal, but behind the terminal sits an <b>interpreter</b>. The command-line interpreter is called a <b>shell</b> because it sits like a shell between the operating system and the terminal. Several shells exist for Linux:</p><div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🏛️</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">Bourne Shell (sh)</div><div style="font-size:26px;color:#333;line-height:1.5;">The oldest Linux shell, written by <b>Stephen Bourne</b> at Bell Labs. It lacks interactive features and rich programming structures, but over the years it became the <b>standard</b>.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🐚</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">BASH</div><div style="font-size:26px;color:#333;line-height:1.5;"><b>B</b>ourne <b>A</b>gain <b>Sh</b>ell — an extended Bourne shell and the <b>default</b> on most distributions. It comes from the GNU project, is open source and resembles the Korn shell.</div></div><div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🌊</div><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">C shell (csh)</div><div style="font-size:26px;color:#333;line-height:1.5;">Created by <b>Bill Joy</b> in Berkeley UNIX; after Bash and Korn it is the third most used shell.</div></div><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🍎</div><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">zsh</div><div style="font-size:26px;color:#333;line-height:1.5;">An extended Bash with extra features borrowed from other languages; it became the default shell on <b>macOS</b> (which used Bash before).</div></div></div>

```python
%%bash
echo "login shell of this user : $SHELL"
echo "shell running this cell  : $(ps -p $$ -o comm=)"
echo "--- shells installed on the system (/etc/shells)"
cat /etc/shells 2>/dev/null || echo "(no /etc/shells here)"
```

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.4 Your First Script</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">The workflow is: <b>write</b> the script in an editor and save it, then do what is needed to <b>interpret and run</b> it. Open your favourite editor and type:</p><div style="display:inline-block;background:#4C8BF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px 10px 0 0;margin-top:10px;">📄 hello.sh</div><div style="background:#1B2A49;border-radius:0 12px 12px 12px;padding:14px 22px;margin:0 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#BB9AF7;">#!/bin/bash</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"Hello World!"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">exit</span><span style="color:#E6EDF7;"> 0</span></div></div><div style="text-align:center;margin:16px 0;"><img src="images/Hello_Script_gedit.jpg" alt="Figure 4-5 — Creating the first script file, hello.sh" style="width:40%;max-width:100%;border-radius:14px;border:3px solid #4C8BF5;box-shadow:0 6px 18px rgba(0,0,0,.18);"><div style="font-size:24px;color:#666;font-style:italic;margin-top:8px;">Figure 4-5 — Creating the first script file, hello.sh</div></div>

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">What do the three lines mean?</h3><div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;"><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🎯</div><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">Line 1 — #!/bin/bash</div><div style="font-size:26px;color:#333;line-height:1.5;">The <b>#</b> (sharp) and <b>!</b> (bang) form the <b>shebang</b>. Remove the line and the system default shell is used.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">💬</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">Line 2 — echo</div><div style="font-size:26px;color:#333;line-height:1.5;"><code>echo</code> prints a message: put the text in quotes after it. This script prints a <b>single</b> message.</div></div><div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🏁</div><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">Line 3 — exit 0</div><div style="font-size:26px;color:#333;line-height:1.5;"><code>exit</code> ends the script. The value <b>0</b> tells the system that the script finished <b>successfully</b>.</div></div></div><div style="background:#E8F0FE;border-left:8px solid #4C8BF5;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#4C8BF5;font-size:30px;margin-bottom:6px;">File extensions</div><div style="font-size:28px;line-height:1.6;">Save the file with any name and the <code>.sh</code> extension. <h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.4.1 Adding Comments</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Start a line with <b>#</b> to add a comment. The interpreter ignores it:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#BB9AF7;">#!/bin/sh</span></div><div style="white-space:pre-wrap;"><span style="color:#7F9BC9;"># this line is a comment and is hidden from the interpreter</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"Hello World!"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">exit</span><span style="color:#E6EDF7;"> 0</span></div></div>

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.4.2 Running the Script</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">A C++ program is compiled into an executable. A shell script is <b>interpreted</b>, so it only needs the <b>execute permission</b> to become runnable. Windows executables mostly end in <code>.exe</code>; Linux pays little attention to extensions.</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">chmod</span><span style="color:#E6EDF7;"> a+x hello.sh</span></div></div><div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;"><div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">a+x</div><div style="font-size:26px;color:#333;line-height:1.5;">Execute permission for <b>all</b> users.</div></div><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">u+x</div><div style="font-size:26px;color:#333;line-height:1.5;">Execute permission for the <b>owner</b> only — use it when only the owner should run the script.</div></div></div><p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">Read and write permission for others is up to you (not recommended for security), but the <b>execute</b> permission (value 1) must be granted to the user who runs the script.</p>

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">Two ways to start a script</h3><div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">📍</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">A) In the current directory</div><div style="font-size:26px;color:#333;line-height:1.5;">Write <code>./</code> and the file name. <code>./</code> tells the interpreter to look <b>here</b> and not in the default system directories stored in the <code>PATH</code> variable.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🗺️</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">B) Somewhere else</div><div style="font-size:26px;color:#333;line-height:1.5;">Give the <b>full path</b> and file name, for example when the script is in the root folder: <code>/root/hello.sh</code>.</div></div></div><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">./hello.sh</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">/root/hello.sh</span></div></div><div style="background:#F3E8FF;border-left:8px solid #9E4CF5;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#9E4CF5;font-size:30px;margin-bottom:6px;">Choosing the interpreter yourself</div><div style="font-size:28px;line-height:1.6;">If the shell is not set at the top of the script, or you want another interpreter, put the interpreter name first: <code>sh script_name.sh</code> or <code>bash script_name.sh</code>.</div></div>

```python
%%bash
mkdir -p /tmp/exp4_lab && cd /tmp/exp4_lab
cat > hello.sh <<'EOF'
#!/bin/bash
# my first script
echo "Hello World!"
exit 0
EOF
echo "--- before chmod:";  ls -l hello.sh
chmod a+x hello.sh
echo "--- after  chmod:";  ls -l hello.sh
echo "--- ./hello.sh";      ./hello.sh
echo "--- bash hello.sh";   bash hello.sh
echo "exit status of the last run: $?"
```

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.5 Variables</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">In Bash you <b>do not declare</b> variables. <h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.5.1 Printing Variables</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;"><code>echo</code> prints messages and variables. <h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.5.2 Reading Input — read</h3><p <h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">4.5.3 Arithmetic on Variables</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Bash has several ways to calculate. The first one is <b>let</b>:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">let</span><span style="color:#E6EDF7;"> x=x+1</span></div></div><div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:30px;margin-bottom:6px;">Notes</div><div style="font-size:28px;line-height:1.6;">No spaces before or after <b>+</b> or <b>=</b> with <code>let</code>. <h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">Arithmetic operators in Bash</h3><table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);"><thead><tr><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:26px;width:18%;">Symbol</th><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:26px;width:27%;">Operator</th><th style="background:#34A853;color:white;padding:12px 16px;text-align:left;font-size:26px;width:55%;">Example</th></tr></thead><tbody><tr><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;font-weight:700;"><code>+</code></td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;">addition</td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;"><code>((x = 5 + 2))</code> → 7</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;font-weight:700;"><code>-</code></td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;">subtraction</td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;"><code>((x = 5 - 2))</code> → 3</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;font-weight:700;"><code>&#42;</code></td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;">multiplication</td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;"><code>((x = 5 &#42; 2))</code> → 10</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;font-weight:700;"><code>/</code></td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;">division</td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;"><code>((x = 7 / 2))</code> → 3 (integer)</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;font-weight:700;"><code>%</code></td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;">remainder</td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;"><code>((x = 7 % 2))</code> → 1</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;font-weight:700;"><code>&#42;&#42;</code></td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;">power</td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#ffffff;color:#222;font-size:25px;"><code>((x = 2 &#42;&#42; 5))</code> → 32</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;font-weight:700;"><code>++</code></td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;">increment</td><td style="padding:10px 16px;border-bottom:1px solid #e3e8f0;background:#E6F4EA;color:#222;font-size:25px;"><code>((x++))</code> adds 1</td></tr><tr><td style="padding:10px 16px;background:#ffffff;color:#222;font-size:25px;font-weight:700;"><code>--</code></td><td style="padding:10px 16px;background:#ffffff;color:#222;font-size:25px;">decrement</td><td style="padding:10px 16px;background:#ffffff;color:#222;font-size:25px;"><code>((x--))</code> subtracts 1</td></tr></tbody></table>

```python
%%bash
x=5
let x=x+1;      echo "let x=x+1    -> $x"
((x = x * 2));  echo "((x = x * 2)) -> $x"
((x++));        echo "((x++))       -> $x"
((x--)); ((x--)); echo "two ((x--))    -> $x"
((p = 2 ** 10)); echo "2 ** 10       -> $p"
((r = 17 % 5));  echo "17 % 5        -> $r"
((q = 17 / 5));  echo "17 / 5        -> $q   (integer division)"
```

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">Example — pass or fail?</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Write a script that prints <b>pass</b> or <b>fail</b> depending on the average grade:</p><div style="display:inline-block;background:#4C8BF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px 10px 0 0;margin-top:10px;">📄 pass_fail.sh</div><div style="background:#1B2A49;border-radius:0 12px 12px 12px;padding:14px 22px;margin:0 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"please enter the average:"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">read</span><span style="color:#E6EDF7;"> ave</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">if</span><span style="color:#E6EDF7;"> [ &#36;ave -ge 10 ]</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">then</span></div><div style="white-space:pre-wrap;"><span bracket and the conditional expression: <code>[ &#36;ave -ge 10 ]</code>. ```python
%%bash
mkdir -p /tmp/exp5_lab && cd /tmp/exp5_lab
cat > pass_fail.sh <<'EOF'
#!/bin/bash
echo "please enter the average:"
read ave
if [ $ave -ge 10 ]
then
    echo "pass"
else
    echo "fail"
fi
exit 0
EOF
for grade in 14 10 7; do echo "--- average = $grade"; echo $grade | bash pass_fail.sh; done
echo "--- missing space before ]  (error on purpose)"
bash -c 'ave=12; if [ $ave -ge 10] ; then echo pass; fi' 2>&1 || true
```

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">Example — positive, negative or zero?</h3><div style="display:inline-block;background:#4C8BF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px 10px 0 0;margin-top:10px;">📄 sign.sh</div><div style="background:#1B2A49;border-radius:0 12px 12px 12px;padding:14px 22px;margin:0 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#BB9AF7;">#!/bin/bash</span></div><div style="white-space:pre-wrap;"><span style="color:#7F9BC9;"># positive or negative?</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"please enter a number:"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">read</span><span style="color:#E6EDF7;"> x</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">if</span><span style="color:#E6EDF7;"> [ &#36;x -gt 0 ]</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">then</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"&#36;x is positive!"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">elif</span><span style="color:#E6EDF7;"> [ &#36;x -lt 0 ]</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">then</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"&#36;x is negative!"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">else</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span ```python
%%bash
cd /tmp/exp5_lab
cat > sign.sh <<'EOF'
#!/bin/bash
echo "please enter a number:"
read x
if [ $x -gt 0 ]
then
    echo "$x is positive!"
elif [ $x -lt 0 ]
then
    echo "$x is negative!"
else
    echo "It's a zero"
fi
exit 0
EOF
for n in 8 -3 0; do echo "--- input $n"; echo $n | bash sign.sh | tail -1; done
```

[ &#36;y -lt 0 ]</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">then</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"the product of &#36;x and &#36;y is positive!"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">elif</span><span style="color:#E6EDF7;"> [ &#36;x -eq 0 ] || [ &#36;y -eq 0 ]</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">then</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"the product of &#36;x and &#36;y is zero!"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">elif</span><span style="color:#E6EDF7;"> [ &#36;x -gt 0 ] &amp;&amp; [ &#36;y -gt 0 ]</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">then</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"the product of &#36;x and &#36;y is positive!"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">else</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"the product of &#36;x and &#36;y is negative!"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">fi</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">exit</span><span style="color:#E6EDF7;"> 0</span></div></div>

```python
%%bash
cd /tmp/exp5_lab
cat > product_sign.sh <<'EOF'
#!/bin/bash
echo "insert two integers:"
read x y
if [ $x -lt 0 ] && [ $y -lt 0 ]
then
    echo "the product of $x and $y is positive!"
elif [ $x -eq 0 ] || [ $y -eq 0 ]
then
    echo "the product of $x and $y is zero!"
elif [ $x -gt 0 ] && [ $y -gt 0 ]
then
    echo "the product of $x and $y is positive!"
else
    echo "the product of $x and $y is negative!"
fi
exit 0
EOF
for pair in "-2 -3" "0 5" "4 6" "-2 5"; do echo "$pair" | bash product_sign.sh | tail -1; done
```

The cure: put the variable <b>inside quotes</b> in the test — <code>[ "&#36;ans" = "y" ]</code>.</div></div>

```python
%%bash
cd /tmp/exp5_lab
cat > ask_bad.sh <<'EOF'
#!/bin/bash
echo "do u want to continue? (y/n)"
read ans
if [ $ans = "y" ]
then
    echo "let's go"
elif [ $ans = "n" ]
then
    echo "see you later"
else
    echo "please type y/n!"
fi
EOF
sed 's/\[ \$ans/[ "$ans"/' ask_bad.sh > ask_good.sh
echo "--- unquoted \$ans, empty answer (error)"; echo "" | bash ask_bad.sh 2>&1 | tail -2
echo "--- quoted \"\$ans\", empty answer";      echo "" | bash ask_good.sh 2>&1 | tail -1
echo "--- quoted, answer y";                      echo "y" | bash ask_good.sh 2>&1 | tail -1
```

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">5.4 The for Loop</h3><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">for</span><span style="color:#E6EDF7;"> variable </span><span style="color:#FF9E64;">in</span><span style="color:#E6EDF7;"> values</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">do</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    commands</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">done</span></div></div><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><b>for</b> repeats the loop once for <b>each value</b> listed after <code>in</code>. <h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">More iterations — two handy forms</h3><div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🔢</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">1) Range with step</div><div style="font-size:26px;color:#333;line-height:1.5;">Give <b>start</b>, <b>end</b> and <b>step</b> in braces separated by <code>..</code>: <code>{start..stop..step}</code>. Brace ranges work from Bash 3.0; the <i>step</i> part needs Bash 4.0 or newer.</div></div><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🧮</div><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">2) C-style loop</div><div style="font-size:26px;color:#333;line-height:1.5;">Start value, end condition and step go between <b>double parentheses</b>, as in the C language. Here we go from 5 to 15 in steps of 3.</div></div></div><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">for</span><span style="color:#E6EDF7;"> i </span><span style="color:#FF9E64;">in</span><span style="color:#E6EDF7;"> {0..10..2}</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">do</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"Even Numbers: &#36;i"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">done</span></div><div style="white-space:pre-wrap;">&nbsp;</div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">for</span><span style="color:#E6EDF7;"> ((i=5; i&lt;=15; i+=3))</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">do</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"&#36;i"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">done</span></div></div><p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">Both forms can also count <b>downwards</b>.</p>

```python
%%bash
echo "--- for x in 1 2 3 4 5";                for x in 1 2 3 4 5; do echo -n "$x "; done; echo
echo "--- for i in {0..10..2}";               for i in {0..10..2}; do echo "Even Numbers: $i"; done
echo "--- for ((i=5; i<=15; i+=3))";         for ((i=5; i<=15; i+=3)); do echo -n "$i "; done; echo
echo "--- counting down: {5..1}";            for i in {5..1}; do echo -n "$i "; done; echo
echo "--- counting down: C style";           for ((i=10; i>=0; i-=5)); do echo -n "$i "; done; echo
```

<h3 style="color:#00A6A6;border-left:8px solid #00A6A6;padding-left:12px;margin:14px 0 10px;font-size:32px;">5.5 The while Loop</h3><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">while</span><span style="color:#E6EDF7;"> [ condition ]</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">do</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    commands</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">done</span></div></div><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><b>Example:</b> read a password and compare it with the correct one (here <b>secret</b>); keep repeating until the right password is typed.</p><div style="display:inline-block;background:#4C8BF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px 10px 0 0;margin-top:10px;">📄 password.sh</div><div style="background:#1B2A49;border-radius:0 12px 12px 12px;padding:14px 22px;margin:0 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#BB9AF7;">#!/bin/bash</span></div><div style="white-space:pre-wrap;"><span style="color:#7F9BC9;"># repeat until the password is correct</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"Enter password:"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">read</span><span style="color:#E6EDF7;"> pass</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">while</span><span style="color:#E6EDF7;"> [ </span><span style="color:#F5D76E;">"&#36;pass"</span><span style="color:#E6EDF7;"> != </span><span style="color:#F5D76E;">"secret"</span><span style="color:#E6EDF7;"> ]</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">do</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"Sorry, try again"</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">read</span><span style="color:#E6EDF7;"> pass</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">done</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">exit</span><span style="color:#E6EDF7;"> 0</span></div></div>

```python
%%bash
cd /tmp/exp5_lab
cat > password.sh <<'EOF'
#!/bin/bash
echo "Enter password:"
read pass
while [ "$pass" != "secret" ]
do
    echo "Sorry, try again"
    read pass
done
echo "access granted"
exit 0
EOF
printf 'abc\n123\nsecret\n' | bash password.sh
```

style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"let's go!"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">elif</span><span style="color:#E6EDF7;"> [ </span><span style="color:#F5D76E;">"&#36;ans"</span><span style="color:#E6EDF7;"> = </span><span style="color:#F5D76E;">"n"</span><span style="color:#E6EDF7;"> ] || [ </span><span style="color:#F5D76E;">"&#36;ans"</span><span style="color:#E6EDF7;"> = </span><span style="color:#F5D76E;">"N"</span><span style="color:#E6EDF7;"> ]</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">then</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    </span><span style="color:#FF9E64;">echo</span><span style="color:#E6EDF7;"> </span><span style="color:#F5D76E;">"see you later!"</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">fi</span></div><div style="white-space:pre-wrap;"><span style="color:#FF9E64;">exit</span><span style="color:#E6EDF7;"> 0</span></div></div><div style="background:#FFEDD9;border-left:8px solid #FF7A00;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#FF7A00;font-size:30px;margin-bottom:6px;">Note</div><div style="font-size:28px;line-height:1.6;">The <code>-p</code> switch of <code>read</code> (print a prompt) is <b>not defined</b> in <code>sh</code>.</div></div>

```python
%%bash
cd /tmp/exp5_lab
cat > ask_loop.sh <<'EOF'
#!/bin/bash
while [ "$ans" != "y" ] && [ "$ans" != "Y" ] && [ "$ans" != "n" ] && [ "$ans" != "N" ]
do
    read -p "do u want to continue?(y/n) " ans
done
if [ "$ans" = "y" ] || [ "$ans" = "Y" ]
then
    echo "let's go!"
elif [ "$ans" = "n" ] || [ "$ans" = "N" ]
then
    echo "see you later!"
fi
exit 0
EOF
echo "(inputs: x, maybe, Y -- the first two are rejected)"
printf 'x\nmaybe\nY\n' | bash ask_loop.sh
```

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">5.6.1 Defining and Changing Environment Variables</h3><div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;"><div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🏠</div><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">Local variable</div><div style="font-size:26px;color:#333;line-height:1.5;">Visible only in the <b>current shell</b>. The manual creates one called <code>MY_VAR</code> with the value <code>Hello</code>.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">📤</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">export — inherited by sub-processes</div><div style="font-size:26px;color:#333;line-height:1.5;">To define or change a variable that <b>every child process</b> can see, use <code>export</code>.</div></div></div><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">&#36; </span><span style="color:#FF9E64;">export</span><span style="color:#E6EDF7;"> MY&#95;VAR2=</span><span style="color:#F5D76E;">"World"</span></div></div><div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:30px;margin-bottom:6px;">⚠️ Careful with “set”</div><div style="font-size:28px;line-height:1.6;">The manual writes <code>set MY_VAR="Hello"</code> (the C-shell way). In <b>Bash</b>, <code>set</code> with arguments does <b>not</b> create a variable — it sets the <i>positional parameters</i>. In Bash, simply write <code>MY_VAR="Hello"</code>.</div></div>

```python
%%bash
echo "--- the C-shell way inside Bash: 'set MY_VAR=Hello' sets \$1, not a variable"
bash -c 'set MY_VAR="Hello"; echo "MY_VAR=[$MY_VAR]   first positional parameter \$1=[$1]"'
echo "--- the Bash way: plain assignment (local)"
MY_VAR="Hello"
echo "this shell sees MY_VAR=[$MY_VAR]"
bash -c 'echo "a child process sees MY_VAR=[$MY_VAR]   (empty: it was not exported)"'
echo "--- export makes it visible to child processes"
export MY_VAR2="World"
bash -c 'echo "a child process sees MY_VAR2=[$MY_VAR2]"'
```

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">Extending PATH</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Suppose you keep personal scripts in a folder inside your home directory. <h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">6.1 What Is an Interrupt?</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Sometimes an event happens <b>outside or inside the processor</b>. The processor temporarily stops the normal execution of the current process, <b>responds to the event</b>, and then returns to the rest of the process. Such an event is called an <b>interrupt</b>.</p><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">At the end of <b>every instruction cycle</b> the processor checks whether an interrupt is pending. If so, it handles it immediately and afterwards goes back to what it was doing.</p><div style="text-align:center;margin:16px 0;"><img src="images/Interrupt_Cycle.jpg" alt="Figure 6-1 — The instruction cycle and the interrupt check" style="width:70%;max-width:100%;border-radius:14px;border:3px solid #EA4335;box-shadow:0 6px 18px rgba(0,0,0,.18);"><div style="font-size:24px;color:#666;font-style:italic;margin-top:8px;">Figure 6-1 — The instruction cycle and the interrupt check</div></div><p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">To respond, the processor runs the <b>interrupt handler</b>, also called the <b>Interrupt Service Routine (ISR)</b>.</p>

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">6.1.1 Types of Interrupts</h3><div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;"><div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🔌</div><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">Hardware interrupts</div><div style="font-size:26px;color:#333;line-height:1.5;">Hardware origin. Sent to the processor as an <b>asynchronous</b> signal.</div></div><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">💻</div><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">Software interrupts (traps)</div><div style="font-size:26px;color:#333;line-height:1.5;">Generated by a <b>running program</b>. They are <b>synchronous</b>.</div></div></div><h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:28px;">What happens when a hardware interrupt arrives?</h3><div style="margin:10px 0;"><div style="display:flex;align-items:flex-start;margin:10px 0;"><div style="flex:0 0 processor receives an <b>interrupt request (IRQ)</b> and confirms it by sending an <b>acknowledge (INTA)</b>.</div></div><div style="display:flex;align-items:flex-start;margin:10px 0;"><div style="flex:0 0 44px;color:#FF7A00;font-size:36px;font-weight:800;line-height:1.5;">2.</div><div style="background:#FFFFFF;border:1px solid #DCE3EF;padding:12px 16px;border-radius:10px;flex:1;font-size:26px;line-height:1.5;color:#222;">The <b>interrupt number</b> is sent to the processor, which identifies the hardware that raised the request.</div></div><div style="display:flex;align-items:flex-start;margin:10px 0;"><div style="flex:0 0 44px;color:#FF7A00;font-size:36px;font-weight:800;line-height:1.5;">3.</div><div style="background:#FFEDD9;padding:12px 16px;border-radius:10px;flex:1;font-size:26px;line-height:1.5;color:#222;">The processor <b>saves the state</b> of the current process and switches to <b>kernel mode</b>.</div></div><div style="display:flex;align-items:flex-start;margin:10px 0;"><div style="flex:0 0 44px;color:#FF7A00;font-size:36px;font-weight:800;line-height:1.5;">4.</div><div style="background:#FFFFFF;border:1px it.</div></div><div style="display:flex;align-items:flex-start;margin:10px 0;"><div style="flex:0 0 44px;color:#FF7A00;font-size:36px;font-weight:800;line-height:1.5;">5.</div><div style="background:#FFEDD9;padding:12px 16px;border-radius:10px;flex:1;font-size:26px;line-height:1.5;color:#222;">Finally it returns to <b>user mode</b> and continues the process.</div></div></div>

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">Examples of hardware interrupts</h3><div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">⏲️ Timer interrupt</div><div style="font-size:26px;color:#333;line-height:1.5;">One of the most basic ones, generated by the system timer. The kernel uses it to manage <b>process scheduling</b> and so prevents one process from monopolising the CPU.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">⌨️ I/O interrupt</div><div style="font-size:26px;color:#333;line-height:1.5;">For example keyboard or printer. Raised when an input/output operation is completed or requested. The processor can do other work instead of waiting for very slow I/O; controllers move data to buffers and then notify the CPU.</div></div><div style="background:#FCE8E6;border-top:6px solid #EA4335;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#EA4335;font-size:28px;margin-bottom:8px;">🛠️ Machine check</div><div style="font-size:26px;color:#333;line-height:1.5;">Error or hardware fault interrupt.</div></div><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">🔁 Restart</div><div style="font-size:26px;color:#333;line-height:1.5;">The restart interrupt.</div></div></div>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">Examples of software interrupts</h3><div style="display:grid;grid-template-columns:repeat(1,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">📞 System call</div><div style="font-size:26px;color:#333;line-height:1.5;">When a user-level program needs a <b>kernel service</b> it raises a software interrupt with a system call. The CPU switches from user mode to kernel mode. This keeps the system secure: user programs cannot touch the kernel directly. Example: saving or opening a file requires disk access, so the process asks the kernel, the CPU serves the request in kernel mode and returns to user mode.</div></div><div style="background:#FFEDD9;border-top:6px solid #FF7A00;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#FF7A00;font-size:28px;margin-bottom:8px;">❌ Program error (exception)</div><div style="font-size:26px;color:#333;line-height:1.5;">Raised when the current process tries an <b>illegal operation</b>, such as an overflow or a division by zero.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">📨 Signal</div><div style="font-size:26px;color:#333;line-height:1.5;">A software interrupt sent by a process, the OS or the user to one or more processes. Signals are used for <b>exception handling</b>, <b>inter-process communication</b> and <b>process control</b>.</div></div></div>

<h3 style="color:#00A6A6;border-left:8px solid #00A6A6;padding-left:12px;margin:14px 0 10px;font-size:32px;">Blocking and non-blocking system calls</h3><p style="font-size:27px;line-height:1.65;color:#222;margin:10px 0;">Think of reading a file from disk. It is an I/O (disk) request started by a <b>system call</b> of the requesting process, and other disk requests may already be queued. To avoid wasting CPU time, the requester is <b>blocked</b>. When the disk access is done, the disk controller announces it with an <b>I/O interrupt</b>. If the process needs the answer to continue it must stay blocked; but some of its tasks do not depend on the I/O answer and could run while it waits. Hence two kinds of system call:</p><div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;"><div style="background:#FCE8E6;border-top:6px solid #EA4335;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#EA4335;font-size:28px;margin-bottom:8px;">⛔ Blocking system call</div><div style="font-size:26px;color:#333;line-height:1.5;">The requesting process becomes <b>blocked</b> and the CPU switches to the next process. When the request completes, a hardware <b>I/O interrupt</b> takes the process out of the blocked state.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">✅ Non-blocking system call</div><div style="font-size:26px;color:#333;line-height:1.5;">Control returns to the requesting process immediately, so it can <b>carry on with its other tasks</b>.</div></div></div>

Naturally a large number.</td></tr></tbody></table><div style="background:#E8F0FE;border-left:8px solid #4C8BF5;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#4C8BF5;font-size:30px;margin-bottom:6px;">Note</div><div style="font-size:26px;line-height:1.6;">Apart from the timer interrupt, the ID numbers of other interrupts may differ from system to system.</div></div><p style="font-size:26px;line-height:1.65;color:#222;margin:10px 0;">To watch the local timer of every processor live:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">watch -n 1 </span><span style="color:#F5D76E;">'cat /proc/interrupts | grep LOC'</span></div></div>

```python
import os
from IPython.display import display, HTML
try:
    rows = []
    for line in open("/proc/interrupts").read().splitlines()[1:]:
        parts = line.split()
        if not parts or not parts[0].endswith(":"):
            continue
        counts = [int(p) for p in parts[1:] if p.isdigit()]
        name = parts[-1] if len(parts) > len(counts) + 1 else "?"
        rows.append((parts[0].rstrip(":"), sum(counts), name))
    rows = sorted(rows, key=lambda r: -r[1])[:8]
    top = max([r[1] for r in rows] + [1])
    cols = ["#4C8BF5", "#34A853", "#FF7A00", "#9E4CF5", "#EA4335", "#00A6A6"]
    bars = ""
    for i, (irq, total, name) in enumerate(rows):
        w = max(2, int(100 * total / top))
        bars += (f'<div style="display:flex;align-items:center;margin:6px 0;font-family:monospace;font-size:20px;">'
                 f'<div style="width:230px;">{irq:>4} {name[:22]}</div>'
                 f'<div style="background:{cols[i % 6]};height:24px;border-radius:6px;width:{w}%;"></div>'
                 f'<div style="margin-left:10px;">{total:,}</div></div>')
    display(HTML('<div style="background:#F7F9FC;border:2px solid #FF7A00;border-radius:12px;padding:14px 18px;">'
                 '<div style="font-weight:800;color:#FF7A00;font-size:24px;margin-bottom:8px;">Busiest interrupts on this machine (all CPUs added up)</div>'
                 + bars + '</div>'))
except Exception as e:
    print("Could not read /proc/interrupts here:", e)
```

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">What each part of the code does</h3><div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🆔</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">1 · getpid()</div><div style="font-size:26px;color:#333;line-height:1.5;">A system call that returns the <b>process ID</b>. <h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">Compile, run and trace</h3><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">gcc sys&#95;call.c -o sys&#95;call</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">./sys&#95;call</span></div></div><p style="font-size:27px;line-height:1.65;color:#222;margin:10px 0;">You can also use <b>strace</b>, a command-line tool that shows the system calls and signals a program makes:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#7F9BC9;"># trace the system calls of a process</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">strace ./sys&#95;call</span></div><div style="white-space:pre-wrap;"><span style="color:#7F9BC9;"># call statistics</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">strace -c ./sys&#95;call</span></div></div>

```python
%%bash
mkdir -p /tmp/exp6_lab && cd /tmp/exp6_lab
cat > sys_call.c <<'EOF'
#include <stdio.h>
#include <unistd.h>
#include <sys/syscall.h>
#include <sys/time.h>
#include <sys/resource.h>
#include <sys/sysinfo.h>
#include <time.h>

void trace_system_calls() {
    printf("\nSystem Call Tracing Example\n");
    printf("getpid() system call:\n");
    pid_t pid = getpid();
    printf("  PID: %d\n", pid);
    printf("getrusage() system call:\n");
    struct rusage usage;
    getrusage(RUSAGE_SELF, &usage);
    printf("  User time: %ld.%06ld\n", usage.ru_utime.tv_sec, usage.ru_utime.tv_usec);
    printf("  System time: %ld.%06ld\n", usage.ru_stime.tv_sec, usage.ru_stime.tv_usec);
    printf("sysinfo() system call:\n");
    struct sysinfo info;
    if (sysinfo(&info) == 0) {
        printf("  Free memory: %lu KB\n", info.freeram / 1024);
        printf("  Number of processes: %d\n", info.procs);
    } else {
        perror("sysinfo failed");
    }
}

void mode_switching() {
    printf("User/Kernel Mode Switching Example\n");
    printf("First; Running in user mode & Performing some simple mathematical calculations \n");
    int result = 0;
    for (int i = 0; i < 1000; i++) {
        result += i * i;
    }
    printf("Calculation result: %d\n", result);
    printf("\nSecond; Switching to kernel mode by executing nanosleep() which is a system call\n");
    struct timespec ts;
    ts.tv_sec = 0;
    ts.tv_nsec = 100000000; // 100ms
    if (nanosleep(&ts, NULL) == 0) {
        printf("Third; Returned to user mode\n");
    } else {
        perror("nanosleep failed\n");
    }
}

int main() {
    trace_system_calls();
    mode_switching();
    return 0;
}
EOF
if ! command -v gcc >/dev/null 2>&1; then echo "gcc is not installed here -> sudo apt install gcc"; exit 0; fi
gcc sys_call.c -o sys_call && ./sys_call
echo
if command -v strace >/dev/null 2>&1; then
  echo "--- strace -c ./sys_call  (statistics, first lines)"
  strace -c ./sys_call 2>&1 >/dev/null | head -12
else
  echo "(strace is not installed here:  sudo apt install strace)"
fi
```

a process: <b>SigPnd</b> / <b>ShdPnd</b> (pending), <b>SigBlk</b> (blocked), <b>SigIgn</b> (ignored) and <b>SigCgt</b> (caught = has a handler).</div></div>

```python
%%bash
mkdir -p /tmp/exp6_lab && cd /tmp/exp6_lab
cat > signal_test.c <<'EOF'
#include <stdio.h>
#include <stdlib.h>
#include <unistd.h>
#include <signal.h>
#include <string.h>

void signal_handler(int sig) {
    printf("\nSignal received: %d (%s)\n", sig, strsignal(sig));
}

void print_signal_status() {
    char command[256];
    snprintf(command, sizeof(command),
             "echo 'Current Signal Status:' && cat /proc/%d/status | grep -A4 Sig",
             getpid());
    system(command);
    printf("\n");
}

int main() {
    printf("Signal Handling Example \n");
    printf("Process ID: %d\n\n", getpid());
    signal(SIGUSR1, signal_handler);
    signal(SIGUSR2, signal_handler);
    signal(SIGTERM, SIG_DFL);
    signal(SIGINT,  SIG_DFL);
    printf("Ready to receive signals.\n\n");
    while (1) {
        pause();
        print_signal_status();
    }
    return 0;
}
EOF
if ! command -v gcc >/dev/null 2>&1; then echo "gcc is not installed here -> sudo apt install gcc"; exit 0; fi
gcc signal_test.c -o signal_test || exit 0
# run it in the background (line-buffered so the log is complete), then play the role of the "other terminal"
stdbuf -oL ./signal_test > signal.log 2>&1 &
PID=$!
sleep 0.5
echo "started signal_test with PID $PID"
kill -USR1 $PID; sleep 0.3
kill -USR2 $PID; sleep 0.3
kill -TERM $PID; wait $PID 2>/dev/null
echo "--- program terminated by SIGTERM (default action).  Its log:"
grep -E "Signal received|SigCgt|Ready" signal.log
```
