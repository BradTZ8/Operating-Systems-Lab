<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">9.1 Semaphore</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">A <b>semaphore</b> is a <b>synchronisation mechanism</b> that controls access to shared resources in concurrent systems such as multitasking operating systems. Semaphores are used to avoid <b>race conditions</b> and to make sure several processes or threads do not enter <b>critical sections</b> of code at the same time.</p><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">A semaphore data structure consists of a <b>queue</b> (the processes or threads waiting on that semaphore) and a <b>counter</b> that stores the wake-up signals. When the counter is zero the process goes to <b>sleep</b>.</p><table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);"><thead><tr><th style="background:#9E4CF5;color:white;padding:12px 16px;text-align:left;font-size:23px;width:16%;">Operation</th><th style="background:#9E4CF5;color:white;padding:12px 16px;text-align:left;font-size:23px;width:24%;">Other names</th><th style="background:#9E4CF5;color:white;padding:12px 16px;text-align:left;font-size:23px;width:60%;">Effect</th></tr></thead><tbody><tr><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#F3E8FF;font-size:23px;color:#222;line-height:1.5;"><b>Wait</b></td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#F3E8FF;font-size:23px;color:#222;line-height:1.5;">Down call · P</td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#F3E8FF;font-size:23px;color:#222;line-height:1.5;">Decreases the semaphore value. If the value is already <b>zero</b>, the caller <b>blocks</b> until the value becomes greater than zero.</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;"><b>Post</b></td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;">Up call · Signal · V</td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;">Increases the semaphore value. If blocked processes exist, <b>one of them is unblocked</b>.</td></tr></tbody></table>

<p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">So <code>wait</code> tries to <b>decrease</b> the counter while <code>post</code> tries to <b>increase</b> it. What makes semaphores special is that a thread that waits on a zero semaphore <b>blocks</b>. Later, when another thread posts, the blocked thread wakes up and the value <b>stays at zero</b>: the post increased it, and the awakened <code>wait</code> immediately decreased it again.</p><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">If several threads are blocked, the system picks one; exactly how is system-dependent. Teaching examples usually assume <b>FIFO</b> order. The order in which threads were blocked is normally not a concern — if it is, the program may not be well designed.</p>

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">9.2 Types of Semaphores</h3><div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">0️⃣1️⃣</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">Binary semaphore</div><div style="font-size:26px;color:#333;line-height:1.5;">Also called a <b>mutex</b>; its value can be only <b>0 or 1</b>. Normally guards <b>a single resource</b>.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">➕</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">Non-negative counting semaphore</div><div style="font-size:26px;color:#333;line-height:1.5;">Any integer from <b>0 to +∞</b>. Manages access to a <b>pool of resources</b> of several instances.</div></div><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">➖</div><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">Integer counting semaphore</div><div style="font-size:26px;color:#333;line-height:1.5;">May also be <b>negative</b>; a negative value means the <b>number of threads waiting</b> in the semaphore queue.</div></div></div><div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:30px;margin-bottom:6px;">Clarification</div><div style="font-size:28px;line-height:1.6;">The POSIX semaphores of this experiment (<code>sem_t</code>) are <b>non-negative</b>: <code>sem_wait</code> blocks at 0 and the value never goes below 0. The “integer counting semaphore” whose negative value counts waiters is a <b>textbook model</b> (some implementations report waiters that way); it is not what <code>sem_t</code> does on Linux.</div></div>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:30px;">Using a semaphore as a mutex</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">A semaphore with initial value <b>1</b> can be used like a mutex. When a thread wants to enter its <b>critical section</b> and access a shared data structure, it performs <b>wait</b>. If no other thread is in its critical section the value is still 1, so the thread does not wait, enters immediately and the value becomes 0. Any other thread that waits meanwhile <b>blocks</b>. When the first thread leaves its critical section it performs <b>post</b>: one waiting thread is unblocked; if nobody waits, the semaphore returns to its initial value 1. It is called a <b>binary semaphore</b> because it has exactly two states.</p><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Since semaphores are integers they can take values greater than one, so they can also <b>count scarce resources</b>. For example, a thread may wait on a semaphore to consume one item of a resource; when no item is left the semaphore is zero and <code>wait</code> blocks. When another thread adds an item it calls <code>post</code>, which either increases the number of items or lets a blocked thread take the item. This is the <b>non-negative counting semaphore</b>.</p>

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:30px;">POSIX semaphores vs. System V IPC</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">The POSIX semaphore API is <b>not really part of the ordinary pthread API</b>; POSIX standardises semaphores under a different API. Traditional UNIX systems supported <b>shared memory, message queues and semaphores</b> as <b>System V IPC</b>. POSIX also provides shared memory, message queues and semaphores as a package that competes with or replaces the older standard. The functionality is similar; the details of the two APIs differ.</p><div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🔀</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">Between processes too</div><div style="font-size:26px;color:#333;line-height:1.5;">Like System V semaphores, POSIX semaphores can synchronise <b>two or more separate processes</b>.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🏷️</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">Named semaphores</div><div style="font-size:26px;color:#333;line-height:1.5;">Because they can be used between processes, they can be <b>given a name</b>: one process creates a semaphore under a name and others open it by that name.</div></div><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🔒</div><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">Unlike pthread mutexes</div><div style="font-size:26px;color:#333;line-height:1.5;">In the manual's words a mutex can be used only by threads of the same process.</div></div></div><div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:30px;margin-bottom:6px;">Clarification</div><div style="font-size:28px;line-height:1.6;">A default <code>pthread_mutex_t</code> is indeed meant for threads of one process, but POSIX allows a mutex in shared memory to be marked <code>PTHREAD_PROCESS_SHARED</code>. Likewise, <code>sem_init(&amp;s, 0, v)</code> with <b>pshared = 0</b> (used in this experiment) shares the semaphore only among threads of one process; a nonzero pshared requires the <code>sem_t</code> to live in shared memory.</div></div>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">9.2.1 Binary Semaphore</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">The example shows the general form of a program that <b>defines</b>, <b>initialises</b>, <b>cleans up</b> and <b>uses</b> a POSIX semaphore. We use it to control the access of <b>two threads</b> to a <b>critical section</b>. The goal is to guarantee that, at any moment, <b>only one</b> thread can be in that section, preventing race conditions.</p>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:30px;">Sem1.c — binary semaphore as a mutex</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><code>mutex</code> starts at 1 (<code>sem_init(&amp;mutex, 0, 1)</code>). Thread <code>t1</code> starts, waits (value 1 → 0), enters the critical section and sleeps 4 s. After 2 s <code>t2</code> is created and blocks in <code>sem_wait</code>. When <code>t1</code> leaves and posts, <code>t2</code> enters.</p><div style="display:inline-block;background:#9E4CF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 Sem1.c</div><div style="font-size:22px;color:#555;">Archive file: <code>01_sem1_binary_semaphore_mutex.c</code> — the next cell writes exactly this source to <code>/tmp/exp9_lab/Sem1.c</code>.</div>

```python
%%writefile /tmp/exp9_lab/Sem1.c
// NOTE: the thread function below has no 'return' statement in the PDF listing.
// The listing is kept unchanged; see README.md ("Issues found in the PDF").
//Sem1.c
#include <stdio.h> // Standard input-output library
#include <pthread.h> // POSIX thread library
#include <semaphore.h> // Semaphore library
#include <unistd.h> // POSIX operating system API

// Define a binary semaphore named mutex
sem_t mutex;

// Thread function definition
void* thread(void* arg) {
    // Decrease the semaphore value; if it is 0 the thread blocks
    sem_wait(&mutex);

    // Print a message: the thread entered the critical section
    printf("\n The thread has entered the critical section \n");

    // Sleep 4 seconds to simulate the time the real work would take
    sleep(4);

    // Print a message: the thread leaves the critical section
    printf("\nExiting thread\n");

    // Increase the semaphore value so other threads may enter the critical section
    sem_post(&mutex);
}

int main() {
    // Give the mutex the initial value 1
    sem_init(&mutex, 0, 1);

    // Define two thread variables
    pthread_t t1, t2;

    // Create the first thread
    pthread_create(&t1, NULL, thread, NULL);
    sleep(2);

    // Create the second thread
    pthread_create(&t2, NULL, thread, NULL);

    // Wait for the threads to terminate
    pthread_join(t1, NULL);
    pthread_join(t2, NULL);

    // Clean up the semaphore
    sem_destroy(&mutex);
    return 0;
}
```

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">9.2.2 Counting Semaphore</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Another difference between a mutex and a counting semaphore: unlike a mutex, a counting semaphore may be <b>posted by a different thread</b> than the one that performed the wait. One thread does <code>wait</code> and another does <code>post</code> on the same semaphore. This is essential when a semaphore counts the instances of a scarce resource.</p><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Consider the classic <b>producer–consumer</b> problem: one thread <b>produces items</b> (for example <code>void *</code> objects pointing to anything), the second <b>consumes</b> them. Semaphores <b>synchronise access</b> to the shared resource, which here is a <b>buffer</b> holding the items while they pass from one thread to the other.</p>

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:30px;">Choosing the number and type of semaphores</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Semaphores guarantee that producer and consumer are <b>synchronised</b> on the buffer, so no race condition occurs. They must <b>not access the buffer at the same time</b>, otherwise reads and writes interfere: while one thread writes or reads, the buffer is locked so no other thread can change it. So <b>at least one semaphore</b> is needed to control concurrent access to the buffer.</p><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Also, neither may <b>overtake</b> the other: the producer must not create a new item when the buffer is <b>full</b> (it would overwrite existing items) and the consumer cannot take an item when the buffer is <b>empty</b>. How many semaphores are needed?</p><div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">1️⃣</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">Only one shared semaphore?</div><div style="font-size:26px;color:#333;line-height:1.5;">To make the producer wait it must be 0 (buffer full); to make the consumer wait it must be 0 (buffer empty) — a <b>contradiction</b>. And if the producer waits on 0, the consumer would wait too (same semaphore) → <b>deadlock</b>.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">2️⃣</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">At least two</div><div style="font-size:26px;color:#333;line-height:1.5;">One counts the <b>free slots</b>, the other the <b>used slots</b>. Initially the buffer is empty: <code>full = 0</code> (consumer waits at once) and <code>empty = BUFFER_SIZE</code> (the producer may produce that many items). When free slots reach zero the producer waits.</div></div><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">3️⃣</div><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">Total: three</div><div style="font-size:26px;color:#333;line-height:1.5;">Two non-negative counting semaphores (<code>empty</code>, <code>full</code>) are needed because the <b>producer must block when the buffer is full</b> and the <b>consumer when it is empty</b>; semaphores block the caller only when it tries to push the value below zero (posting never blocks). Plus one <b>binary semaphore</b> (<code>mutex</code>) for mutual exclusion.</div></div></div>

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:30px;">Producer–consumer problem (the listing has no file name)</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">The producer makes 10 items; each one needs a free slot (<code>sem_wait(&amp;empty)</code>) and the mutex, then stores the item, prints it, sleeps 1 s and posts <code>mutex</code> and <code>full</code>. The consumer does the mirror image with <code>full</code> / <code>empty</code> and removes one item per round. <code>main</code> initialises <code>empty = BUFFER_SIZE (5)</code>, <code>full = 0</code>, <code>mutex = 1</code>, creates both threads, joins them and destroys the semaphores.</p><div style="display:inline-block;background:#34A853;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 prod_cons.c</div><div style="font-size:22px;color:#555;">Archive file: <code>02_producer_consumer_counting_semaphores.c</code> — the next cell writes exactly this source to <code>/tmp/exp9_lab/prod_cons.c</code>.</div>

```python
%%writefile /tmp/exp9_lab/prod_cons.c
// NOTE: the producer/consumer functions have no 'return' statement in the PDF listing.
// The listing is kept unchanged; see README.md ("Issues found in the PDF").
// Producer-Consumer Problem
#include <stdio.h>
#include <unistd.h>
#include <pthread.h>
#include <semaphore.h>

// Buffer size
#define BUFFER_SIZE 5

// Array buffer
int buffer[BUFFER_SIZE];

// Counter of the items inside the buffer
int count = 0;

// Three semaphores for the empty state, the full state and mutual exclusion
sem_t empty, full, mutex;

// Producer thread function
void* producer(void* arg) {
    // Loop that produces 10 items
    for (int i = 0; i < 10; i++) {
        // Decrease 'empty'; if the buffer is full, the producer blocks
        sem_wait(&empty);

        // Decrease the mutex to enter the critical section
        sem_wait(&mutex);

        // Add an item to the buffer and increase the counter
        buffer[count++] = i;

        // Print the number of the produced item
        printf("Produced: %d\n", i);

        // One second delay (optional)
        sleep (1);

        // Increase the mutex to leave the critical section
        sem_post(&mutex);

        // Increase 'full': one item was produced. If the consumer was blocked, wake it up
        sem_post(&full);
    }
}

// Consumer thread function
void* consumer(void* arg) {
    for (int i = 0; i < 10; i++) {
        // Decrease 'full'; if the buffer is empty, the consumer blocks
        sem_wait(&full);

        // Decrease the mutex to enter the critical section
        sem_wait(&mutex);

        // Remove an item from the buffer and decrease the counter
        int item = buffer[--count];

        // Print the consumed item
        printf("Consumed: %d\n", item);
        sleep (1);

        // Increase the mutex to leave the critical section
        sem_post(&mutex);

        // Increase 'empty': one item was consumed, so the producer can leave its wait
        sem_post(&empty);
    }
}

int main() {
    // Two thread variables for the producer and the consumer
    pthread_t prod, cons;

    // Initialise the 'empty' semaphore with the buffer size
    sem_init(&empty, 0, BUFFER_SIZE);

    // Initialise the 'full' semaphore with zero
    sem_init(&full, 0, 0);

    // Initialise the mutex semaphore with 1
    sem_init(&mutex, 0, 1);

    // Create the producer and consumer threads
    pthread_create(&prod, NULL, producer, NULL);
    pthread_create(&cons, NULL, consumer, NULL);

    // Wait for the threads to finish
    pthread_join(prod, NULL);
    pthread_join(cons, NULL);

    // Clean up the semaphores
    sem_destroy(&empty);
    sem_destroy(&full);
    sem_destroy(&mutex);
    return 0;
}
```
