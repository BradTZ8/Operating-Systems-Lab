<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:30px;">Example 1 — thread1.c: a thread created inside the main thread</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">This is the general form of a program that creates a thread. After creating the new thread, the <b>main thread waits for its end</b>. Both threads only print a message so you can see the <b>order of execution</b>. <b>Run the code several times</b> to observe the order.</p>

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:30px;">thread1.c</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Main prints a message, creates a thread with <code>pthread_create</code>, prints a second message while the new thread runs, and calls <code>pthread_join</code> before the last message.</p><div style="display:inline-block;background:#4C8BF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 thread1.c</div><div style="font-size:22px;color:#555;">Archive file: <code>01_thread1_create_and_join.c</code> — the next cell writes exactly this source to <code>/tmp/exp8_lab/thread1.c</code>.</div>

```python
%%writefile /tmp/exp8_lab/thread1.c
// thread1.c : create a new thread inside another thread
/*run this code several times to observe the order of threads being executed */
#include <stdio.h>
#include <stdlib.h>
#include <pthread.h>

// Function to be executed by the new thread
void* thread_function(void* arg) {
    printf("Hello from new thread!\n");
    return NULL;
}

int main() {
    pthread_t thread_ID;
    printf("Hello from main thread!\n");

    // Create a new thread
    pthread_create(&thread_ID, NULL, thread_function, NULL);

    // The main program continues while the other thread executes.
    printf("Main thread is running parallel with the new thread !\n");

    // Wait for the thread to terminate.
    pthread_join(thread_ID, NULL);

    // Only the main thread is running now.
    printf("Now, only the main thread is running!\n");
    return 0;
}
```

<div style="display:inline-block;background:#4C8BF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px 10px 0 0;margin-top:10px;">⚙️ Compile and run</div><div style="background:#1B2A49;border-radius:0 12px 12px 12px;padding:14px 22px;margin:0 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">gcc thread1.c -o thread1 -pthread</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">./thread1</span></div></div><div style="background:#E8F0FE;border-left:8px solid #4C8BF5;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#4C8BF5;font-size:30px;margin-bottom:6px;">Expected behaviour</div><div style="font-size:28px;line-height:1.6;">Four lines. <i>Hello from main thread!</i> is always first and <i>Now, only the main thread is running!</i> is always last (because of <code>pthread_join</code>). The two middle lines — <i>Hello from new thread!</i> and <i>Main thread is running parallel…</i> — can appear in either order.</div></div>

```python
%%bash
cd /tmp/exp8_lab
gcc thread1.c -o thread1 -pthread
./thread1
echo "--- order of the 2nd output line over 20 runs (shows the non-determinism):"
for i in $(seq 1 20); do ./thread1 | sed -n 2p; done | sort | uniq -c
```

<h3 style="color:#4C8BF5;border-left:8px solid #4C8BF5;padding-left:12px;margin:14px 0 10px;font-size:28px;">8.2.3 Ending threads — joining the main thread</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Both threads run in parallel or pseudo-parallel. At some point we must <b>wait</b> until the new thread ends and let the main thread <b>collect its result</b> with <code>pthread_join()</code>. It has two parameters: the <b>thread ID</b> and the <b>value returned by the thread function</b> (see the next example). With several threads call <code>pthread_join</code> <b>once per thread ID</b>. After that only the main thread is running.</p><h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:28px;">8.2.4 Detached threads</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Another way is a <b>detached thread</b>: it has an independent execution state and needs <b>no synchronisation through join</b>. Its returned value is <b>discarded</b>, and when it ends it releases its system resources <b>automatically</b>, without <code>pthread_join()</code>.</p><div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">👍</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">Good for</div><div style="font-size:26px;color:#333;line-height:1.5;">Background jobs needing no supervision, independent services, tasks returning no result, event handlers in network programs.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">👎</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">Not usable when</div><div style="font-size:26px;color:#333;line-height:1.5;">You must collect results, synchronise with the thread or check how it finished.</div></div><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">⚠️</div><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">Danger</div><div style="font-size:26px;color:#333;line-height:1.5;">You never know when it finishes. When <code>main()</code> executes <code>return</code>, every running thread is <b>cut off abruptly</b>, possibly leaving important work unfinished.</div></div></div><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">A thread can be created detached by changing its <b>attributes</b>, or an existing thread can be detached with:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">pthread_detach(thread);</span></div></div><div style="background:#FFEDD9;border-left:8px solid #FF7A00;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#FF7A00;font-size:30px;margin-bottom:6px;">Note</div><div style="font-size:28px;line-height:1.6;">A detached thread can <b>no longer be joined</b>.</div></div><div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:30px;margin-bottom:6px;">Clarification</div><div style="font-size:28px;line-height:1.6;">The manual says the result cannot be returned because of a “non-shared memory model”. That is imprecise: threads <b>do share memory</b>. The real reason is that nobody <b>waits for</b> (joins) a detached thread, so nobody receives its return value or exit status.</div></div>

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:28px;">8.2.5 Cancelling — <code>pthread_cancel()</code></h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">To destroy a thread <b>before</b> its function returns normally, use <code>pthread_cancel()</code>. It is risky: you must be sure the thread has <b>released the resources</b> it acquired before it dies. Memory allocated dynamically but not freed before cancelling causes a <b>memory leak</b> and gradually slows the program or even the whole system.</p><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">This differs from killing a whole <b>process</b>: the operating system cleans up the resources of a process (the whole address space is returned). It does <b>not</b> do that for individual threads, because the threads of a process <b>share</b> resources and the system knows memory allocated by one thread may still be used by another. That is why cancelling threads correctly is difficult.</p>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">8.3 Passing an Argument to a Thread and Returning Results</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">In the next example an integer <code>number</code> is read from the user and passed as the <b>fourth argument</b> of <code>pthread_create</code>, so it reaches the thread function as its input. Because the thread function parameter is a <code>void *</code>, it must first be <b>cast to an <code>int *</code></b> and then dereferenced to get the value. (In practice a pointer to a <b>structure</b> holding several arguments can be passed too.)</p>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:30px;">thread4.c — argument in, result out</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">The thread doubles the number, stores the result in memory obtained with <code>malloc</code> and returns that pointer. <code>main</code> receives it with <code>pthread_join(thread_ID, (void**)&amp;thread_result)</code>, prints it and <b>frees</b> the memory.</p><div style="display:inline-block;background:#9E4CF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 thread4.c</div><div style="font-size:22px;color:#555;">Archive file: <code>02_thread4_argument_and_return_value.c</code> — the next cell writes exactly this source to <code>/tmp/exp8_lab/thread4.c</code>.</div>

```python
%%writefile /tmp/exp8_lab/thread4.c
//thread4.c
#include <stdio.h>
#include <stdlib.h>
#include <pthread.h>

// Function to be executed by the new thread
void* thread_function(void* arg) {
    int number = *((int*)arg); // Cast the argument
    int* result = malloc(sizeof(int)); // Allocate memory for the result
    *result = number * 2; // Example computation: double the number
    return (void*)result; // Return the result as a void pointer
}

int main() {
    pthread_t thread_ID; // Declare a thread identifier
    int number; // The integer to be passed to the thread
    int* thread_result; // Pointer to store the result from the thread

    // Get the integer from the user
    printf("Enter an integer: ");
    scanf("%d", &number);

    // Create a new thread that will run 'thread_function'
    pthread_create(&thread_ID, NULL, thread_function, &number);

    // Wait for the thread to terminate and get the result
    pthread_join(thread_ID, (void**)&thread_result);

    // Print the result
    printf("Thread returned value: %d\n", *thread_result);

    // Free the allocated memory
    free(thread_result);
    return 0;
}
```

To receive an <b>integer</b> result, <code>main</code> declares an <code>int *thread_result</code>; the address of this pointer, <code>&amp;thread_result</code>, is a pointer to a pointer. <code>void *</code> can point to <b>any</b> data type, and with <code>void **</code> a variable holding an address is converted to a “pointer to a pointer” of unspecified data type:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">pthread_join(thread_ID, (void**)&amp;thread_result);</span></div></div><div style="background:#FFEDD9;border-left:8px solid #FF7A00;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#FF7A00;font-size:30px;margin-bottom:6px;">Warning from the manual</div><div style="font-size:28px;line-height:1.6;">Thread functions return a <code>void *</code>. Converting an integer to a pointer and back works only on machines where it loses no information, and is dangerous on some machines. <h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">8.4 Several Threads with the Same Function</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Here two threads share the <b>same thread function</b>, each with a <b>different argument</b> (and possibly a different result). To create more threads, use a <b>loop</b> and keep the identifiers in an <b>array</b>.</p>

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:30px;">thread6.c — two threads, one function</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Both threads run <code>print_message</code>, which casts its <code>void *</code> argument back to <code>char *</code> and prints it. Main joins both, then prints a final line.</p><div style="display:inline-block;background:#34A853;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 thread6.c</div><div style="font-size:22px;color:#555;">Archive file: <code>03_thread6_same_function_two_threads.c</code> — the next cell writes exactly this source to <code>/tmp/exp8_lab/thread6.c</code>.</div>

```python
%%writefile /tmp/exp8_lab/thread6.c
//thread6.c
//Multi threading: create two threads, both threads use the same function.
#include <stdio.h>
#include <pthread.h>

// Function to be executed by the threads
void* print_message(void* arg) {
    char* message = (char*)arg;
    printf("%s\n", message);
    return NULL;
}

int main() {
    pthread_t thread1, thread2;

    // Messages to be printed by the threads
    char* message1 = "Hello from Thread 1!";
    char* message2 = "Hello from Thread 2!";

    // Create two threads
    pthread_create(&thread1, NULL, print_message, (void*)message1);
    pthread_create(&thread2, NULL, print_message, (void*)message2);

    // Wait for both threads to finish
    pthread_join(thread1, NULL);
    pthread_join(thread2, NULL);

    printf("Both threads have finished execution.\n");
    return 0;
}
```

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">8.4.1 Static buffer</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Making the buffer <b>static</b> keeps it alive after the thread function ends, which solves the dangling-pointer problem:</p><div style="display:inline-block;background:#4C8BF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px 10px 0 0;margin-top:10px;">⚠️ fragment — not compiled</div><div style="background:#1B2A49;border-radius:0 12px 12px 12px;padding:14px 22px;margin:0 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">void *thread_function ( void *)</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">{</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    static char buffer [ 64 ] ;</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    // Fill up the buffer with something good.</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">    return buffer ;</span></div><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">}</span></div></div><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">This may be acceptable sometimes, but not when <b>several threads run the same function</b>: the second thread <b>overwrites</b> the static buffer and destroys the data left by the first. <b>Global data</b> suffers the same problem because it always has a fixed lifetime.</p><div style="background:#FCE8E6;border-left:8px solid #EA4335;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#EA4335;font-size:30px;margin-bottom:6px;">Status of the fragment</div><div style="font-size:28px;line-height:1.6;">Again an <b>incomplete fragment</b> (illustration only, same unnamed-parameter remark as above); the complete, compilable demonstration is <code>thread7.c</code>.</div></div>

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:30px;">thread7.c — shared static buffer (race condition)</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Both threads fill the <b>same</b> static <code>buffer</code> character by character with their own string (<code>Thread1</code> / <code>Thread2</code>). <code>usleep(100)</code> simulates work and raises the chance of a <b>context switch</b>, so the two writes interleave.</p><div style="display:inline-block;background:#FF7A00;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 thread7.c</div><div style="font-size:22px;color:#555;">Archive file: <code>04_thread7_shared_static_buffer_race.c</code> — the next cell writes exactly this source to <code>/tmp/exp8_lab/thread7.c</code>.</div>

```python
%%writefile /tmp/exp8_lab/thread7.c
//thread7.c
//multi thread, one function, static buffer
#include <stdio.h>
#include <pthread.h>
#include <unistd.h>
#include <string.h>

#define BUFFER_SIZE 100

// Static buffer
static char buffer[BUFFER_SIZE];

// Function to be called by threads
void* write_to_buffer(void* arg) {
    char* str = (char*)arg; //arg is a ptr to string
    for (int i = 0; i < BUFFER_SIZE - 1; ++i) {
        // fill the buffer with the chars from the arg string:
        // each char of the buffer is set to a char from the str
        buffer[i] = str[i % strlen(str)];
        // Sleep (micro seconds) to simulate work and increase the chance of
        // context switch to demonstrate concurrency issues in multi threading
        usleep(100);
    }
    // set the last char of the buffer to Null terminator,
    // marking the end of the str in the buffer
    buffer[BUFFER_SIZE - 1] = '\0';
    printf("Buffer content from thread: %s\n", buffer);
    return NULL;
}

int main() {
    pthread_t thread1, thread2;

    // Create two threads
    pthread_create(&thread1, NULL, write_to_buffer, "Thread1");
    pthread_create(&thread2, NULL, write_to_buffer, "Thread2");

    // Wait for both threads to finish
    pthread_join(thread1, NULL);
    pthread_join(thread2, NULL);

    // Print final buffer content
    printf("\nFinal buffer content: %s\n", buffer);
    return 0;
}
```

Because both threads write into the same memory, the buffer holds a <b>random mixture</b> of <code>Thread1</code> and <code>Thread2</code> pieces, and the text differs from run to run.</div></div>

```python
%%bash
cd /tmp/exp8_lab
gcc thread7.c -o thread7 -pthread
for r in 1 2 3; do echo "=== run $r"; ./thread7; done
```

<div style="background:#DDF6F6;border-left:8px solid #00A6A6;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#00A6A6;font-size:30px;margin-bottom:6px;">What to check</div><div style="font-size:28px;line-height:1.6;">After the run, read the printed content. One thread writes part of the buffer, then sleeps (or yields for any reason); the other writes; then the first continues from where it stopped, and so on. The buffer is therefore a random mix of what both threads wrote. The cause is a <b>shared resource</b> (the static buffer) used without protection — a <b>race condition</b>. To prevent it we need <b>synchronisation</b>, which is covered in the next experiment (semaphores).</div></div><div style="background:#FFEDD9;border-left:8px solid #FF7A00;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#FF7A00;font-size:30px;margin-bottom:6px;">Note</div><div style="font-size:28px;line-height:1.6;"><code>usleep()</code> is used on purpose to simulate the real execution time of threads and so increase the chance of switching between them.</div></div>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">8.4.2 Dynamic buffer</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">This time every thread allocates <b>its own buffer</b> with <code>malloc</code>. The contents of the two threads are no longer mixed or overwritten because <b>there is no shared resource</b>. Access to such a buffer is lost when the thread ends, but there is still a risk of a <b>memory leak</b> — especially if the main function ends without waiting for its threads — because when <code>main</code> finishes, all running threads are cut off and what they were doing stays unfinished.</p>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:30px;">thread8.c — a private dynamic buffer per thread</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Each call of <code>write_to_buffer</code> allocates its own 100-byte buffer, fills it, prints it and <code>free()</code>s it; <code>main</code> joins both threads so the cleanup always happens.</p><div style="display:inline-block;background:#9E4CF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 thread8.c</div><div style="font-size:22px;color:#555;">Archive file: <code>05_thread8_dynamic_buffer_per_thread.c</code> — the next cell writes exactly this source to <code>/tmp/exp8_lab/thread8.c</code>.</div>

```python
%%writefile /tmp/exp8_lab/thread8.c
//thread8.c
//multi thread , one func, dynamic memory allocation
#include <stdio.h>
#include <stdlib.h>
#include <pthread.h>
#include <unistd.h>
#include <string.h>

#define BUFFER_SIZE 100

// Function to be called by threads
void* write_to_buffer(void* arg) {
    char* str = (char*)arg;
    char* buffer = (char*)malloc(BUFFER_SIZE * sizeof(char));
    if (buffer == NULL) {
        perror("Failed to allocate memory");
        return NULL;
    }
    for (int i = 0; i < BUFFER_SIZE - 1; ++i) {
        buffer[i] = str[i % strlen(str)];
        usleep(100); // Sleep to simulate work and increase the chance of context switch
    }
    buffer[BUFFER_SIZE - 1] = '\0';
    printf("Buffer content from thread: %s\n", buffer);
    free(buffer);
    return NULL;
}

int main() {
    pthread_t thread1, thread2;

    // Create two threads
    pthread_create(&thread1, NULL, write_to_buffer, "Thread1");
    pthread_create(&thread2, NULL, write_to_buffer, "Thread2");

    // Wait for both threads to finish
    pthread_join(thread1, NULL);
    pthread_join(thread2, NULL);
    return 0;
}
```

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:32px;">8.5 Nested Threads</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">In this example the function of the first thread <b>creates another thread</b>.</p>

<h3 style="color:#EA4335;border-left:8px solid #EA4335;padding-left:12px;margin:14px 0 10px;font-size:30px;">thread9.c — nested threads</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><code>main</code> creates a thread running <code>main_thread_function</code>; that thread creates and joins the nested thread, then prints a message; <code>main</code> joins the first thread and prints the last message.</p><div style="display:inline-block;background:#EA4335;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 thread9.c</div><div style="font-size:22px;color:#555;">Archive file: <code>06_thread9_nested_threads.c</code> — the next cell writes exactly this source to <code>/tmp/exp8_lab/thread9.c</code>.</div>

```python
%%writefile /tmp/exp8_lab/thread9.c
//thread9.c
//nested threads: one thread creates another thread
#include <stdio.h>
#include <pthread.h>

// Function to be executed by the nested thread
void* nested_thread_function(void* arg) {
    printf("Hello from the nested thread!\n");
    return NULL;
}

// Function to be executed by the main thread
void* main_thread_function(void* arg) {
    pthread_t nested_thread;

    // Create the nested thread
    pthread_create(&nested_thread, NULL, nested_thread_function, NULL);

    // Wait for the nested thread to finish
    pthread_join(nested_thread, NULL);
    printf("Nested thread has finished execution.\n");
    return NULL;
}

int main() {
    pthread_t main_thread;

    // Create the main thread
    pthread_create(&main_thread, NULL, main_thread_function, NULL);

    // Wait for the main thread to finish
    pthread_join(main_thread, NULL);
    printf("Main thread has finished execution.\n");
    return 0;
}
```
