<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">7.1 What Is a Process?</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">A <b>process</b> is a <b>running program</b> with its own <b>address space</b> and resources. Every process has a unique <b>process ID</b> (<b>PID</b>). When a process is created by another process, the created one is the <b>child process</b> and the creator is the <b>parent process</b>. A child has its own PID and also keeps the <b>parent process ID</b> (<b>PPID</b>) in its record.</p><div style="display:grid;grid-template-columns:repeat(3,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">👶</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">Orphan process</div><div style="font-size:26px;color:#333;line-height:1.5;">A <b>child whose parent has terminated</b> while the child is still running.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🧟</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">Zombie process</div><div style="font-size:26px;color:#333;line-height:1.5;">A process that <b>has finished</b> but whose record (<b>PCB</b>) is <b>not yet removed</b> from the process table — usually because the parent has not yet called <code>wait</code> for it.</div></div><div style="background:#F3E8FF;border-top:6px solid #9E4CF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">1️⃣</div><div style="font-weight:800;color:#9E4CF5;font-size:28px;margin-bottom:8px;">The first process</div><div style="font-size:26px;color:#333;line-height:1.5;">At boot the kernel creates the first process, <code>systemd</code>, with <b>PID 1</b> (called <code>init</code> on older systems). It is the ancestor of all other processes.</div></div></div>

<p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">To see the first process of your distribution, run:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">ps -p 1</span></div></div>

```python
%%bash
echo "--- the first process of this system (PID 1)"
ps -p 1 || echo "(ps could not show PID 1 here)"
echo "--- the same process with more columns"
ps -p 1 -o pid,ppid,comm 2>/dev/null || true
```

It is the parent of two other managers: one for <b>login shells</b> and one for <b>user sessions</b>.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">⌨️</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">getty@tty1</div><div style="font-size:26px;color:#333;line-height:1.5;">Manages a <b>virtual terminal</b> and lets users log in. It shows the <b>login prompt</b>, then creates a child named <code>login</code> that <b>authenticates</b> the user; <code>login</code> creates one more child — the user's <b>default shell</b>, which provides the command line.</div></div></div><div style="background:#DDF6F6;border-left:8px solid #00A6A6;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#00A6A6;font-size:30px;margin-bottom:6px;">Remember</div><div style="font-size:28px;line-height:1.6;">Process tree: <b>systemd (1)</b> → <b>getty</b> → <b>login</b> → <b>shell</b> → every command you type.</div></div>

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">7.2 Viewing Processes</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">To list all processes of the system, run:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">$ ps -aux</span></div></div><div style="background:#FFEDD9;border-left:8px solid #FF7A00;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#FF7A00;font-size:30px;margin-bottom:6px;">Note — tty and pts</div><div style="font-size:28px;line-height:1.6;">In the output you can also see <b>tty</b>s. For some processes you see <b>pts</b> instead: it stands for <b>Pseudo Terminal Slave</b> and belongs to <i>emulated</i> terminals (network / remote logins, terminal windows). <b>tty</b>s belong to <b>direct consoles</b> and classic physical or virtual terminals of the system.</div></div><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">As seen in earlier experiments, <code>top</code> or <code>htop</code> show processes <b>interactively</b>; <code>pstree</code> shows the <b>tree structure</b>; and to get a PID from a process name use:</p><div style="background:#1B2A49;border-radius:12px;padding:14px 22px;margin:12px 0 12px 0;font-family:monospace;font-size:24px;line-height:1.6;color:#E6EDF7;"><div style="white-space:pre-wrap;"><span style="color:#E6EDF7;">$ pgrep process_name</span></div></div>

```python
%%bash
echo "--- ps aux  (first 5 lines)"
ps aux | head -n 5
echo "--- pgrep: PID of the processes whose name matches 'bash'"
pgrep bash | head -n 3 || true
echo "--- pstree (if installed)"
if command -v pstree >/dev/null 2>&1; then pstree -p -s $$ | head -n 3; else echo "(pstree is not installed on this machine)"; fi
```

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:30px;">proc1.c — current PID and parent PID</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Uses <code>getpid()</code> and <code>getppid()</code> (type <code>pid_t</code>) and prints both numbers.</p><div style="display:inline-block;background:#9E4CF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 proc1.c</div><div style="font-size:22px;color:#555;">Archive file: <code>01_proc1_get_pid_ppid.c</code> — the next cell writes exactly this source to <code>/tmp/exp7_lab/proc1.c</code>.</div>

```python
%%writefile /tmp/exp7_lab/proc1.c
//proc1.c
#include <stdio.h>
#include <unistd.h>
#include <sys/types.h>

int main() {
    pid_t pid, ppid;
    pid = getpid();
    ppid = getppid();
    printf("Current Process PID: %d, PPID: %d\n", pid, ppid);
    return 0;
}
```

The PID is different on every run; the PPID is the process that started <code>p1</code> (your shell).</div></div>

```python
%%bash
cd /tmp/exp7_lab
gcc proc1.c -o p1
./p1
echo "--- second run: the PID changes"
./p1
```

The child prints its own PID and its PPID.</p><div style="display:inline-block;background:#4C8BF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 proc2.c</div><div style="font-size:22px;color:#555;">Archive file: <code>02_proc2_fork_child.c</code> — the next cell writes exactly this source to <code>/tmp/exp7_lab/proc2.c</code>.</div>

```python
%%writefile /tmp/exp7_lab/proc2.c
//proc2.c
#include <stdio.h>
#include <unistd.h>
#include <sys/types.h>
#include <sys/wait.h>

int main() {
    pid_t pid = fork();
    if (pid == 0) {
        // Child Process Code goes here
        printf("Child Process PID: %d, PPID: %d\n", getpid(), getppid());
    }
    else {
        // Parent Process Code goes here
        printf("Parent Process PID: %d, Child Process PID: %d \n", getpid(), pid);
        wait(NULL); // Parent waits till the Child terminates.
    }
    return 0;
}
```

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">Replacing the code of the child — exec</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Here the child uses <code>execlp()</code> to <b>replace</b> the code it inherited from the parent with new code — in this example the command <code>ls</code>. <code>execve()</code> could be used too, but <code>execlp()</code> has a simpler structure.</p>

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:30px;">proc3.c — fork() + execlp("ls")</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">The child prints a message and calls <code>execlp("ls", "ls", "-l", NULL)</code>. If <code>execlp</code> succeeds, the lines after it never run (the child <i>is</i> now <code>ls</code>); <code>return 1;</code> is reached only when the exec fails. The parent waits and prints a final message.</p><div style="display:inline-block;background:#FF7A00;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 proc3.c</div><div style="font-size:22px;color:#555;">Archive file: <code>03_proc3_exec_ls.c</code> — the next cell writes exactly this source to <code>/tmp/exp7_lab/proc3.c</code>.</div>

```python
%%writefile /tmp/exp7_lab/proc3.c
//proc3.c
#include <stdio.h>
#include <unistd.h>
#include <sys/types.h>
#include <sys/wait.h>

int main() {
    pid_t pid = fork();
    if (pid == 0) {
        // Child Process
        printf("Child process will run ls command \n");
        execlp("ls", "ls", "-l", NULL);
        return 1;
    }
    else {
        // Parent Process
        printf("Parent Process: waiting for child termination\n");
        wait(NULL);
        printf("Child Process Has terminated\n");
    }
    return 0;
}
```

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:32px;">7.5.1 IPC through a Pipe</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">A <b>pipe</b> (a “pseudo-file”) is one of the simplest and oldest IPC methods. It creates a <b>one-way channel</b> that connects two processes: the output of one becomes the input of the other. Two kinds exist:</p><div style="display:grid;grid-template-columns:repeat(2,1fr);gap:16px;margin:14px 0;"><div style="background:#E8F0FE;border-top:6px solid #4C8BF5;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🕳️</div><div style="font-weight:800;color:#4C8BF5;font-size:28px;margin-bottom:8px;">Unnamed pipe</div><div style="font-size:26px;color:#333;line-height:1.5;">Works only between <b>related</b> processes (e.g. parent and child) and disappears when they end. The child gets it <b>by inheritance</b>.</div></div><div style="background:#E6F4EA;border-top:6px solid #34A853;border-radius:12px;padding:18px;box-shadow:0 3px 10px rgba(0,0,0,.12);"><div style="font-size:50px;">🏷️</div><div style="font-weight:800;color:#34A853;font-size:28px;margin-bottom:8px;">Named pipe (FIFO)</div><div style="font-size:26px;color:#333;line-height:1.5;">Appears as a <b>file in the file system</b>, lets <b>unrelated</b> processes communicate and stays until it is removed.</div></div></div><div style="background:#FFEDD9;border-left:8px solid #FF7A00;border-radius:10px;padding:16px 22px;margin:14px 0;color:#222;"><div style="font-weight:800;color:#FF7A00;font-size:30px;margin-bottom:6px;">Limits</div><div style="font-size:28px;line-height:1.6;">Communication is <b>one-way</b>, and an unnamed pipe is restricted to related processes. Still, pipes are very common on the command line to connect the output of one command to the input of another.</div></div><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">In the example the <b>parent sends a message</b> to the child. The child calls <code>read()</code>, a <b>blocking call</b>: it waits until data exists, then prints it.</p>

<h3 style="color:#34A853;border-left:8px solid #34A853;padding-left:12px;margin:14px 0 10px;font-size:30px;">Pipe between parent and child</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;"><code>pipe(fd)</code> creates the channel (<code>fd[0]</code> = read end, <code>fd[1]</code> = write end). After <code>fork()</code>, each side <b>closes the end it does not use</b>: the child closes <code>fd[1]</code> and reads; the parent closes <code>fd[0]</code>, writes the message (including the terminating <code>\0</code>), closes the write end and waits. <i>The PDF listing has no file name for this program.</i></p><div style="display:inline-block;background:#34A853;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 pipe_parent_to_child.c</div><div style="font-size:22px;color:#555;">Archive file: <code>04_pipe_parent_to_child.c</code> — the next cell writes exactly this source to <code>/tmp/exp7_lab/pipe_parent_to_child.c</code>.</div>

```python
%%writefile /tmp/exp7_lab/pipe_parent_to_child.c
// pipe_parent_to_child.c  (the PDF listing has no file name; this name was chosen for the archive)
#include <stdio.h>
#include <unistd.h>
#include <sys/types.h>
#include <sys/wait.h>
#include <string.h>

int main() {
    int fd[2];
    char message[] = "Hello from Parent!\n";
    char buffer[100];

    pipe(fd); //Create pipe

    pid_t pid = fork();

    if (pid == 0) {
        // Child Process
        close(fd[1]);
        read(fd[0], buffer, sizeof(buffer)); //Read from pipe
        printf("Child Process received this message:%s\n", buffer);
        close(fd[0]); } // Close the Read End
    else {
        // Parent Process
        close(fd[0]);
        write(fd[1], message, strlen(message)+1); //write in pipe
        close(fd[1]); // Close the Write End
        wait(NULL); }
    return 0;
}
```

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:32px;">7.5.2 IPC through Shared Memory</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">Two or more processes <b>share a part of memory</b> and read / write data <b>directly</b> in it. To avoid a <b>race condition</b> some synchronisation is needed. Because no data is copied, this is the <b>fastest form of IPC</b>.</p><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">The example has a <b>producer</b> and a <b>consumer</b>. Returns a <code>void *</code> that is cast to <code>char *</code> so the memory behaves like a character array.</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;"><code>shmdt(str)</code></td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;"><b>Detaches</b> the segment from the process.</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#F3E8FF;font-size:23px;color:#222;line-height:1.5;"><code>shmctl(shmid, IPC_RMID, NULL)</code></td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#F3E8FF;font-size:23px;color:#222;line-height:1.5;">Marks the segment for <b>destruction</b> (done by the producer at the end).</td></tr></tbody></table>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:30px;">producer.c — Process 1</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Creates <code>shmfile</code>, builds the key, creates and attaches the segment, reads one line with <code>fgets</code>, prints it and polls once per second until the first byte becomes <code>*</code>.</p><div style="display:inline-block;background:#9E4CF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 producer.c</div><div style="font-size:22px;color:#555;">Archive file: <code>05_shm_producer.c</code> — the next cell writes exactly this source to <code>/tmp/exp7_lab/producer.c</code>.</div>

```python
%%writefile /tmp/exp7_lab/producer.c
// IPC with Shared Memory; Process 1: The Producer
// producer.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <unistd.h>
#include <sys/shm.h>
#include <sys/ipc.h>

#define SHM_SIZE 1024 // Size of shared memory segment

int main () {
    // Create a file to be used in key production
    FILE *f = fopen("shmfile", "w");
    if (f) fclose(f);

    // Generate unique key for shared memory
    key_t key = ftok("shmfile", 65);

    // Create shared memory segment
    int shmid = shmget(key, SHM_SIZE, 0666 | IPC_CREAT);

    // Attach shared memory to process address space
    char *str = (char*) shmat(shmid, NULL, 0);

    // Get input from user and write to shared memory
    printf("Producer Process; Enter data: ");
    fgets(str, SHM_SIZE, stdin);

    printf("Data is written to shared memory: %s\n", str);

    // Wait for the consumer to read data
    // Consumer will signal by writing '*' to first byte
    printf("Waiting for consumer to read data...\n");
    while (*str != '*') {
        sleep(1); // Sleep for 1 second
    }

    // Detach from shared memory
    shmdt(str);

    // Mark shared memory segment for destruction
    shmctl(shmid, IPC_RMID, NULL);

    return 0;
}
```

It ends only after the consumer has run.</div></div>

<h3 style="color:#9E4CF5;border-left:8px solid #9E4CF5;padding-left:12px;margin:14px 0 10px;font-size:30px;">consumer.c — Process 2</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Builds the <b>same key</b>, opens the existing segment (no <code>IPC_CREAT</code>), polls until the first byte is non-zero, prints the data and writes <code>*</code> to signal the producer.</p><div style="display:inline-block;background:#9E4CF5;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 consumer.c</div><div style="font-size:22px;color:#555;">Archive file: <code>06_shm_consumer.c</code> — the next cell writes exactly this source to <code>/tmp/exp7_lab/consumer.c</code>.</div>

```python
%%writefile /tmp/exp7_lab/consumer.c
// IPC with Shared Memory; Process 2: The Consumer
// consumer.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <unistd.h>
#include <sys/shm.h>
#include <sys/ipc.h>

#define SHM_SIZE 1024 // Size of shared memory segment

int main() {
    // Generate the same key as producer
    key_t key = ftok("shmfile", 65);

    // Get existing shared memory segment
    int shmid = shmget(key, SHM_SIZE, 0666);

    // Attach shared memory to process address space
    char *str = (char*) shmat(shmid, NULL, 0);

    // Wait for producer to write data
    // Check if first byte contains data (not null)
    printf("Consumer Process; Waiting for data...\n");
    while (*str == 0) {
        sleep(1); // Sleep for 1 second
    }

    // Read and display data from shared memory
    printf("Data read from shared memory: %s\n", str);

    // Signal producer by '*' that writing is finished
    *str = '*';

    // Detach from shared memory
    shmdt(str);

    return 0;
}
```

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:32px;">7.5.3 IPC through a Socket</h3><p style="font-size:30px;line-height:1.65;color:#222;margin:10px 0;">A <b>socket</b> is a <b>network-based</b> IPC mechanism following the <b>client–server</b> model. It works between processes on one machine or on different machines and supports protocols such as <b>TCP</b> and <b>UDP</b>.</p><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">The <b>server</b> creates a local socket of type <b>UNIX domain</b> and binds it to a local address — a constant defined in the program (<code>SOCKET_PATH</code>) that works for processes on the same machine. For network sockets you would use the <code>AF_INET</code> family instead. The server listens, accepts a client, receives its message, sends a reply and closes the connection.</p><table style="border-collapse:separate;border-spacing:0;width:100%;margin:14px 0;border-radius:12px;overflow:hidden;box-shadow:0 3px 12px rgba(0,0,0,.14);"><thead><tr><th style="background:#FF7A00;color:white;padding:12px 16px;text-align:left;font-size:23px;width:38%;">Call</th><th style="background:#FF7A00;color:white;padding:12px 16px;text-align:left;font-size:23px;width:62%;">Role</th></tr></thead><tbody><tr><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFEDD9;font-size:23px;color:#222;line-height:1.5;"><code>socket(AF_UNIX, SOCK_STREAM, 0)</code></td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFEDD9;font-size:23px;color:#222;line-height:1.5;">Creates the socket (local, reliable stream; other arguments left at default 0).</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;"><code>bind()</code></td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;">Attaches the socket to the address stored in the <code>sockaddr_un</code> structure (a file in <code>/tmp</code>).</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFEDD9;font-size:23px;color:#222;line-height:1.5;"><code>listen(fd, 5)</code></td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFEDD9;font-size:23px;color:#222;line-height:1.5;">Waits for connection requests; at most <b>5</b> may wait in the queue.</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;"><code>accept()</code></td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;">Accepts a request and creates a <b>new file descriptor</b> for that client.</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFEDD9;font-size:23px;color:#222;line-height:1.5;"><code>read()</code> / <code>write()</code></td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFEDD9;font-size:23px;color:#222;line-height:1.5;">Receive the client's message and write the reply.</td></tr><tr><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;"><code>unlink()</code></td><td style="padding:10px 16px;border-bottom:1px solid #DCE3EF;background:#FFFFFF;font-size:23px;color:#222;line-height:1.5;">Removes the socket file.</td></tr></tbody></table>

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:30px;">server.c — Process 1</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Creates, binds and listens on <code>/tmp/my_socket</code>, accepts one client, prints the received text, replies <code>Hello from server!</code>, closes everything and removes the socket file.</p><div style="display:inline-block;background:#FF7A00;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 server.c</div><div style="font-size:22px;color:#555;">Archive file: <code>07_socket_server.c</code> — the next cell writes exactly this source to <code>/tmp/exp7_lab/server.c</code>.</div>

```python
%%writefile /tmp/exp7_lab/server.c
// IPC with Socket; Process 1: Server
// server.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <unistd.h>
#include <sys/socket.h>
#include <sys/un.h>

#define SOCKET_PATH "/tmp/my_socket" // Path for UNIX domain socket

int main () {
    // file descriptors for clients and server sockets
    int server_fd, client_fd;
    // Address Structures
    struct sockaddr_un server_addr, client_addr;
    //Define client Address size
    socklen_t client_len = sizeof(client_addr);
    char buffer[256]; // buffer to put data in

    // Create socket file descriptor
    // AF_UNIX: Unix domain socket, SOCK_STREAM: TCP-like reliable socket
    server_fd = socket(AF_UNIX, SOCK_STREAM, 0);
    if (server_fd == -1) {
        perror("Socket creation failed");
        exit(1); }

    // Configure server address structure
    memset(&server_addr, 0, sizeof(server_addr));
    server_addr.sun_family = AF_UNIX;
    // server socket address = your socket address
    strcpy(server_addr.sun_path, SOCKET_PATH);

    // Remove socket file if it exists from previous run
    unlink(SOCKET_PATH);

    // Bind socket to the specified address
    if (bind(server_fd, (struct sockaddr*)&server_addr, sizeof(server_addr)) == -1) {
        perror("Socket bind failed");
        exit(1); }

    // Listen for incoming connections
    // Backlog queue size: 5
    if (listen(server_fd, 5) == -1) {
        perror("Socket listen failed");
        exit (1); }
    printf("Server: Listening for connections...\n");

    // Accept incoming connection
    client_fd = accept(server_fd, (struct sockaddr*)&client_addr, &client_len);
    if (client_fd == -1) {
        perror("Socket accept failed");
        exit(1); }

    printf("Server: Client connected\n");

    // Read data from client
    int n = read(client_fd, buffer, sizeof(buffer)-1);
    if (n > 0) {
        buffer[n] = '\0'; // Null-terminate the string
        printf("Server: Received: %s\n", buffer);

        // Send response to client
        char *response = "Hello from server!";
        write(client_fd, response, strlen(response)); }

    // Close connections and clean up
    close(client_fd);
    close(server_fd);
    unlink(SOCKET_PATH); // Remove socket file

    return 0; }
```

<h3 style="color:#FF7A00;border-left:8px solid #FF7A00;padding-left:12px;margin:14px 0 10px;font-size:30px;">client.c — Process 2</h3><p style="font-size:28px;line-height:1.65;color:#222;margin:10px 0;">Creates a local socket, <code>connect()</code>s to the same address, sends <code>Hello from client!</code>, prints the server's response and closes the socket.</p><div style="display:inline-block;background:#FF7A00;color:white;font-weight:700;font-size:22px;padding:5px 16px;border-radius:10px;margin:6px 0;">📄 client.c</div><div style="font-size:22px;color:#555;">Archive file: <code>08_socket_client.c</code> — the next cell writes exactly this source to <code>/tmp/exp7_lab/client.c</code>.</div>

```python
%%writefile /tmp/exp7_lab/client.c
// IPC with Socket; Process 2: Client
// client.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <unistd.h>
#include <sys/socket.h>
#include <sys/un.h>

#define SOCKET_PATH "/tmp/my_socket" // Must match server path

int main() {
    int sockfd;
    struct sockaddr_un server_addr;
    char buffer[256];

    // Create socket file descriptor
    sockfd = socket(AF_UNIX, SOCK_STREAM, 0);
    if (sockfd == -1) {
        perror("Socket creation failed");
        exit(1); }

    // Configure server address structure
    memset(&server_addr, 0, sizeof(server_addr));
    server_addr.sun_family = AF_UNIX;
    strcpy(server_addr.sun_path, SOCKET_PATH);

    // Connect to server
    if (connect(sockfd, (struct sockaddr*)&server_addr, sizeof(server_addr)) == -1) {
        perror("Socket connect failed");
        exit(1); }

    printf("Client: Connected to server\n");

    // Send message to server
    char *message = "Hello from client!";
    write(sockfd, message, strlen(message));

    // Read response from server
    int n = read(sockfd, buffer, sizeof(buffer)-1);
    if (n > 0) {
        buffer[n] = '\0'; // Null-terminate the string
        printf("Client: Server response: %s\n", buffer); }

    // Close socket connection
    close(sockfd);

    return 0; }
```
